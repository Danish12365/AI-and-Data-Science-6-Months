# 🔷 1. Data Visualization in Tableau Public

👉 Data Visualization = turning data into **charts + dashboards**

In Tableau:

> Drag → Drop → Visualize

---

## 📊 Common Charts You’ll Use

* Bar Chart
* Line Chart
* Scatter Plot
* Maps
* Pie Chart

---

## 🧠 Basic Workflow

1. Load dataset
2. Drag dimensions (categories) → Columns
3. Drag measures (numbers) → Rows
4. Choose chart type

---

# 🔷 2. Scatter Plots in Tableau

👉 Scatter Plot = relationship between **two numeric variables**

---

## 🧩 Example:

* X-axis → Sales
* Y-axis → Profit

---

## 🪜 Steps:

1. Drag **Sales → Columns**
2. Drag **Profit → Rows**
3. Change Marks → Circle

---

## 🔥 Add More Power:

* Drag **Category → Color**
* Drag **Region → Detail**

👉 Now you see patterns

---

## 📊 Use Case:

* Correlation analysis
* Outlier detection

---

# 🔷 3. Dynamic Scatter Plot (VERY IMPORTANT)

👉 Dynamic = user can change values (interactive)

---

## 🧩 Method 1: Filters

1. Drag **Category → Filters**
2. Right click → Show Filter

👉 User can switch categories

---

## 🧩 Method 2: Parameters (Advanced)

1. Create Parameter (e.g., select measure)
2. Use calculated field

---

## 💡 Example:

User chooses:

* Sales
* Profit
* Quantity

👉 Chart updates automatically

---

# 🔷 4. Flip Your Charts in Tableau

👉 Flip = swap X and Y axis

---

## 🪜 Method:

Click **Swap Button (⇄)** in toolbar

---

## 📌 Example:

Before:

* Sales → Rows
* Category → Columns

After flip:

* Category → Rows
* Sales → Columns

---

## 💡 Why Use?

* Better readability
* Horizontal bar charts

---

# 🔷 5. Bar Plots and Sorting

---

## 📊 Create Bar Chart

1. Drag **Category → Columns**
2. Drag **Sales → Rows**

---

## 🔄 Sorting

### 🟢 Method 1:

Click **Sort Icon**

---

### 🟢 Method 2:

Right click → Sort

---

## 📌 Example:

Sort by:

* Sales (descending)

👉 Shows top-performing categories

---

## 🔥 Add Enhancements:

* Color by region
* Label values
* Add filters

---

# 🔷 6. GEO Maps in Tableau

👉 Tableau has built-in **geographic intelligence**

---

## 🌍 Example Data:

* Country
* City
* State

---

## 🪜 Steps:

1. Drag **Country → Canvas**
2. Tableau auto-detects map

---

## 📊 Add Data:

* Drag **Sales → Color**
* Drag **Profit → Size**

---

## 🔥 Result:

Interactive world map

---

## 📌 Types of Maps:

* Symbol Map (points)
* Filled Map (regions)

---

## 💡 Example:

* Sales by country
* Profit by region

---
