Choosing the right plot is about matching your **data type** to your **analytical goal**. In data science, we don't pick plots because they look good; we pick them because they reveal a specific relationship.

Here is a professional guide to selecting the best visualization for your projects.

---

## 1. The "Big Four" Goals
Most visualizations fall into one of these four categories:

| Goal | Description | Best Plots |
| :--- | :--- | :--- |
| **Comparison** | Comparing values across categories. | Bar Chart, Grouped Bar, Radar Chart |
| **Distribution** | Seeing how data is spread out (spread, skew, outliers). | Histogram, Box Plot, Violin Plot |
| **Relationship** | Finding correlations between two or more variables. | Scatter Plot, Heatmap, Bubble Chart |
| **Composition** | Seeing how a "whole" is broken into parts. | Stacked Bar, Treemap (Avoid Pie Charts!) |

---

## 2. Guide by Data Type

### A. Numerical vs. Numerical (Regression)
When you want to see if one number affects another (e.g., `total_bill` vs. `tip`).
* **Scatter Plot:** The gold standard for relationships.
* **Line Plot:** Use this if the X-axis is **Time** (Time Series).
* **Joint Plot:** Shows the scatter plot plus the distribution of both variables on the sides.


### B. Categorical vs. Numerical (Classification/Analysis)
When you want to compare groups (e.g., `Day of Week` vs. `Average Tip`).
* **Bar Chart:** Best for comparing total or average values.
* **Box Plot (Whisker Plot):** Crucial for seeing **Outliers** and the **Median**.
* **Violin Plot:** Shows the "Density" (PDF) of the data. Use this if you want to see if the data has two peaks (bimodal).


### C. Categorical vs. Categorical
When you want to see how groups overlap (e.g., `Smoker` vs. `Day`).
* **Heatmap:** Excellent for seeing intensity (e.g., a Correlation Matrix).
* **Stacked Bar Chart:** Shows the composition of one category within another.

---

## 3. Professional "Rules of Thumb"

### ✅ Do This:
* **Use Box Plots for Outliers:** Before training your Bank Churn model, use a box plot on "Balance" to see if a few wealthy customers are skewing your data.
* **Use Histograms for Normality:** Use a Histogram (with a KDE curve) to check if your features are normally distributed before applying a `QuantileTransformer`.
* **Keep it Simple:** If a Bar Chart explains it, don't use a 3D Radar Chart.


### ❌ Avoid This:
* **Pie Charts:** Humans are bad at comparing angles. Use a **Horizontal Bar Chart** or a **Treemap** instead.
* **Too Many Colors:** Only use color to represent a variable, not just to make it look "pretty."
* **3D Plots:** 3D charts usually distort the data and make it harder to read.

---

## 4. Quick Selection Matrix

| If you want to see... | Use this Plot |
| :--- | :--- |
| **Outliers** | Box Plot |
| **Correlation between all features** | Heatmap |
| **How a variable changes over time** | Line Chart |
| **The shape of a distribution** | Histogram or KDE |
| **Proportion of a whole** | Treemap or Stacked Bar |
| **Relationship between 2 numbers** | Scatter Plot |

## 1. Distribution Plots (The "Health Check")
Use these first to see if your data is skewed or has outliers.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# A. Histogram with KDE (Is it Normal/Gaussian?)
plt.figure(figsize=(8, 5))
sns.histplot(df['total_bill'], kde=True, color='blue')
plt.title('Distribution of Total Bill')
plt.show()

# B. Box Plot (Where are the Outliers?)
sns.boxplot(x='day', y='tip', data=df, palette='Set2')
plt.title('Tips by Day: Identifying Outliers')
plt.show()
```


---

## 2. Relationship Plots (The "Correlation Check")
Use these to see if one variable drives another.

```python
# C. Scatter Plot with Regression Line
sns.regplot(x='total_bill', y='tip', data=df, scatter_kws={'alpha':0.5})
plt.title('Total Bill vs Tip (with Trendline)')
plt.show()

# D. Heatmap (Correlation Matrix)
# Essential for feature selection!
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlation Matrix')
plt.show()
```


---

## 3. Comparison & Composition (The "Category Check")
Use these to compare groups.

```python
# E. Violin Plot (Distribution density across categories)
sns.violinplot(x='day', y='total_bill', hue='sex', split=True, data=df)
plt.title('Bill Distribution: Day vs Gender')
plt.show()

# F. Bar Chart (Aggregated values)
sns.barplot(x='time', y='tip', data=df, estimator='mean', errorbar=None)
plt.title('Average Tip: Lunch vs Dinner')
plt.show()
```


---

## 4. Multi-Plot Grid (The "Pro Move")
If you want to see everything at once, use a `PairPlot`.

```python
# G. Pairplot (Relationships between all numeric variables)
sns.pairplot(df, hue='smoker', diag_kind='kde')
plt.show()
```

### Professional Tips for your Visuals:
1.  **Tight Layout:** Always use `plt.tight_layout()` before `plt.show()` to prevent labels from overlapping.
2.  **Color Palettes:** Use `palette='viridis'` or `cmap='magma'` for a modern, accessible look.
3.  **Remove Spines:** To make your plots look cleaner (like in high-end journals), use `sns.despine()`.
4.  **Save for Reports:** Use `plt.savefig('my_plot.png', dpi=300)` to save high-quality images for your GitHub README.
