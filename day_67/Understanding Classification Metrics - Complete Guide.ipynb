## 🎯 Core Metrics Explained

### **1. Support**
**What it is:** The number of actual samples in each class.

In your example:
- Class 0: 82 samples
- Class 1: 53 samples  
- Class 2: 49 samples
- **Total: 184 samples**

**Why it matters:** Shows if you have **imbalanced classes**. Your dataset is slightly imbalanced (82 vs 49).

---

### **2. Precision**
**Formula:** Precision = True Positives / (True Positives + False Positives)

**What it means:** "When the model says something is Class X, how often is it correct?"

**In your example:**
- **Class 0 precision = 0.83** → When model predicts "0", it's correct 83% of the time
- **Class 1 precision = 0.60** → When model predicts "1", it's correct only 60% of the time
- **Class 2 precision = 0.62** → When model predicts "2", it's correct 62% of the time

**Real-world example:**
- **Spam filter:** High precision = Few legitimate emails marked as spam
- **Medical test:** High precision = Few healthy people diagnosed as sick

**When to prioritize:** When **False Positives are costly**
- Banking fraud detection (don't want to block legitimate transactions)
- Product recommendations (don't want to annoy users with bad suggestions)

---

### **3. Recall (Sensitivity)**
**Formula:** Recall = True Positives / (True Positives + False Negatives)

**What it means:** "Of all the actual Class X samples, how many did the model find?"

**In your example:**
- **Class 0 recall = 0.85** → Model catches 85% of actual "0"s
- **Class 1 recall = 0.45** → Model catches only 45% of actual "1"s ⚠️ (BAD!)
- **Class 2 recall = 0.76** → Model catches 76% of actual "2"s

**Real-world example:**
- **Spam filter:** High recall = Catches most spam emails
- **Medical test:** High recall = Detects most sick patients

**When to prioritize:** When **False Negatives are costly**
- Cancer screening (can't miss actual cancer cases)
- Fraud detection (must catch fraudulent transactions)
- Fire alarms (better safe than sorry)

---

### **4. F1-Score**
**Formula:** F1 = 2 × (Precision × Recall) / (Precision + Recall)

**What it means:** The **harmonic mean** of precision and recall. Balances both metrics.

**In your example:**
- **Class 0 F1 = 0.84** → Good balance
- **Class 1 F1 = 0.52** → Poor (low recall hurts this)
- **Class 2 F1 = 0.68** → Decent

**Why harmonic mean?** It punishes extreme imbalances. If either precision or recall is low, F1 will be low.

**Example:**
- Precision = 0.9, Recall = 0.9 → F1 = 0.90 ✅
- Precision = 0.9, Recall = 0.1 → F1 = 0.18 ⚠️ (punished heavily)

---

## 🔢 Overall Metrics Explained

### **5. Accuracy**
**Formula:** Accuracy = Correct Predictions / Total Predictions

**In your example:** 0.71 = 71% of all predictions were correct

**When it's good:** Balanced datasets
**When it's misleading:** Imbalanced datasets

**Example of misleading accuracy:**
- Dataset: 95% Class 0, 5% Class 1
- Model always predicts Class 0 → Accuracy = 95%! (But useless for Class 1)

---

### **6. Macro Average**
**How it's calculated:** Simple average across all classes (treats all classes equally)

**In your example:**
- **Macro avg precision = (0.83 + 0.60 + 0.62) / 3 = 0.68**
- **Macro avg recall = (0.85 + 0.45 + 0.76) / 3 = 0.69**
- **Macro avg F1 = (0.84 + 0.52 + 0.68) / 3 = 0.68**

**When to use:** When **all classes are equally important**
- Example: Classifying rare diseases where each disease matters equally

**Insight from your data:** Macro avg (0.68) is lower than accuracy (0.71) because **Class 1 performs poorly** (F1=0.52), dragging down the average.

---

### **7. Weighted Average**
**How it's calculated:** Average weighted by the number of samples in each class (support)

**In your example:**
```
Weighted precision = (0.83×82 + 0.60×53 + 0.62×49) / 184 = 0.71
Weighted recall = (0.85×82 + 0.45×53 + 0.76×49) / 184 = 0.71
Weighted F1 = (0.84×82 + 0.52×53 + 0.68×49) / 184 = 0.71
```

**When to use:** When **larger classes are more important**
- Example: Customer classification where you have 80% regular customers, 20% VIP

**Insight from your data:** Weighted avg (0.71) equals accuracy because **Class 0 (largest class) performs well**, balancing out poor Class 1 performance.

## 🎯 Which Metric Should You Use?

### **Use Accuracy when:**
- Classes are balanced
- All errors are equally costly
- Simple overall measure needed

### **Use Precision when:**
- False Positives are expensive
- Examples: Spam filter, fraud alerts, medical screening

### **Use Recall when:**
- False Negatives are expensive
- Examples: Cancer detection, fire alarms, security systems

### **Use F1-Score when:**
- Need balance between precision and recall
- Classes are imbalanced
- **Most common choice for imbalanced data**

### **Use Macro Avg when:**
- All classes equally important
- Want to detect problems in minority classes
- Evaluating fairness across classes

### **Use Weighted Avg when:**
- Larger classes are more important
- Want metric aligned with business impact
- Reporting to stakeholders

## 🎓 Quick Reference

| Metric | Formula | Question It Answers |
|--------|---------|---------------------|
| **Precision** | TP/(TP+FP) | How many predictions were correct? |
| **Recall** | TP/(TP+FN) | How many actual cases did we find? |
| **F1** | 2×P×R/(P+R) | What's the balance of both? |
| **Accuracy** | Correct/Total | What % of all predictions correct? |
| **Macro Avg** | Mean(metric) | How well for all classes equally? |
| **Weighted Avg** | Weighted mean | How well considering class sizes? |
