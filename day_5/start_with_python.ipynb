## **Conda Environments (A-Z)**

### **What is Conda?**
A **package manager AND environment manager** for Python. Think of it as a "container" system for your Python projects.

### **Why Use Environments?**
Imagine these scenarios:
1. Project A needs TensorFlow 2.0
2. Project B needs TensorFlow 1.0
Without environments, they conflict. **Environments keep them separate.**

### **Complete Conda Commands Guide:**

```bash
# 1. CHECK CONDDA VERSION
conda --version

# 2. CREATE NEW ENVIRONMENT
conda create --name ai_env python=3.9
# 'ai_env' = your environment name
# 'python=3.9' = Python version

# 3. ACTIVATE ENVIRONMENT
# On Windows:
conda activate ai_env
# On Mac/Linux:
source activate ai_env

# 4. INSTALL PACKAGES
conda install numpy pandas matplotlib jupyter
# OR for specific version:
conda install tensorflow=2.8

# 5. LIST ALL PACKAGES IN CURRENT ENV
conda list

# 6. LIST ALL ENVIRONMENTS
conda env list
# OR
conda info --envs

# 7. DEACTIVATE ENVIRONMENT
conda deactivate

# 8. REMOVE ENVIRONMENT
conda remove --name ai_env --all

# 9. CLONE ENVIRONMENT (Make a copy)
conda create --name ai_env_copy --clone ai_env

# 10. EXPORT ENVIRONMENT (Share with others)
conda env export > environment.yml
# This creates a file with ALL dependencies

# 11. CREATE FROM YML FILE
conda env create -f environment.yml

# 12. UPDATE PACKAGE
conda update numpy

# 13. SEARCH FOR PACKAGE
conda search tensorflow

# 14. REMOVE PACKAGE
conda remove numpy
```

## **Operators in Python**

### **1. Arithmetic Operators** (For calculations)
```python
# Basic math
a = 10
b = 3

print(a + b)   # 13  - Addition
print(a - b)   # 7   - Subtraction  
print(a * b)   # 30  - Multiplication
print(a / b)   # 3.333... - Division (float)
print(a // b)  # 3   - Floor Division (integer)
print(a % b)   # 1   - Modulus (remainder)
print(a ** b)  # 1000 - Exponentiation (power)

# AI Example: Calculate accuracy
correct_predictions = 950
total_samples = 1000
accuracy = correct_predictions / total_samples  # 0.95
```

### **2. Comparison Operators** (Return True/False)
```python
# For conditions and model evaluation
x = 10
y = 5

print(x == y)   # False - Equal to
print(x != y)   # True  - Not equal to
print(x > y)    # True  - Greater than
print(x < y)    # False - Less than
print(x >= y)   # True  - Greater than or equal
print(x <= y)   # False - Less than or equal

# AI Example: Check if model is good enough
model_accuracy = 0.85
threshold = 0.80
is_acceptable = model_accuracy >= threshold  # True
```

### **3. Logical Operators** (Combine conditions)
```python
# and, or, not
temp = 25
humidity = 60

# AI Example: Check if conditions are good for training
good_temp = temp > 20 and temp < 30      # True
high_humidity = humidity > 70            # False
train_model = good_temp and not high_humidity  # True

# Another example
accuracy = 0.92
training_time = 120  # minutes

# Deploy model if accuracy > 90% AND training time < 2 hours
deploy = (accuracy > 0.90) and (training_time < 120)  # True
```

### **4. Assignment Operators**
```python
# Shorthand operations
x = 10
x += 5   # x = x + 5 → 15
x -= 3   # x = x - 3 → 12
x *= 2   # x = x * 2 → 24
x /= 4   # x = x / 4 → 6.0

# AI Example: Update learning rate during training
learning_rate = 0.1
learning_rate *= 0.9   # Reduce by 10% each epoch
```

### **5. Membership Operators** (Check if item exists)
```python
# Useful for data validation
features = ['age', 'salary', 'education', 'experience']

# Check if feature exists
has_age = 'age' in features        # True
has_gender = 'gender' not in features  # True
```
