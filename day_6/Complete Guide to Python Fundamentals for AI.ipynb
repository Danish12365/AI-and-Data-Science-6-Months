## **1. Indentations and if Conditions in Python**

### **Indentation is CRITICAL in Python!**
Unlike other languages that use braces `{}`, Python uses **indentation (spaces/tabs)** to define code blocks.

```python
# CORRECT - Proper indentation
if True:
    print("This is inside the if block")  # 4 spaces/tab
    print("Still inside")                 # same indentation
print("Outside the if block")             # back to left edge

# WRONG - Will cause error!
if True:
print("This will give IndentationError")  # No indentation!
```

### **If Conditions - Decision Making in Code**
```python
# Basic if statement
accuracy = 0.92

if accuracy > 0.90:
    print("✅ Model performance is excellent!")
    print("Ready for deployment.")

# if-else statement
model_accuracy = 0.85

if model_accuracy >= 0.90:
    print("✅ Deploy to production")
else:
    print("⚠️ Needs more training")
    print("Increase epochs or adjust hyperparameters")

# if-elif-else (multiple conditions)
accuracy = 0.88

if accuracy >= 0.90:
    grade = "A"
    action = "Deploy immediately"
elif accuracy >= 0.80:  # "else if" in other languages
    grade = "B"
    action = "Minor improvements needed"
elif accuracy >= 0.70:
    grade = "C"
    action = "Needs significant retraining"
else:
    grade = "F"
    action = "Start from scratch"

print(f"Grade: {grade}")
print(f"Action: {action}")

# Multiple conditions with logical operators
temperature = 75
humidity = 60
dataset_size = 10000

# AND - All conditions must be True
if temperature < 80 and humidity < 70:
    print("Good conditions for server room")

# OR - At least one condition must be True
if accuracy > 0.95 or dataset_size > 50000:
    print("✅ Model is production-ready")

# NOT - Inverts the condition
is_trained = False
if not is_trained:
    print("Model needs training")

# Complex AI example
model_accuracy = 0.87
training_time = 45  # minutes
resource_usage = "high"

if model_accuracy > 0.85 and training_time < 60:
    if resource_usage == "low":
        print("✅ Perfect model! Deploy everywhere.")
    elif resource_usage == "medium":
        print("✅ Good model. Deploy with monitoring.")
    else:
        print("⚠️ Good accuracy but resource-heavy. Optimize first.")
else:
    print("❌ Model needs improvement before deployment.")
```

## **2. User Input Program in Python**

### **Getting Input from Users**
```python
# Basic input
name = input("Enter your name: ")
print(f"Hello, {name}! Welcome to AI World.")

# Input for numbers (input always returns string)
age = input("Enter your age: ")
age = int(age)  # Convert to integer
print(f"In 10 years, you'll be {age + 10} years old.")

# Multiple inputs in one line
x, y = input("Enter two numbers (separated by space): ").split()
x = float(x)
y = float(y)
print(f"Sum: {x + y}")

# AI Model Configuration Input
print("=== AI Model Configuration ===")
learning_rate = float(input("Enter learning rate (e.g., 0.001): "))
epochs = int(input("Enter number of epochs: "))
batch_size = int(input("Enter batch size (32, 64, 128): "))

print(f"\nConfiguration saved:")
print(f"Learning Rate: {learning_rate}")
print(f"Epochs: {epochs}")
print(f"Batch Size: {batch_size}")

# Complete AI Assistant Program
def ai_assistant():
    print("🤖 Welcome to AI Assistant!")
    
    # Get user requirements
    task = input("What AI task do you need? (classification/regression/clustering): ").lower()
    
    if task == "classification":
        classes = input("How many classes? ")
        print(f"I'll prepare a {classes}-class classifier for you!")
        
    elif task == "regression":
        print("I'll set up a regression model for continuous predictions.")
        
    elif task == "clustering":
        clusters = input("How many clusters? ")
        print(f"I'll use K-Means with {clusters} clusters.")
        
    else:
        print("Sorry, I don't support that task yet.")
    
    # Get dataset info
    samples = input("Approximate dataset size (number of samples): ")
    print(f"\nPerfect! I'll prepare an AI model for {task} with {samples} samples.")

# Run the assistant
# ai_assistant()

# Continuous input with while loop
def data_collection_bot():
    print("📊 Data Collection Bot Started")
    data_points = []
    
    while True:
        value = input("Enter a data point (or 'done' to finish): ")
        
        if value.lower() == 'done':
            break
        
        try:
            num_value = float(value)
            data_points.append(num_value)
            print(f"Added: {num_value}")
        except:
            print("Please enter a valid number!")
    
    print(f"\nData Collection Complete!")
    print(f"Total points: {len(data_points)}")
    print(f"Average: {sum(data_points)/len(data_points):.2f}")
    print(f"Data: {data_points}")

# Run the bot
# data_collection_bot()
```

## **4. Control Flow Statements in Python**

### **For Loops**
```python
# Basic for loop
ai_frameworks = ["TensorFlow", "PyTorch", "Keras", "Scikit-learn"]

for framework in ai_frameworks:
    print(f"Learning {framework}...")

# With range()
print("Training epochs:")
for epoch in range(1, 6):  # 1 to 5
    print(f"  Epoch {epoch}: Training...")

# AI Example: Training simulation
print("\n🎯 Training Progress:")
accuracies = []

for epoch in range(1, 11):  # 10 epochs
    # Simulate accuracy improvement
    accuracy = 0.70 + (epoch * 0.025)
    accuracies.append(accuracy)
    print(f"Epoch {epoch:2d}: Accuracy = {accuracy:.2%}")

print(f"\nFinal accuracy: {accuracies[-1]:.2%}")

# Looping with index
datasets = ["MNIST", "CIFAR-10", "IMDB", "Boston Housing"]

for i, dataset in enumerate(datasets, 1):
    print(f"{i}. {dataset}")

# Break and Continue
print("\nFinding first good model:")
model_accuracies = [0.65, 0.72, 0.85, 0.88, 0.91]

for accuracy in model_accuracies:
    if accuracy < 0.75:
        continue  # Skip to next iteration
    elif accuracy >= 0.90:
        print(f"⭐ Found excellent model: {accuracy:.2%}")
        break  # Exit loop
    else:
        print(f"✅ Good model: {accuracy:.2%}")
```

### **While Loops**
```python
# Basic while loop
epoch = 1
target_accuracy = 0.90
current_accuracy = 0.70

print("Training until target accuracy reached:")
while current_accuracy < target_accuracy:
    print(f"Epoch {epoch}: Accuracy = {current_accuracy:.2%}")
    current_accuracy += 0.05  # Simulate improvement
    epoch += 1
    if epoch > 20:  # Safety check
        print("⚠️ Stopping - too many epochs!")
        break

print(f"\nTraining completed in {epoch-1} epochs!")
print(f"Final accuracy: {current_accuracy:.2%}")

# Real-time monitoring simulation
print("\n📊 Real-time Model Monitoring:")
data_stream = [0.65, 0.68, 0.72, 0.75, 0.78, 0.82, 0.85, 0.88, 0.90]
index = 0

while index < len(data_stream):
    accuracy = data_stream[index]
    print(f"Batch {index+1}: Accuracy = {accuracy:.2%}")
    
    if accuracy >= 0.85:
        print("  ✅ Reached acceptable accuracy!")
    
    index += 1

# User-controlled loop
def model_training_interface():
    print("🤖 AI Training Interface")
    print("Commands: train, status, stop")
    
    is_training = False
    epoch = 0
    
    while True:
        command = input("\nEnter command: ").lower()
        
        if command == "train":
            is_training = True
            print("Starting training...")
            
        elif command == "status":
            if is_training:
                epoch += 1
                accuracy = 0.70 + (epoch * 0.02)
                print(f"Epoch {epoch}: Accuracy = {accuracy:.2%}")
            else:
                print("Training not started.")
                
        elif command == "stop":
            print("Training stopped.")
            break
            
        else:
            print("Unknown command. Try: train, status, stop")

# Run the interface
# model_training_interface()
```

## **5. Nested Loops in Python**

### **Basic Nested Loops**
```python
# Grid pattern
print("Neural Network Grid:")
for i in range(3):          # Rows
    for j in range(5):      # Columns
        print(f"({i},{j})", end=" ")
    print()  # New line after each row

# Multiplication table (AI parameter tuning)
print("\nLearning Rate x Batch Size combinations:")
learning_rates = [0.1, 0.01, 0.001]
batch_sizes = [16, 32, 64, 128]

print("LR\\BS", end="\t")
for bs in batch_sizes:
    print(f"BS={bs}", end="\t")
print()

for lr in learning_rates:
    print(f"LR={lr}", end="\t")
    for bs in batch_sizes:
        # Simulate some metric
        score = round(lr * bs * 100, 2)
        print(score, end="\t")
    print()
```

### **AI-Specific Examples**

```python
# 1. Hyperparameter Grid Search
print("🔍 Hyperparameter Grid Search:")
learning_rates = [0.1, 0.01, 0.001]
batch_sizes = [32, 64, 128]
optimizers = ['sgd', 'adam', 'rmsprop']

best_score = 0
best_params = {}

for lr in learning_rates:
    for bs in batch_sizes:
        for opt in optimizers:
            # Simulate training score
            score = round(0.7 + (lr * 10) + (bs/100) + 0.05, 3)
            
            print(f"LR={lr}, BS={bs}, Opt={opt} → Score: {score}")
            
            if score > best_score:
                best_score = score
                best_params = {'lr': lr, 'bs': bs, 'opt': opt}

print(f"\n🏆 Best Parameters: {best_params}")
print(f"Best Score: {best_score}")

# 2. Image Processing (Pixel Grid)
print("\n🎨 Simple Image Matrix (3x3):")
image_matrix = [
    [255, 0, 0],    # Red pixel
    [0, 255, 0],    # Green pixel
    [0, 0, 255]     # Blue pixel
]

for row in image_matrix:
    for pixel_value in row:
        # Print colored representation
        if pixel_value == 255:
            print("█", end=" ")  # Bright pixel
        else:
            print("░", end=" ")  # Dark pixel
    print()

# 3. Neural Network Layer Weights
print("\n⚡ Neural Network Weights (3x4x2):")
layer1_neurons = 3
layer2_neurons = 4
layer3_neurons = 2

# Initialize weights (small random values)
import random

print("Layer 1 → Layer 2 connections:")
for i in range(layer1_neurons):
    for j in range(layer2_neurons):
        weight = round(random.uniform(-0.5, 0.5), 3)
        print(f"w{i+1}{j+1} = {weight:7}", end="  ")
    print()

# 4. Confusion Matrix (Classification Evaluation)
print("\n📊 Confusion Matrix (Actual vs Predicted):")
actual = [0, 1, 0, 1, 0, 1, 1, 0]      # Actual labels
predicted = [0, 1, 1, 1, 0, 0, 1, 0]   # Predicted labels
classes = [0, 1]  # Binary classification

print("   Predicted:")
print("    0   1")
print("   ───────")

for actual_class in classes:
    print(f"{actual_class}|", end=" ")
    for pred_class in classes:
        # Count occurrences
        count = 0
        for a, p in zip(actual, predicted):
            if a == actual_class and p == pred_class:
                count += 1
        print(f"{count:2}", end=" ")
    print()
```

### **Nested Loop with Break/Continue**
```python
print("🔬 Finding Optimal Parameters:")
param_space = [
    {'lr': 0.1, 'bs': 32},
    {'lr': 0.01, 'bs': 64},
    {'lr': 0.001, 'bs': 128},
    {'lr': 0.0001, 'bs': 256}
]

for param_set in param_space:
    print(f"\nTesting: LR={param_set['lr']}, BS={param_set['bs']}")
    
    # Simulate training for 5 epochs
    for epoch in range(1, 6):
        # Simulate accuracy
        accuracy = 0.5 + (param_set['lr'] * 10) + (epoch * 0.05)
        
        if accuracy < 0.6:
            print(f"  Epoch {epoch}: {accuracy:.2%} → Too low, skipping rest")
            break  # Break inner loop
        
        print(f"  Epoch {epoch}: {accuracy:.2%}")
        
        if accuracy > 0.85:
            print(f"  ⭐ Excellent! Found good parameters.")
            break  # Break inner loop
    
    if accuracy > 0.85:
        print(f"\n✅ Optimal found: LR={param_set['lr']}, BS={param_set['bs']}")
        break  # Break outer loop
```

### **Practical Exercise: Model Evaluation**
```python
# Evaluate multiple models on multiple datasets
models = ['Logistic Regression', 'Random Forest', 'SVM', 'Neural Network']
datasets = ['Iris', 'Digits', 'Wine', 'Breast Cancer']
results = {}

for model in models:
    print(f"\n📈 Evaluating {model}:")
    model_scores = []
    
    for dataset in datasets:
        # Simulate evaluation score
        base_score = 0.7
        if model == 'Neural Network':
            base_score += 0.1
        if dataset in ['Digits', 'Breast Cancer']:
            base_score += 0.05
        
        score = round(base_score + random.uniform(-0.05, 0.05), 3)
        model_scores.append(score)
        
        print(f"  {dataset:15} → {score:.2%}")
    
    avg_score = sum(model_scores) / len(model_scores)
    results[model] = avg_score
    print(f"  Average: {avg_score:.2%}")

# Find best model
best_model = max(results, key=results.get)
print(f"\n🏆 Best Performing Model: {best_model} ({results[best_model]:.2%})")
```

## **Quick Practice Problems**

1. **Temperature Monitor:**
```python
# Write a program that:
# 1. Takes 5 temperature readings
# 2. Categorizes each as "Normal", "High", or "Critical"
# 3. If 3+ are "Critical", alert the user

temperatures = []
for i in range(5):
    temp = float(input(f"Enter temperature {i+1}: "))
    temperatures.append(temp)

critical_count = 0
for temp in temperatures:
    if temp > 100:
        print(f"{temp}°C: 🔴 Critical!")
        critical_count += 1
    elif temp > 80:
        print(f"{temp}°C: 🟡 High")
    else:
        print(f"{temp}°C: 🟢 Normal")

if critical_count >= 3:
    print("🚨 ALERT: Multiple critical readings!")
```
