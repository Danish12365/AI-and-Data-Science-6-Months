When you first get a dataset, here's a practical order that works well:

**1. Basic overview first**
- Look at the shape (rows, columns)
- Check column names and data types
- Get a quick sense of what you're working with

**2. Missing values**
- This is usually your first real check because missing data affects almost everything else
- See which columns have nulls and how many
- Decide early if you need to drop, impute, or flag them

**3. Data distribution**
- Look at summary statistics (mean, median, min, max)
- Check distributions of key variables
- Spot obvious outliers or weird values

**4. Data quality issues**
- Duplicates
- Invalid values (like negative ages)
- Inconsistent formatting
- Range violations

**5. Relationships and correlations** (if doing modeling)
- How features relate to each other
- How they relate to your target variable

The reason missing values comes before distribution is practical: if 80% of a column is missing, you might drop it entirely rather than spend time analyzing its distribution. Also, many statistical measures don't handle missing values well, so you need to know about them upfront.

That said, steps 1-4 often happen somewhat simultaneously during exploratory data analysis. You're not wrong to look at distributions early - just make sure you're aware of the missing values as you do so.

## **Detecting Outliers**

### What are outliers?
Outliers are data points that are significantly different from the rest of your data. They can be:
- **Legitimate but extreme values** (a billionaire in an income dataset)
- **Errors** (data entry mistakes, sensor malfunctions)
- **Impossible values** (negative age, 200-year-old person)

### Why do outliers matter?
- They can **skew your statistics** (mean, standard deviation)
- They can **mess up models**, especially linear regression
- They might represent **important edge cases** you need to understand
- They could indicate **data quality problems**

### Types of Outliers

**1. Univariate outliers** - extreme in one variable
- Someone who is 7 feet tall
- An income of $10 million when most people make $50k

**2. Multivariate outliers** - unusual combination of values
- A 5-year-old with a PhD (each value alone might be fine, but together they're suspicious)

### Methods to Detect Outliers

**1. Visual Methods**
- **Box plots**: Show you the quartiles and flag points beyond the "whiskers"
- **Scatter plots**: See points that don't fit the pattern
- **Histograms**: Spot isolated bars far from the main distribution

**2. Statistical Methods**

**Z-Score Method:**
- Measures how many standard deviations away from the mean a point is
- Rule of thumb: |z-score| > 3 is often considered an outlier
- Formula: z = (x - mean) / standard deviation
- **Problem**: Assumes normal distribution, sensitive to extreme outliers affecting the mean

**IQR (Interquartile Range) Method:**
- More robust than z-scores
- IQR = Q3 - Q1 (75th percentile minus 25th percentile)
- Outliers are typically defined as:
  - Below Q1 - 1.5×IQR
  - Above Q3 + 1.5×IQR
- This is what box plots use

**Example:**
```
Data: [10, 12, 13, 14, 15, 16, 17, 100]
Q1 = 12.5, Q3 = 16.5
IQR = 4
Lower fence = 12.5 - 1.5(4) = 6.5
Upper fence = 16.5 + 1.5(4) = 22.5
→ 100 is an outlier
```

**3. Domain-Based Detection**
- Age < 0 or > 120: impossible
- Pregnancy duration > 12 months: biologically implausible
- Negative prices (usually): logically impossible
- This is often the most reliable method!

### What to do with outliers?

**DON'T automatically delete them!**

1. **Investigate first**: Why is this value extreme?
2. **Correct if it's an error**: Age of 220 might be a typo for 22
3. **Keep if legitimate**: Real extreme values contain information
4. **Transform the data**: Use log transformation to reduce impact
5. **Use robust methods**: Median instead of mean, robust regression models
6. **Cap/winsorize**: Replace extremes with a max/min threshold
7. **Separate analysis**: Analyze outliers separately to understand them

## **Checking Skewness**

### What is skewness?
Skewness measures the **asymmetry** of your data distribution. It tells you if your data is bunched up on one side or evenly distributed.

### Types of Skewness

**1. Right-skewed (positive skew)**
- Tail extends to the right
- Most data bunched on the left (lower values)
- Mean > Median > Mode
- **Examples**: Income (most people earn modest amounts, few billionaires), house prices, reaction times

```
    *
   ***
  *****
 *******
**********_______________
^         ^
mode/median  mean →
```

**2. Left-skewed (negative skew)**
- Tail extends to the left
- Most data bunched on the right (higher values)
- Mean < Median < Mode
- **Examples**: Age at death (most people live to old age), test scores (if the test is easy)

```
_______________**********
               *******
                *****
                 ***
                  *
      mean ←  ^
          median/mode
```

**3. Symmetric (no skew)**
- Balanced on both sides
- Mean ≈ Median ≈ Mode
- **Example**: Height in a population (roughly normal distribution)

### Measuring Skewness

**Skewness coefficient:**
- **0**: Perfectly symmetric
- **-0.5 to 0.5**: Roughly symmetric
- **0.5 to 1** or **-1 to -0.5**: Moderately skewed
- **> 1** or **< -1**: Highly skewed

### Why does skewness matter?

**1. Statistical tests**
- Many tests (t-tests, ANOVA) assume normality
- Skewed data violates this assumption
- Results might be unreliable

**2. Model performance**
- Linear regression assumes normally distributed residuals
- Skewed features can hurt model performance
- Some algorithms (like linear models) are sensitive to it

**3. Interpretation**
- Mean is misleading in skewed data
- In right-skewed income data: "average income $75k" might sound good, but if median is $35k, most people earn much less

**4. Outliers**
- Highly skewed data often contains outliers
- The skew itself might be caused by outliers

### How to check skewness

**1. Visual inspection**
- Histogram: See the shape directly
- Box plot: If median isn't centered in the box, data is skewed
- Q-Q plot: Compare your data to normal distribution

**2. Calculate skewness coefficient**
Most statistical software can calculate this for you

**3. Compare mean and median**
- Mean ≈ Median → symmetric
- Mean > Median → right-skewed
- Mean < Median → left-skewed

### What to do about skewness?

**1. Transformations** (for right-skewed data)
- **Log transformation**: log(x) - very common for income, prices
- **Square root**: √x - gentler than log
- **Box-Cox transformation**: Finds optimal transformation

**2. For left-skewed data**
- Square transformation: x²
- Exponential transformation
- Or reflect, transform, then reflect back

**3. Use appropriate statistics**
- Report median instead of mean
- Use non-parametric tests (don't assume normality)

**4. Keep it as-is**
- Some models handle skewness well (tree-based models)
- Sometimes the skewness is meaningful and should be preserved

## **Practical Workflow**

When analyzing new data:

1. **Plot histograms** of all numeric variables
2. **Create box plots** to spot outliers visually
3. **Calculate summary stats**: mean, median, min, max, std dev
4. **Check for impossible values** based on domain knowledge
5. **Calculate skewness** for each variable
6. **Investigate outliers**: Are they errors or real?
7. **Decide on transformations** if needed

**Dataset: Employee Salaries**

```
Salaries: $30k, $35k, $40k, $42k, $45k, $48k, $52k, $55k, $500k
```

**Analysis:**
- **Outlier**: $500k (CEO salary) - legitimate but extreme
- **Skewness**: Highly right-skewed (mean $94k, median $45k)
- **Mean is misleading**: "Average salary $94k" sounds great, but most employees make around $45k
- **Action**: Either report median salary, use log transformation for modeling, or analyze CEO separately
