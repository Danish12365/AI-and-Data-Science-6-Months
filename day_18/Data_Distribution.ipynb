To build a solid data foundation, you need to understand both what your data points are (outliers) and how they are arranged (skewness). 

1. Detecting Outliers: Finding the "Odd Ones Out"
Outliers are data points that differ significantly from the rest of your dataset. They can be caused by human error, equipment failure, or rare but legitimate events. 

Visual Detection:
Box Plots: The most common tool. Any points appearing as dots outside the "whiskers" of the box are considered potential outliers.

Scatter Plots: Best for seeing how two variables relate. If one dot is "floating" far away from the main cloud of data, it’s an outlier.

Statistical Detection:
Z-Score: Measures how many standard deviations a point is from the mean. A rule of thumb is that any point with a Z-score greater than 3 or less than -3 is an outlier.
IQR Method: Calculates the "middle 50%" of your data. Any value 1.5 times the IQR above the 75th percentile or below the 25th percentile is flagged. 

2. Checking Skewness: The "Lean" of Your Data
Skewness measures the asymmetry of your data distribution. While a "normal" distribution looks like a symmetric bell curve, real-world data often "leans" one way. 
Positive (Right) Skew:
The Look: The "tail" of the graph stretches out to the right (toward higher numbers).
The Trap: The Mean is pulled higher than the Median by a few large values.
Example: Household income. Most people earn a middle range, but a few billionaires pull the average way up.

Negative (Left) Skew:
The Look: The "tail" stretches out to the left (toward lower numbers).
The Trap: The Mean is dragged down by a few low values, making it lower than the Median.
Example: Retirement age. Most people retire in their 60s, but some retire very early, creating a tail on the left.

How to measure it:
Skewness Coefficient: A value of 0 is perfectly symmetric. Values between -0.5 and 0.5 are considered fairly symmetric, while values beyond ±1 are highly skewed. 
Why These Matter Together
If your data is highly skewed, you cannot rely on the Mean (Average) to describe your data because it's too sensitive to those long tails. In these cases, use the Median as your "typical" value.
