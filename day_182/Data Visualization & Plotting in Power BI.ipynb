# 🔷 2. Column / Bar Charts in Power BI

👉 Most used chart in real-world dashboards

---

## 📊 Difference

| Chart        | Orientation |
| ------------ | ----------- |
| Column Chart | Vertical    |
| Bar Chart    | Horizontal  |

---

## 🪜 Create Column Chart

1. Go to **Visualizations Pane**
2. Select **Clustered Column Chart**
3. Drag:

   * Category → Axis
   * Sales → Values

---

## 🪜 Create Bar Chart

1. Select **Clustered Bar Chart**
2. Drag same fields

---

## 📌 Example

* Axis → Category
* Values → Total Sales

# 🔷 3. Create a Bar Plot Like a Professional

👉 This is where beginners vs professionals differ 👇

---

## 🧩 Step 1: Clean Data

* Remove nulls
* Correct data types

---

## 🧩 Step 2: Sort Data

👉 Click **Sort Descending (Sales)**

📌 Always show:

> Highest → Lowest

---

## 🧩 Step 3: Add Data Labels

* Turn ON “Data Labels”
* Show actual values

---

## 🧩 Step 4: Use Meaningful Colors

👉 Example:

* Profit → Green
* Loss → Red

---

## 🧩 Step 5: Rename Titles

❌ Bad:

```text
Sum of Sales by Category
```

✅ Good:

```text
Top Performing Categories by Sales
```

---

## 🧩 Step 6: Add Tooltips

👉 Hover → more info

---

## 🧩 Step 7: Remove Clutter

* Gridlines (optional)
* Unnecessary labels

---
