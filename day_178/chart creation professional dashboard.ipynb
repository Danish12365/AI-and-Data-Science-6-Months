# 🔷 1. Area Charts in Tableau Public

👉 **Area Chart = Line chart with filled area**

### 📊 Best for:

* Trends over time
* Showing volume (not just direction)

---

## 🪜 Steps to Create

1. Drag **Date → Columns**
2. Drag **Sales (or any measure) → Rows**
3. From “Marks” → select **Area**

---

## 🔥 Make It Better

* Drag **Category → Color** → creates stacked area
* Add **Filter (Region/Year)**

---

## 📌 Types

### 🟢 Simple Area Chart

* Single measure over time

### 🔵 Stacked Area Chart

* Multiple categories stacked

---

## ⚠️ Tip

Avoid too many categories → becomes messy

---

# 🔷 2. Group Charts in Tableau

👉 Grouping = combining similar values into one group

---

## 🧩 Example

You have categories:

* Phones
* Laptops
* Tablets

👉 Group them as:

* “Electronics”

---

## 🪜 Steps

1. Select multiple values
2. Right click → **Group**
3. Rename group

---

## 🔥 Use Cases

* Combine small categories
* Simplify visualization
* Clean dashboards

---

## 💡 Alternative: Sets

👉 Sets = dynamic grouping (advanced)

# 🔷 3. Tableau Interactive Dashboard Development

👉 This is the **MOST IMPORTANT skill for jobs**

---

## 🧠 What is Dashboard?

> Multiple charts + filters + interactivity in one screen

---

## 🪜 Step-by-Step

---

### 🟢 Step 1: Create Charts

Make 3–5 charts:

* Bar chart → Category vs Sales
* Line/Area → Time trend
* Scatter → Profit vs Sales
* Map → Region

---

### 🔵 Step 2: Create Dashboard

1. Click **New Dashboard**
2. Drag charts into layout

---

### 🟡 Step 3: Add Filters

1. Drag field → Filters
2. Right click → Show Filter

👉 Example:

* Region
* Category
* Date

---

### 🔴 Step 4: Add Interactivity

👉 This is what makes it powerful

---

## 🔥 Dashboard Actions

### 🧩 Filter Action

Click one chart → update others

Steps:

1. Dashboard → Actions
2. Add Action → Filter

---

### 🧩 Highlight Action

Hover → highlight related data

---

### 🧩 URL Action (Advanced)

Click → open webpage

---

---

## 🎨 Design Tips

* Keep layout clean
* Use consistent colors
* Add titles
* Use spacing
