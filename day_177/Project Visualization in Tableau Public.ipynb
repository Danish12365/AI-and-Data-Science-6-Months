# 🔷 0. Project Visualization in Tableau Public
---

# 🔷 1. Your Own Project Visualization in Tableau Public

👉 This means:

> Taking your dataset → creating meaningful charts → building a dashboard

---

## 🧠 Step 1: Choose a Project Dataset

Use something **real + useful**

### 💡 Good Ideas (for YOU):

* Sales data
* COVID data (you already worked on this 👍)
* Hotel booking dataset
* Customer churn

---

# 🔷 2. Define Your Goal (IMPORTANT)

Before visualization, ask:

👉 What do I want to show?

---

## 📊 Example Goals:

* Sales trends over time
* Top-performing categories
* Regional performance
* Profit analysis

---

📌 Without goal → dashboard becomes messy

---

# 🔷 3. Create Key Visualizations

---

## 🟢 1. Bar Chart

👉 Show comparison

* Category vs Sales
* Product vs Profit

---

## 🔵 2. Line Chart

👉 Show trends

* Sales over months
* Cases over time

---

## 🟡 3. Scatter Plot

👉 Show relationships

* Sales vs Profit

---

## 🔴 4. Map

👉 Show location-based data

* Sales by country

---

# 🔷 4. Build Dashboard

---

## 🪜 Steps:

1. Click **New Dashboard**
2. Drag sheets into dashboard
3. Arrange layout

---

## 🎨 Design Tips:

* Keep it clean
* Use consistent colors
* Add titles

---

## 📌 Add Interactivity

👉 Add filters:

* Region
* Category
* Date

👉 Users can explore data

---

# 🔷 5. Add Filters & Actions

---

## 🧩 Filter Example:

1. Drag “Region” → Filters
2. Show filter

---

## 🔥 Dashboard Action:

👉 Click chart → updates other charts

---

# 🔷 6. Add Titles & Labels

👉 Make it professional:

* “Sales Dashboard 2024”
* “Profit Analysis by Region”

---

# 🔷 7. Final Project Example (Structure)

---

## 📊 Dashboard Contains:

* Bar chart → Sales by category
* Line chart → Monthly trend
* Scatter plot → Profit vs Sales
* Map → Regional sales

---

👉 This is **portfolio-ready**

---

# 🔷 8. Saving Your Plots in Tableau Public

---

## 💾 Option 1: Save Workbook

👉 File → Save to Tableau Public

---

## 🪜 Steps:

1. Login to Tableau Public
2. Give project name
3. Click Save

---

👉 Your dashboard is now:

* Online
* Public
* Shareable

---

## 🌐 You Get:

* Public link
* Portfolio showcase

---

# 🔷 9. Export Options

## 📌 Save as PDF

👉 File → Print to PDF

## 📌 Share Link

# 🔷 10. Best Practice (VERY IMPORTANT)

---

## ✅ Do This:

* Use clean dataset
* Add filters
* Write clear titles
* Keep dashboard simple

---

## ❌ Avoid:

* Too many charts
* Too many colors
* Confusing layout
