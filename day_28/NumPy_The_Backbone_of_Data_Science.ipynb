# NumPy: The Backbone of Data Science

If Python is the "glue" language of data science, **NumPy** (Numerical Python) is the engine. It is the fundamental library for scientific computing in Python, providing support for large, multi-dimensional arrays and matrices, along with a massive collection of high-level mathematical functions to operate on them.

Think of it as Python lists on espresso—faster, stronger, and much better at math.

---

## 1. The Core: The `ndarray`

The heart of NumPy is the **ndarray** (n-dimensional array). Unlike standard Python lists, which can hold different data types, NumPy arrays are **homogeneous** (all elements must be the same type). This allows them to be packed densely in memory, making operations incredibly fast.

* **1D Array:** A vector (like a single row or column).
* **2D Array:** A matrix (like an Excel spreadsheet).
* **3D Array:** A "cube" of data (often used for images where you have height, width, and color channels).

---

## 2. Why use NumPy instead of Python Lists?

| Feature | Python Lists | NumPy Arrays |
| --- | --- | --- |
| **Data Type** | Heterogeneous (can mix types) | Homogeneous (fixed type) |
| **Speed** | Slower (iterates through objects) | Blazing Fast (utilizes C-based optimizations) |
| **Memory** | High overhead | Low memory footprint |
| **Functionality** | Basic operations | Advanced math, linear algebra, and Fourier transforms |

### The "Vectorization" Magic

In standard Python, to add two lists, you'd need a `for` loop. In NumPy, you use **Vectorization**. This allows you to perform operations on entire arrays at once without writing loops in Python.

> **Direct Comparison:**
> Instead of `[x + y for x, y in zip(list_a, list_b)]`, you simply do `array_a + array_b`.

---

## 3. Key Concepts to Know

### Broadcasting

Broadcasting is a powerful mechanism that allows NumPy to work with arrays of different shapes during arithmetic operations. For example, if you multiply a 2D matrix by a single number (a scalar), NumPy "broadcasts" that number across the entire matrix.

### Universal Functions (ufuncs)

These are functions that operate element-wise on arrays. Examples include:

* `np.sin()`, `np.cos()`
* `np.exp()` (exponential)
* `np.sqrt()` (square root)

## 🛠️ NumPy Quick Reference Cheat Sheet

### 1. Creating Arrays

| Method | Description |
| --- | --- |
| `np.array([1, 2, 3])` | Create a 1D array from a list |
| `np.zeros((3, 4))` | Create a 3x4 array of zeros |
| `np.ones((2, 2))` | Create a 2x2 array of ones |
| `np.full((2, 2), 7)` | Create a 2x2 array filled with the number 7 |
| `np.arange(0, 10, 2)` | Array from 0 to 10 with a step of 2 (`[0, 2, 4, 6, 8]`) |
| `np.linspace(0, 1, 5)` | 5 evenly spaced numbers between 0 and 1 |
| `np.eye(3)` | Create a 3x3 **Identity Matrix** |

### 2. Inspecting Your Array

Suppose you have an array `a`:

* **`a.shape`**: Returns a tuple of dimensions (e.g., `(3, 2)`).
* **`a.ndim`**: Number of dimensions (1, 2, or 3).
* **`a.size`**: Total number of elements in the array.
* **`a.dtype`**: The data type (e.g., `int64`, `float32`).

---

### 3. Slicing and Indexing

NumPy uses a `[row, column]` syntax for indexing.

* **`a[0]`**: Get the first row.
* **`a[:, 1]`**: Get the entire second column.
* **`a[1:3, 0:2]`**: Get rows 1 to 2 and columns 0 to 1 (sub-matrix).
* **`a[a > 5]`**: **Boolean Indexing**—returns all elements greater than 5.

---

### 4. Mathematical Operations

Operations are performed **element-wise**.

* **`a + b`** or **`np.add(a, b)`**: Addition.
* **`a * b`**: Element-wise multiplication.
* **`a @ b`** or **`np.dot(a, b)`**: **Matrix multiplication** (dot product).
* **`np.sqrt(a)`**: Square root of every element.
* **`np.exp(a)`**: Exponential () of every element.

---

### 5. Aggregation Functions

These help you summarize your data:

* **`a.sum()`**: Sum of all elements.
* **`a.mean()`**: Average of all elements.
* **`a.max(axis=0)`**: Maximum values along the columns.
* **`a.std()`**: Standard deviation.

---

### 6. Reshaping and Transposing

* **`a.reshape(rows, cols)`**: Changes the shape without changing the data.
> *Note: The new shape must have the same total number of elements.*


* **`a.T`**: Transposes the matrix (swaps rows and columns).
* **`a.flatten()`**: Collapses a multi-dimensional array into 1D.

---

### 💡 Pro-Tip: Broadcasting

Broadcasting allows NumPy to perform math on arrays of different sizes. If you add a 1D array to a 2D matrix, NumPy "stretches" the smaller array to match the larger one, provided their dimensions are compatible.
