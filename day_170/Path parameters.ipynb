Path parameters are one of the **core features** of APIs in FastAPI. They allow you to pass **dynamic values inside the URL**.

Let’s break it down step-by-step 👇

---

# 🔷 1. What are Path Parameters?

👉 Path parameters = values inside the URL path

### 📌 Example URL:

```text
/books/10
```

👉 Here:

* `10` = path parameter (book_id)

---

## 🧠 Simple Definition

> Path parameters are variables in the URL used to identify specific resources.

---

# 🔷 2. Basic Example

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/books/{book_id}")
def get_book(book_id: int):
    return {"book_id": book_id}
```

---

## 🔍 How it Works

👉 URL:

```text
/books/5
```

👉 Output:

```json
{
  "book_id": 5
}
```

---

# 🔷 3. Data Types in Path Parameters

FastAPI automatically validates types 👇

```python
@app.get("/items/{item_id}")
def get_item(item_id: int):
    return {"item_id": item_id}
```

---

## 📌 If wrong type:

```text
/items/abc
```

👉 Error:

* Validation error (FastAPI handles it automatically)

---

# 🔷 4. Multiple Path Parameters

```python
@app.get("/users/{user_id}/orders/{order_id}")
def get_order(user_id: int, order_id: int):
    return {"user_id": user_id, "order_id": order_id}
```

---

## 📌 Example:

```text
/users/2/orders/45
```

---

# 🔷 5. Path Parameters with Query Parameters

```python
@app.get("/books/{book_id}")
def get_book(book_id: int, detail: bool = False):
    return {"book_id": book_id, "detail": detail}
```

---

## 📌 Example:

```text
/books/5?detail=true
```

---

# 🔷 6. Path Parameters with Pydantic Model

```python
from pydantic import BaseModel

class Book(BaseModel):
    title: str
    price: float

@app.put("/books/{book_id}")
def update_book(book_id: int, book: Book):
    return {"book_id": book_id, "book": book}
```

---

# 🔷 7. Path Parameter Validation (Advanced)

```python
from fastapi import Path

@app.get("/books/{book_id}")
def get_book(book_id: int = Path(..., gt=0, lt=100)):
    return {"book_id": book_id}
```

---

## 📌 Meaning:

* `gt=0` → greater than 0
* `lt=100` → less than 100

---

# 🔷 8. Add Description for Swagger

```python
@app.get("/books/{book_id}")
def get_book(
    book_id: int = Path(..., description="The ID of the book", example=1)
):
    return {"book_id": book_id}
```

👉 Shows in Swagger UI

---

# 🔷 9. Enum Path Parameters (Limited Choices)

```python
from enum import Enum

class Category(str, Enum):
    fiction = "fiction"
    science = "science"

@app.get("/books/{category}")
def get_books(category: Category):
    return {"category": category}
```

---

## 📌 Example:

```text
/books/fiction
```

---

# 🔷 10. Path vs Query Parameters

| Feature  | Path Parameter | Query Parameter   |
| -------- | -------------- | ----------------- |
| Location | URL path       | After `?`         |
| Required | Yes            | Optional          |
| Example  | `/books/5`     | `/books?limit=10` |
