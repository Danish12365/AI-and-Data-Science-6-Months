# Text functions in MySQL

**Text functions** in MySQL allow you to manipulate and work with string data, while filtering the output refers to retrieving specific rows based on certain conditions. By combining text functions with `WHERE` clauses, you can filter and refine your query results based on string patterns, lengths, and other textual characteristics.

### 1. Common Text Functions in MySQL

Here are some of the most commonly used text functions in MySQL:

- **`UPPER()` / `LOWER()`**: Converts a string to uppercase or lowercase.
- **`LENGTH()`**: Returns the length of a string.
- **`SUBSTRING()`**: Extracts a substring from a string.
- **`CONCAT()`**: Concatenates multiple strings into one.
- **`TRIM()`**: Removes leading and trailing spaces from a string.
- **`REPLACE()`**: Replaces occurrences of a substring within a string.
- **`INSTR()`**: Returns the position of the first occurrence of a substring within a string.
- **`LEFT()` / `RIGHT()`**: Extracts a specified number of characters from the left or right of a string.
- **`REVERSE()`**: Reverses the characters in a string.
- **`LPAD()` / `RPAD()`**: Pads a string on the left or right with a specified character to a certain length.

### 2. Filtering Output Using Text Functions

Text functions can be used in `WHERE` clauses to filter query results based on string conditions. Let's explore some examples.

#### 2.1 Filtering Based on Case

Suppose you have a table `customers` and you want to filter out all customers whose names are stored in uppercase.

```sql
SELECT * FROM customers
WHERE UPPER(name) = name;
```

- This query retrieves all rows where the `name` column is already in uppercase.

#### 2.2 Filtering Based on Length

You may want to filter rows based on the length of a string. For example, select all customers whose names are longer than 10 characters.

```sql
SELECT * FROM customers
WHERE LENGTH(name) > 10;
```

- This query returns all customers with names longer than 10 characters.

#### 2.3 Filtering Using Substrings

If you want to filter results based on a part of a string, you can use the `SUBSTRING()` function.

```sql
SELECT * FROM customers
WHERE SUBSTRING(phone_number, 1, 3) = '555';
```

- This query retrieves all customers whose phone numbers start with '555'.

#### 2.4 Filtering Using `LIKE` with Text Functions

The `LIKE` operator can be combined with text functions to filter rows based on patterns.

```sql
SELECT * FROM customers
WHERE UPPER(name) LIKE 'J%';
```

- This query retrieves all customers whose names (converted to uppercase) start with the letter 'J'.

#### 2.5 Filtering Using `INSTR()`

You can use the `INSTR()` function to filter rows where a substring exists within a string.

```sql
SELECT * FROM customers
WHERE INSTR(email, '@gmail.com') > 0;
```

- This query retrieves all customers whose email addresses contain '@gmail.com'.

#### 2.6 Filtering Based on Replacement

To filter rows based on whether a specific substring can be replaced:

```sql
SELECT * FROM products
WHERE REPLACE(description, 'Free', '') <> description;
```

- This query retrieves all products whose description contains the word 'Free'.

### 3. Combining Multiple Filters

You can combine multiple text-based filters using `AND` or `OR` operators.

```sql
SELECT * FROM employees
WHERE LENGTH(last_name) > 5
AND LOWER(first_name) LIKE 'a%'
AND INSTR(email, 'company.com') > 0;
```

- This query retrieves all employees whose last names are longer than 5 characters, first names start with 'a', and email contains 'company.com'.

### 4. Using Text Functions in `ORDER BY`

You can also use text functions in the `ORDER BY` clause to sort the results.

```sql
SELECT * FROM products
ORDER BY LENGTH(product_name) DESC;
```

- This query sorts the products by the length of the product name in descending order.

Concatenation in MySQL involves combining two or more strings into one continuous string. This is often used to merge text fields, create custom outputs, or format data for reporting or display purposes.

### 1. Basic Concatenation with `CONCAT()`

The primary function for concatenation in MySQL is `CONCAT()`. This function can take two or more string arguments and returns a single concatenated string.

**Syntax:**

```sql
CONCAT(string1, string2, ..., stringN)
```

- `string1`, `string2`, ..., `stringN`: These are the strings you want to concatenate.

**Example:**

```sql
SELECT CONCAT(first_name, ' ', last_name) AS full_name
FROM customers;
```

- This query concatenates the `first_name` and `last_name` columns with a space (`' '`) in between and returns the full name of each customer.

### 2. Concatenating with Static Text

You can also concatenate columns with static text or other string literals.

**Example:**

```sql
SELECT CONCAT('Customer: ', first_name, ' ', last_name) AS customer_details
FROM customers;
```

- This query creates a string like "Customer: John Doe" by concatenating static text with the `first_name` and `last_name`.

### 3. Handling `NULL` Values with `CONCAT()`

When any of the arguments to `CONCAT()` are `NULL`, the result is `NULL`. However, MySQL provides the `CONCAT_WS()` function to handle this situation.

### 4. Using `CONCAT_WS()` for Safe Concatenation

`CONCAT_WS()` stands for "Concatenate With Separator" and allows you to specify a separator between the strings you want to concatenate. Importantly, `CONCAT_WS()` ignores `NULL` values, which prevents `NULL` from causing the entire result to be `NULL`.

**Syntax:**

```sql
CONCAT_WS(separator, string1, string2, ..., stringN)
```

- `separator`: The string to insert between each of the other strings (e.g., a comma, space, or hyphen).
- `string1`, `string2`, ..., `stringN`: The strings you want to concatenate.

**Example:**

```sql
SELECT CONCAT_WS(', ', first_name, last_name, email) AS customer_info
FROM customers;
```

- This query concatenates the `first_name`, `last_name`, and `email` columns, separated by a comma and a space. If any of these columns contain `NULL`, they will be ignored in the output.

### 5. Practical Examples of Concatenation

#### 5.1 Creating Full Addresses

You might want to combine multiple address components into a single string.

**Example:**

```sql
SELECT CONCAT_WS(', ', street_address, city, state, postal_code) AS full_address
FROM customers;
```

- This query concatenates the `street_address`, `city`, `state`, and `postal_code` into a full address, with each component separated by a comma and a space.

#### 5.2 Formatting Phone Numbers

Concatenation can also be used to format phone numbers.

**Example:**

```sql
SELECT CONCAT('(', SUBSTRING(phone_number, 1, 3), ') ', SUBSTRING(phone_number, 4, 3), '-', SUBSTRING(phone_number, 7)) AS formatted_phone
FROM customers;
```

- This query formats a phone number like "5551234567" into "(555) 123-4567" using `CONCAT()` and `SUBSTRING()`.

### 6. Combining Concatenation with Other Functions

You can combine `CONCAT()` with other MySQL functions to create complex string manipulations.

**Example:**

```sql
SELECT CONCAT(UPPER(first_name), ' ', LOWER(last_name)) AS styled_name
FROM customers;
```

- This query concatenates the uppercase version of the `first_name` with the lowercase version of the `last_name`.

- **Generating URLs**: Concatenate base URLs with dynamic paths or query parameters.
- **Building Descriptions**: Merge different fields to create a product or service description.

In MySQL, combining date, time, or other components into a readable format often involves the use of functions like `CONCAT()`, `DATE_FORMAT()`, and `TIME_FORMAT()`. These functions allow you to format dates, times, and other fields in a more user-friendly or customized way.

### 1. Formatting Date and Time Using `DATE_FORMAT()`

`DATE_FORMAT()` allows you to format `DATE` or `DATETIME` values according to a specific pattern.

**Syntax:**

```sql
DATE_FORMAT(date, format)
```

- `date`: The `DATE` or `DATETIME` column you want to format.
- `format`: The string format you want to apply, where special placeholders like `%Y` (year), `%m` (month), `%d` (day), etc., are used.

### Example 1: Formatting a Date

Let’s say you want to convert a `DATE` column (`order_date`) into a more readable format such as "March 10, 2024".

```sql
SELECT DATE_FORMAT(order_date, '%M %d, %Y') AS formatted_date
FROM orders;
```

- `%M`: Full month name (e.g., "March")
- `%d`: Day of the month, two digits (e.g., "10")
- `%Y`: Full year (e.g., "2024")

### Output:

| order_date  | formatted_date     |
|-------------|--------------------|
| 2024-03-10  | March 10, 2024     |
| 2024-04-22  | April 22, 2024     |

### Example 2: Formatting a Date and Time

If the `orders` table has a `DATETIME` column (`order_timestamp`), and you want to display both date and time in a specific format:

```sql
SELECT DATE_FORMAT(order_timestamp, '%M %d, %Y at %h:%i %p') AS formatted_datetime
FROM orders;
```

- `%M %d, %Y`: Full date format ("March 10, 2024")
- `%h`: Hour (12-hour format)
- `%i`: Minutes
- `%p`: AM/PM

### Output:

| order_timestamp      | formatted_datetime         |
|----------------------|----------------------------|
| 2024-03-10 14:35:00  | March 10, 2024 at 02:35 PM |
| 2024-04-22 09:15:00  | April 22, 2024 at 09:15 AM |

### 2. Combining Date and Time Components with `CONCAT()`

You can also combine the components of `DATE`, `TIME`, or `DATETIME` fields using `CONCAT()` and `DATE_FORMAT()` for more customized formatting.

### Example 3: Concatenating Date and Time Fields

Assume you have separate `date` and `time` columns in the `appointments` table, and you want to combine them into one formatted string.

```sql
SELECT CONCAT(
    DATE_FORMAT(appointment_date, '%M %d, %Y'), ' at ', 
    TIME_FORMAT(appointment_time, '%h:%i %p')
) AS formatted_appointment
FROM appointments;
```

- `DATE_FORMAT(appointment_date, '%M %d, %Y')`: Formats the date as "Month Day, Year".
- `TIME_FORMAT(appointment_time, '%h:%i %p')`: Formats the time as "Hour:Minute AM/PM".
- The `CONCAT()` function is used to join the formatted date and time with the text `" at "`.

### Output:

| appointment_date | appointment_time | formatted_appointment        |
|------------------|------------------|------------------------------|
| 2024-03-10       | 14:30:00         | March 10, 2024 at 02:30 PM   |
| 2024-04-22       | 09:15:00         | April 22, 2024 at 09:15 AM   |

### 3. Formatting Date Components Separately

You can also format date components (year, month, day) separately and combine them using `CONCAT()`.

### Example 4: Custom Date Format

Suppose you want to display a date as `Day-Month-Year`:

```sql
SELECT CONCAT(DAY(order_date), '-', MONTH(order_date), '-', YEAR(order_date)) AS custom_date
FROM orders;
```

- `DAY(order_date)`: Extracts the day part of the date.
- `MONTH(order_date)`: Extracts the month part of the date.
- `YEAR(order_date)`: Extracts the year part of the date.

### Output:

| order_date  | custom_date |
|-------------|-------------|
| 2024-03-10  | 10-3-2024   |
| 2024-04-22  | 22-4-2024   |

### 4. Formatting Time Using `TIME_FORMAT()`

Just as `DATE_FORMAT()` handles date formatting, `TIME_FORMAT()` allows for custom time formatting.

**Syntax:**

```sql
TIME_FORMAT(time, format)
```

### Example 5: Formatting Time

You want to display the `time` column in a `HH:MM AM/PM` format:

- `%p`: AM/PM

### 5. Combining Date, Time, and Other Components

You can use `CONCAT()` to combine date, time, and additional text or information into a single string.

### Example 6: Full Report of Date and Time

If you need a custom report combining both date and time components along with other details, you can do something like this:

```sql
SELECT CONCAT('Order placed on ', 
              DATE_FORMAT(order_date, '%M %d, %Y'),
              ' at ', 
              TIME_FORMAT(order_time, '%h:%i %p')
) AS order_report
FROM orders;
```

You can customize the format to match your requirements, such as showing dates as "Month Day, Year" or times as "Hour:Minute AM/PM".

MySQL provides several text functions that are useful for manipulating and analyzing string data. Here are some other important text functions, along with examples of how they can be used:

### 1. **`TRIM()`**

Removes leading and trailing spaces from a string.

**Syntax:**

```sql
TRIM([LEADING | TRAILING | BOTH] trim_string FROM string)
```

**Example:**

```sql
SELECT TRIM('   Hello World   ') AS trimmed_string;
```

- **Output:** `Hello World`

### 2. **`LENGTH()`**

Returns the length of a string in bytes.

**Syntax:**

```sql
LENGTH(string)
```

**Example:**

```sql
SELECT LENGTH('Hello World') AS string_length;
```

- **Output:** `11` (since 'Hello World' has 11 characters)

### 3. **`CHAR_LENGTH()` / `CHARACTER_LENGTH()`**

Returns the number of characters in a string (counts multi-byte characters as one).

**Syntax:**

```sql
CHAR_LENGTH(string)
```

**Example:**

```sql
SELECT CHAR_LENGTH('Hello World') AS char_length;
```

- **Output:** `11`

### 4. **`UPPER()` / `LOWER()`**

Converts all characters in a string to uppercase or lowercase.

**Syntax:**

```sql
UPPER(string)
LOWER(string)
```

**Example:**

```sql
SELECT UPPER('Hello World') AS upper_case, LOWER('Hello World') AS lower_case;
```

- **Output:**

  | upper_case    | lower_case    |
  |---------------|---------------|
  | HELLO WORLD   | hello world   |

### 5. **`REPLACE()`**

Replaces occurrences of a substring with another substring.

**Syntax:**

```sql
REPLACE(string, old_substring, new_substring)
```

**Example:**

```sql
SELECT REPLACE('Hello World', 'World', 'MySQL') AS replaced_string;
```

- **Output:** `Hello MySQL`

### 6. **`LEFT()` / `RIGHT()`**

Extracts a specified number of characters from the beginning or end of a string.

**Syntax:**

```sql
LEFT(string, length)
RIGHT(string, length)
```

**Example:**

```sql
SELECT LEFT('Hello World', 5) AS left_string, RIGHT('Hello World', 5) AS right_string;
```

- **Output:**

  | left_string | right_string |
  |-------------|--------------|
  | Hello       | World        |

### 7. **`SUBSTRING()` / `SUBSTR()`**

Extracts a substring from a string starting at a specified position.

**Syntax:**

```sql
SUBSTRING(string, start, length)
SUBSTR(string, start, length)
```

**Example:**

```sql
SELECT SUBSTRING('Hello World', 7, 5) AS substring;
```

- **Output:** `World`

### 8. **`LOCATE()`**

Returns the position of the first occurrence of a substring within a string.

**Syntax:**

```sql
LOCATE(substring, string, [start_position])
```

**Example:**

```sql
SELECT LOCATE('World', 'Hello World') AS position;
```

- **Output:** `7` (since 'World' starts at the 7th position in 'Hello World')

### 9. **`FIND_IN_SET()`**

Returns the position of a string within a comma-separated list of strings.

**Syntax:**

```sql
FIND_IN_SET(string, list)
```

**Example:**

```sql
SELECT FIND_IN_SET('apple', 'banana,apple,orange') AS position;
```

- **Output:** `2` (since 'apple' is the 2nd item in the list)

### 10. **`CONCAT_WS()`**

Concatenates strings with a separator, and skips `NULL` values.

**Syntax:**

```sql
CONCAT_WS(separator, string1, string2, ..., stringN)
```

**Example:**

```sql
SELECT CONCAT_WS(' - ', 'January', 'February', 'March') AS formatted_string;
```

- **Output:** `January - February - March`

### 11. **`REPEAT()`**

Repeats a string a specified number of times.

**Syntax:**

```sql
REPEAT(string, count)
```

**Example:**

```sql
SELECT REPEAT('abc', 3) AS repeated_string;
```

- **Output:** `abcabcabc`

### 12. **`TRIM()`**

Removes specified prefixes or suffixes from a string. If no prefix or suffix is specified, it defaults to trimming spaces.

**Syntax:**

```sql
TRIM([[LEADING | TRAILING | BOTH] trim_string FROM] string)
```

**Example:**

```sql
SELECT TRIM(LEADING 'x' FROM 'xxxHello Worldxxx') AS trimmed_string;
```

- **Output:** `Hello Worldxxx`

### 13. **`ASCII()`**

Returns the ASCII value of the first character in a string.

**Syntax:**

```sql
ASCII(string)
```

**Example:**

```sql
SELECT ASCII('A') AS ascii_value;
```

- **Output:** `65` (ASCII value for 'A')

### 14. **`CHAR()`**

Returns the string that represents the specified ASCII value.

**Syntax:**

```sql
CHAR(number)
```

**Example:**

```sql
SELECT CHAR(65) AS character;
```

- **Output:** `A` (character for ASCII value 65)

### Conclusion

These text functions are powerful tools for string manipulation in MySQL. Whether you need to format, search, or transform text data, these functions provide a wide range of capabilities to suit your needs. Combining these functions effectively can help you manage and present textual data more efficiently in your database queries.
