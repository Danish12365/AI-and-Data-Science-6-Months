   ```sql
   SELECT CEIL(6);   -- Output: 6
   SELECT CEIL(-3);  -- Output: -3
   ```

### **Explanation of the `ROUND()` Function with a Negative Precision Value:**

When you use the `ROUND()` function in MySQL with a **negative precision** (i.e., a negative value for the second argument), the function rounds the number to the left of the decimal point, rather than rounding the decimal places.

```sql
ROUND(number, decimal_places)
```

- **number**: The number you want to round.
- **decimal_places**: The number of decimal places to round to. If negative, it rounds the digits to the left of the decimal point.

#### **How Negative Precision Works:**
- A **positive value** in `decimal_places` rounds the number to the right of the decimal point.
- A **negative value** rounds to the nearest 10, 100, 1000, etc., depending on the magnitude of the negative value.

#### **Example:**
```sql
SELECT ROUND(123.456, -1);  -- Output: 120
```

- The number `123.456` is being rounded to **one digit to the left of the decimal point** because the second argument is `-1`.
- This means it will round to the nearest 10 (ten's place).
- **123.456** rounds down to **120** because **3** (the digit in the tens place) is closer to **120** than to **130**.


### **Summary:**
- When you pass a negative value as the second argument in `ROUND()`, it rounds the number to the left of the decimal point.

### **RADIANS() in MySQL**

#### **Description:**
The `RADIANS()` function in MySQL is used to convert a value measured in **degrees** into **radians**. This is commonly used in trigonometric calculations, as many mathematical functions such as `SIN()`, `COS()`, and `TAN()` in most programming languages, including MySQL, expect the angle to be in radians rather than degrees.

#### **Degrees and Radians:**
To understand `RADIANS()`, it's important to know the difference between **degrees** and **radians**.

- **Degrees**: A degree is a unit for measuring angles. There are **360 degrees** in a full circle. It’s the traditional way of expressing angles.
- **Radians**: A radian is another way to measure angles, and it is more commonly used in mathematical contexts, particularly in calculus and trigonometry. There are **2π radians** in a full circle.

The relationship between degrees and radians can be expressed by the following formula:
```text
Radians = Degrees * (π / 180)
```

#### **Syntax:**
```sql
RADIANS(degree_value)
```

#### **How It Works:**
- You provide an angle in degrees as input, and the `RADIANS()` function converts that angle to its equivalent in radians.

#### **Examples:**

1. **Converting 180 degrees to radians**:
   ```sql
   SELECT RADIANS(180);  -- Output: 3.1416 (approximately π)
   ```
   Explanation: 
   - \( 180^\circ \) is equivalent to \( \pi \) radians, and the function returns the value of \( \pi \) (~3.1416).

2. **Converting 90 degrees to radians**:
   ```sql
   SELECT RADIANS(90);  -- Output: 1.5708
   ```
   Explanation:
   - \( 90^\circ \) is one-quarter of a full circle. In radians, this is \( \pi/2 \), which is approximately 1.5708.

3. **Converting 45 degrees to radians**:
   ```sql
   SELECT RADIANS(45);  -- Output: 0.7854
   ```
   Explanation:
   - \( 45^\circ \) is one-eighth of a full circle. In radians, this is \( \pi/4 \), which is approximately 0.7854.

4. **Converting 360 degrees to radians**:
   ```sql
   SELECT RADIANS(360);  -- Output: 6.2832
   ```
   Explanation:
   - A full circle in radians is \( 2\pi \), which is approximately 6.2832.

#### **Concept Behind Radians:**
Radians provide a more direct and mathematically elegant way to describe angles because they are directly related to the arc length of a circle. For example, if you have a circle with radius 1 (a unit circle):
- An angle of **1 radian** will subtend an arc on the circle whose length is equal to the radius (i.e., 1 unit of arc length).
- A full rotation around the circle (360 degrees) corresponds to \( 2\pi \) radians, since the circumference of a circle is \( 2\pi r \), and for a unit circle, \( r = 1 \).

Using radians is especially useful when you are working with trigonometric functions, derivatives, and integrals, as many of the formulas and relationships in trigonometry are simpler when expressed in radians.

#### **Use Cases in MySQL:**
- **Trigonometric Calculations**: If you are using functions like `SIN()`, `COS()`, `TAN()`, and their inverses (e.g., `ASIN()`, `ACOS()`), you need to work with angles in radians, so the `RADIANS()` function is essential to convert your degree-based angles.
  
Example of using `RADIANS()` in a trigonometric function:
```sql
SELECT SIN(RADIANS(90));  -- Output: 1 (because sin(π/2) = 1)
```

This converts 90 degrees to \( \pi/2 \) radians, and the sine of \( \pi/2 \) is 1.

#### **Additional Example:**
You can use the `RADIANS()` function to convert multiple angles at once:
```sql
SELECT angle_in_degrees, RADIANS(angle_in_degrees) AS angle_in_radians
FROM angles_table;
```
This will convert all the angle values stored in the `angles_table` from degrees to radians.

#### **Summary:**
- The `RADIANS()` function converts degrees to radians using the formula \( Radians = Degrees \times (\pi / 180) \).
- Radians are the standard angular unit for most mathematical and trigonometric functions.
- This function is critical in MySQL when you need to perform trigonometric operations, as those functions expect inputs in radians.

### **Concept of Trigonometry:**

**Trigonometry** is a branch of mathematics that studies the relationships between the angles and sides of triangles. It has applications in various fields including geometry, physics, engineering, and even computer graphics. The core functions of trigonometry are the **trigonometric functions**, which relate angles to side lengths in right-angled triangles and circular motion.

#### **1. Basics of Trigonometry:**

1. **Right-Angled Triangle**:
   - A right-angled triangle is a triangle with one angle exactly equal to 90 degrees.
   - The side opposite the right angle is called the **hypotenuse**.
   - The side opposite the angle of interest is called the **opposite** side.
   - The side adjacent to the angle of interest (but not the hypotenuse) is called the **adjacent** side.

2. **Basic Trigonometric Functions**:
   In a right-angled triangle with an angle \( \theta \):
   - **Sine (sin)**: Ratio of the length of the opposite side to the hypotenuse.
     \[
     \sin(\theta) = \frac{\text{Opposite}}{\text{Hypotenuse}}
     \]
   - **Cosine (cos)**: Ratio of the length of the adjacent side to the hypotenuse.
     \[
     \cos(\theta) = \frac{\text{Adjacent}}{\text{Hypotenuse}}
     \]
   - **Tangent (tan)**: Ratio of the length of the opposite side to the adjacent side.
     \[
     \tan(\theta) = \frac{\text{Opposite}}{\text{Adjacent}}
     \]

3. **Reciprocal Functions**:
   - **Cosecant (csc)**: Reciprocal of sine.
     \[
     \csc(\theta) = \frac{1}{\sin(\theta)}
     \]
   - **Secant (sec)**: Reciprocal of cosine.
     \[
     \sec(\theta) = \frac{1}{\cos(\theta)}
     \]
   - **Cotangent (cot)**: Reciprocal of tangent.
     \[
     \cot(\theta) = \frac{1}{\tan(\theta)}
     \]

#### **2. Unit Circle:**

The unit circle is a circle with a radius of 1 centered at the origin of a coordinate system. It is used to define trigonometric functions for all angles, not just those in right-angled triangles.

- **Angle Measurement**:
  - **Degrees**: A full circle is 360 degrees.
  - **Radians**: A full circle is \(2\pi\) radians. Conversion: \( \text{Radians} = \text{Degrees} \times \frac{\pi}{180} \).

- **Function Values**:
  On the unit circle, any point \((x, y)\) on the circle corresponding to an angle \( \theta \) can be used to find:
  \[
  \cos(\theta) = x
  \]
  \[
  \sin(\theta) = y
  \]

#### **3. Key Trigonometric Identities:**

1. **Pythagorean Identities**:
   - \[
     \sin^2(\theta) + \cos^2(\theta) = 1
     \]
   - \[
     1 + \tan^2(\theta) = \sec^2(\theta)
     \]
   - \[
     1 + \cot^2(\theta) = \csc^2(\theta)
     \]

2. **Angle Sum and Difference Identities**:
   - \[
     \sin(a \pm b) = \sin(a) \cos(b) \pm \cos(a) \sin(b)
     \]
   - \[
     \cos(a \pm b) = \cos(a) \cos(b) \mp \sin(a) \sin(b)
     \]

3. **Double-Angle Identities**:
   - \[
     \sin(2\theta) = 2 \sin(\theta) \cos(\theta)
     \]
   - \[
     \cos(2\theta) = \cos^2(\theta) - \sin^2(\theta)
     \]

4. **Half-Angle Identities**:
   - \[
     \sin^2(\frac{\theta}{2}) = \frac{1 - \cos(\theta)}{2}
     \]
   - \[
     \cos^2(\frac{\theta}{2}) = \frac{1 + \cos(\theta)}{2}
     \]

#### **4. Applications of Trigonometry:**

1. **Geometry**:
   - Used to determine lengths of sides and angles in triangles.
   - Essential for solving problems involving polygons and circles.

2. **Physics**:
   - Analyzes oscillations, wave functions, and periodic phenomena.
   - Used in mechanics, optics, and electromagnetism.

3. **Engineering**:
   - Essential in designing and analyzing structures, electrical circuits, and mechanical systems.

4. **Computer Graphics**:
   - Used to perform rotations, translations, and scaling of objects.

5. **Navigation and Surveying**:
   - Helps in determining distances and angles in land surveying and navigation.

#### **5. Trigonometric Functions in Calculus:**

- **Derivatives**: The derivatives of sine and cosine functions are used in calculus.
  - \[
    \frac{d}{dx}[\sin(x)] = \cos(x)
    \]
  - \[
    \frac{d}{dx}[\cos(x)] = -\sin(x)
    \]

- **Integrals**: Trigonometric functions are also integrated in calculus.
  - \[
    \int \sin(x) \, dx = -\cos(x) + C
    \]
  - \[
    \int \cos(x) \, dx = \sin(x) + C
    \]

Trigonometry is a foundational mathematical discipline that explores the relationships between angles and sides in triangles and extends to the unit circle for all angles. Its functions and identities are pivotal in various scientific and engineering fields.
