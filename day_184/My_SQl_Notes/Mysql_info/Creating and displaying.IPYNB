# Creating and displaying databases in MySQL

Creating and displaying databases in MySQL is one of the fundamental tasks for any database administrator or developer. Here, I'll cover this topic comprehensively, including syntax, use cases, examples, and best practices.

## 1. **Creating Databases in MySQL**

### 1.1 Basic Syntax

To create a database in MySQL, you use the `CREATE DATABASE` statement.

#### Syntax:

```sql
CREATE DATABASE database_name;
```

#### Example:

```sql
CREATE DATABASE school;
```

This command creates a database named `school`.

### 1.2 Specifying Character Set and Collation

When creating a database, you can specify the character set and collation, which determine how text data is stored and compared.

#### Syntax:

```sql
CREATE DATABASE database_name
CHARACTER SET charset_name
COLLATE collation_name;
```

#### Example:

```sql
CREATE DATABASE school
CHARACTER SET utf8mb4
COLLATE utf8mb4_unicode_ci;
```

In this example, the `utf8mb4` character set is used, which supports a wide range of characters including emojis, and `utf8mb4_unicode_ci` is the collation that determines how text is sorted and compared.

### 1.3 IF NOT EXISTS Clause

To avoid errors if the database already exists, you can use the `IF NOT EXISTS` clause.

#### Syntax:

```sql
CREATE DATABASE IF NOT EXISTS database_name;
```

#### Example:

```sql
CREATE DATABASE IF NOT EXISTS school;
```

This command will only create the database if it doesn’t already exist, preventing an error message.

## 2. **Displaying Databases**

Once you have databases created, you might want to list or display them.

### 2.1 SHOW DATABASES Command

The `SHOW DATABASES` command lists all databases available on the MySQL server.

#### Syntax:

```sql
SHOW DATABASES;
```

#### Example:

```sql
SHOW DATABASES;
```

This command will return a list of all databases on the server.

### 2.2 Displaying a Specific Database's Information

To get detailed information about a specific database, you can use the `SHOW CREATE DATABASE` command.

#### Syntax:

```sql
SHOW CREATE DATABASE database_name;
```

#### Example:

```sql
SHOW CREATE DATABASE school;
```

This will show the `CREATE DATABASE` statement that was used to create the database, including any character set or collation settings.

### 2.3 Selecting a Database

Before you can work with the tables or other objects in a database, you need to select it.

#### Syntax:

```sql
USE database_name;
```

#### Example:

```sql
USE school;
```

This command sets the `school` database as the current database for the session. All subsequent commands will be executed against this database unless another database is selected.

## 3. **Modifying a Database**

Although MySQL doesn’t provide a direct command to modify an existing database (e.g., to change its character set), you can achieve similar results by creating a new database with the desired settings and copying the data from the old one.

### 3.1 Changing Database Character Set

To change the character set of a database, you can use the following steps:

1. Create a new database with the desired character set.
2. Export the data from the old database.
3. Import the data into the new database.

#### Example:

```sql
CREATE DATABASE new_school
CHARACTER SET utf8mb4
COLLATE utf8mb4_unicode_ci;

-- Export data from old database
mysqldump -u username -p old_school > old_school.sql

-- Import data into new database
mysql -u username -p new_school < old_school.sql
```

## 4. **Dropping a Database**

To delete a database, you use the `DROP DATABASE` statement. This operation is irreversible and will remove all data, tables, and other objects within the database.

### 4.1 Basic Syntax

#### Syntax:

```sql
DROP DATABASE database_name;
```

#### Example:

```sql
DROP DATABASE school;
```

This command deletes the `school` database and all of its contents.

### 4.2 Using IF EXISTS Clause

To avoid errors if the database doesn’t exist, use the `IF EXISTS` clause.

#### Syntax:

```sql
DROP DATABASE IF EXISTS database_name;
```

#### Example:

```sql
DROP DATABASE IF EXISTS school;
```

This will drop the database only if it exists, otherwise, it will do nothing.

## 5. **Database Naming Conventions and Best Practices**

When creating databases, it's important to follow certain best practices to avoid future issues.

### 5.1 Naming Conventions

- **Use descriptive names**: The name should clearly indicate the purpose of the database, like `school`, `inventory`, or `ecommerce`.
- **Avoid reserved keywords**: Do not use MySQL reserved keywords as database names.
- **Be consistent**: Stick to a consistent naming convention (e.g., all lowercase, underscores for spaces).
- **Keep it short and simple**: Avoid unnecessarily long names.

#### Examples of Good Names:

- `school_management`
- `ecommerce_db`
- `hr_system`

#### Examples of Bad Names:

- `SchoolManagementSystem` (too long, mixed case)
- `123database` (starting with a number)
- `select` (reserved keyword)

### 5.2 Backup Considerations

Always ensure you have a backup strategy in place before dropping databases or making significant changes. Use tools like `mysqldump` or automated backups provided by your hosting service.

## 6. **Database Access and Permissions**

Controlling who can create, modify, and drop databases is crucial for security.

### 6.1 Granting Permissions

You can grant permissions for creating or dropping databases using the `GRANT` statement.

#### Example:

```sql
GRANT CREATE, DROP ON *.* TO 'username'@'localhost';
```

This grants the user `username` the ability to create and drop databases.

### 6.2 Revoking Permissions

To remove permissions, use the `REVOKE` statement.

#### Example:

```sql
REVOKE CREATE, DROP ON *.* FROM 'username'@'localhost';
```

This removes the ability to create and drop databases from the user `username`.

## 7. **Practical Example: Creating and Managing a Database**

### Step 1: Create a Database

```sql
CREATE DATABASE ecommerce
CHARACTER SET utf8mb4
COLLATE utf8mb4_unicode_ci;
```

### Step 2: Select the Database

```sql
USE ecommerce;
```

### Step 3: Create a Table in the Database

```sql
CREATE TABLE products (
    product_id INT AUTO_INCREMENT PRIMARY KEY,
    product_name VARCHAR(100),
    price DECIMAL(10, 2),
    stock INT
);
```

### Step 4: Insert Data into the Table

```sql
INSERT INTO products (product_name, price, stock)
VALUES ('Laptop', 999.99, 50),
       ('Smartphone', 499.99, 100);
```

### Step 5: Query the Data

```sql
SELECT * FROM products;
```

### Step 6: Drop the Database

```sql
DROP DATABASE IF EXISTS ecommerce;
```

Creating and displaying databases in MySQL is a foundational task that allows you to structure and manage your data efficiently. By understanding the commands, options, and best practices associated with creating, displaying, and managing databases, you can ensure that your databases are well-organized, secure, and optimized for your application's needs. Always remember to handle database operations with care, especially when it comes to modifying or dropping databases, as these actions can have significant consequences.
