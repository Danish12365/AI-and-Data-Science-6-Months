In MySQL, you have three primary ways to write comments, including one that is identical to Python. To use these in a Jupyter Notebook, you should ensure your cell is set to use a SQL kernel or use a "SQL magic" command (like `%%sql`) if you are using a Python kernel with an extension like `ipython-sql`.

Here is the content formatted for a Jupyter Notebook cell:

---

```sql
%%sql

/*
  METHOD 1: THE HASH SIGN (#)
  This is exactly like Python. 
  Everything after the # on the same line is ignored.
*/

# This is a single-line comment (Python-style)
SELECT * FROM employees; 


/*
  METHOD 2: DOUBLE DASHES (--)
  This is the standard SQL way. 
  CRITICAL: In MySQL, you MUST have a space after the second dash.
*/

-- This is a single-line comment (Standard SQL-style)
SELECT * FROM products; -- Note the space after the dashes


/*
  METHOD 3: MULTI-LINE (/* ... */)
  Used for long explanations or blocking out chunks of code.
*/

/* Author: Muhammad Danish Azeem
   Purpose: This query analyzes 
   the hotel booking dataset.
*/
SELECT hotel, arrival_date_month, lead_time 
FROM bookings 
LIMIT 10;


/*
  BONUS: THE EXPLAIN COMMAND
  If you want to 'explain' the logic of how MySQL runs a query 
  (performance analysis), use the EXPLAIN keyword.
*/

EXPLAIN SELECT * FROM bookings WHERE is_canceled = 1;

```

### Quick Comparison Table

| Feature | Python | MySQL |
| --- | --- | --- |
| **Single-line** | `# comment` | `# comment` **OR** `-- comment` |
| **Multi-line** | `''' comment '''` | `/* comment */` |
| **Analysis** | (Debugging tools) | `EXPLAIN SELECT ...` |
