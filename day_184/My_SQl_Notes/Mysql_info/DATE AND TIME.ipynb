In MySQL, handling **date and time** is essential for working with time-sensitive data such as event logs, transactions, orders, etc. MySQL provides a variety of data types, functions, and formatting options to manage date and time effectively.

### **Date and Time Data Types:**

1. **DATE**:  
   Stores a date value in `YYYY-MM-DD` format.
   - Range: `'1000-01-01'` to `'9999-12-31'`

2. **TIME**:  
   Stores a time value in `HH:MM:SS` format.
   - Range: `'-838:59:59'` to `'838:59:59'`

3. **DATETIME**:  
   Stores both date and time in `YYYY-MM-DD HH:MM:SS` format.
   - Range: `'1000-01-01 00:00:00'` to `'9999-12-31 23:59:59'`

4. **TIMESTAMP**:  
   Stores both date and time as a Unix timestamp (`YYYY-MM-DD HH:MM:SS`), which represents the number of seconds since 1970-01-01 00:00:00 UTC. It’s used for tracking changes and logs.
   - Range: `'1970-01-01 00:00:01'` to `'2038-01-19 03:14:07'`

5. **YEAR**:  
   Stores a year value in either two-digit or four-digit format.
   - Range: `'70'` to `'69'` (1970 to 2069) for two-digit; `'1901'` to `'2155'` for four-digit.

### **Useful Date and Time Functions:**

#### **1. CURDATE()**  
Returns the current date.
```sql
SELECT CURDATE();
-- Output: '2024-09-14' (depends on the current date)
```

#### **2. CURTIME()**  
Returns the current time.
```sql
SELECT CURTIME();
-- Output: '14:30:15' (depends on the current time)
```

#### **3. NOW()**  
Returns the current date and time.
```sql
SELECT NOW();
-- Output: '2024-09-14 14:30:15' (depends on the current date and time)
```

#### **4. ADDDATE()**  
Adds a specified number of days to a date.
```sql
SELECT ADDDATE('2024-09-10', INTERVAL 5 DAY);
-- Output: '2024-09-15'
```

#### **5. DATE_ADD()**  
Adds a specified interval to a date or datetime. The interval can be in days, hours, minutes, seconds, etc.
```sql
SELECT DATE_ADD('2024-09-10', INTERVAL 2 MONTH);
-- Output: '2024-11-10'
```

#### **6. SUBDATE()**  
Subtracts a specified number of days from a date.
```sql
SELECT SUBDATE('2024-09-10', INTERVAL 5 DAY);
-- Output: '2024-09-05'
```

#### **7. DATEDIFF()**  
Returns the number of days between two dates.
```sql
SELECT DATEDIFF('2024-09-15', '2024-09-10');
-- Output: 5
```

#### **8. EXTRACT()**  
Extracts parts (like year, month, day, etc.) from a date or datetime.
```sql
SELECT EXTRACT(YEAR FROM '2024-09-14');
-- Output: 2024
```

#### **9. DATE_FORMAT()**  
Formats a date or datetime value according to the given format.
```sql
SELECT DATE_FORMAT('2024-09-14 14:30:15', '%W, %M %e, %Y');
-- Output: 'Saturday, September 14, 2024'
```

#### **10. STR_TO_DATE()**  
Parses a string into a date based on the specified format.
```sql
SELECT STR_TO_DATE('14-09-2024', '%d-%m-%Y');
-- Output: '2024-09-14'
```

#### **11. TIMESTAMPDIFF()**  
Calculates the difference between two `TIMESTAMP` values.
```sql
SELECT TIMESTAMPDIFF(HOUR, '2024-09-14 10:00:00', '2024-09-14 18:00:00');
-- Output: 8
```

#### **12. UNIX_TIMESTAMP()**  
Converts a date or datetime value to a Unix timestamp.
```sql
SELECT UNIX_TIMESTAMP('2024-09-14 14:30:00');
-- Output: 1726354200
```

#### **13. FROM_UNIXTIME()**  
Converts a Unix timestamp to a `DATETIME` format.
```sql
SELECT FROM_UNIXTIME(1726354200);
-- Output: '2024-09-14 14:30:00'
```

#### **14. LAST_DAY()**  
Returns the last day of the month for a given date.
```sql
SELECT LAST_DAY('2024-09-14');
-- Output: '2024-09-30'
```

### **Example Usage of Date and Time Functions:**

#### 1. **Getting the Current Date and Time:**
```sql
SELECT NOW() AS current_datetime, CURDATE() AS current_date, CURTIME() AS current_time;
```

#### 2. **Adding and Subtracting Dates:**
```sql
SELECT 
    DATE_ADD('2024-09-10', INTERVAL 10 DAY) AS add_10_days,
    SUBDATE('2024-09-10', INTERVAL 10 DAY) AS subtract_10_days;
```

#### 3. **Calculating the Age of a Person from a Date of Birth:**
```sql
SELECT 
    FLOOR(DATEDIFF(CURDATE(), '1990-05-15') / 365) AS age;
```

#### 4. **Calculating the Days Between Two Dates:**
```sql
SELECT DATEDIFF('2024-12-31', '2024-01-01') AS days_between;
```

#### 5. **Formatting Dates:**
```sql
SELECT DATE_FORMAT(NOW(), '%W, %M %e, %Y') AS formatted_date;
```

The **`UNIX_TIMESTAMP()`** function in MySQL is used to convert a date or `DATETIME` value into a Unix timestamp. A **Unix timestamp** is the number of seconds that have elapsed since **January 1, 1970 (00:00:00 UTC)**, also known as the **Epoch**.

### **Syntax:**
```sql
UNIX_TIMESTAMP([datetime])
```

- **`datetime`**: This is the date or datetime value you want to convert into a Unix timestamp. If no argument is provided, it returns the current Unix timestamp.

### **Example 1: Converting a Specific Date to a Unix Timestamp**
```sql
SELECT UNIX_TIMESTAMP('2024-09-14 14:30:00') AS unix_time;
```
**Output:**
```
unix_time
------------
1726354200
```
- In this example, the `DATETIME` value `'2024-09-14 14:30:00'` is converted to the Unix timestamp `1726354200`.

### **Example 2: Getting the Current Unix Timestamp**
```sql
SELECT UNIX_TIMESTAMP() AS current_unix_time;
```
**Output:**
```
current_unix_time
------------------
1726355400
```
- This example returns the current Unix timestamp, representing the number of seconds since January 1, 1970, at the current moment in time.

### **Example 3: Converting a Date (Without Time) to a Unix Timestamp**
```sql
SELECT UNIX_TIMESTAMP('2024-09-14') AS unix_time;
```
**Output:**
```
unix_time
------------
1726272000
```
- In this case, only the date `'2024-09-14'` is provided. The time is assumed to be `00:00:00`, resulting in the Unix timestamp `1726272000`.

### **Example 4: Using `UNIX_TIMESTAMP()` with Time Zones**
The Unix timestamp is always in UTC (Universal Time Coordinated). Time zone differences are not considered in the calculation, so:
```sql
SELECT UNIX_TIMESTAMP('2024-09-14 14:30:00') AS unix_time_utc;
```
will return the same value regardless of the MySQL server's time zone setting.

---

### **Why Use Unix Timestamps?**
- **Precision**: Unix timestamps provide second-level precision and are commonly used for time-sensitive data.
- **Consistency**: They are independent of time zones, making them ideal for comparing times across different geographic regions.
- **Efficiency**: Working with Unix timestamps can be faster for date and time comparisons, especially when indexing.

---

### **Converting Back from Unix Timestamp to Date:**

To convert back from a Unix timestamp to a readable date or datetime format, you can use the **`FROM_UNIXTIME()`** function:

```sql
SELECT FROM_UNIXTIME(1726354200) AS readable_date;
```
**Output:**
```
readable_date
-------------------
2024-09-14 14:30:00
```
