# Querying data in MySQL

Querying data in MySQL is the process of retrieving and manipulating data from one or more tables in a database. MySQL uses the SQL (Structured Query Language) to perform various operations on the data. This can range from simple queries that retrieve data from a single table to complex queries that involve multiple tables and conditions.

Here's a comprehensive guide to querying data in MySQL:

### 1. **Basic SELECT Statement**

The `SELECT` statement is used to query data from a database. The data returned is stored in a result table, sometimes called the result set.

#### Syntax:

```sql
SELECT column1, column2, ...
FROM table_name;
```

#### Example:

```sql
SELECT first_name, last_name
FROM employees;
```

This query selects the `first_name` and `last_name` columns from the `employees` table.

### 2. **Selecting All Columns**

To select all columns from a table, you can use the `*` wildcard.

#### Example:

```sql
SELECT * 
FROM employees;
```

This query retrieves all columns from the `employees` table.

### 3. **Filtering Rows with WHERE**

The `WHERE` clause is used to filter records that meet a certain condition.

#### Syntax:

```sql
SELECT column1, column2, ...
FROM table_name
WHERE condition;
```

#### Example:

```sql
SELECT first_name, last_name
FROM employees
WHERE department = 'Sales';
```

This query retrieves the first and last names of employees who work in the Sales department.

### 4. **Using Logical Operators**

You can use logical operators like `AND`, `OR`, and `NOT` to combine multiple conditions in a `WHERE` clause.

#### Example with `AND`:

```sql
SELECT first_name, last_name
FROM employees
WHERE department = 'Sales' AND salary > 50000;
```

This query retrieves the first and last names of employees who work in Sales and have a salary greater than $50,000.

#### Example with `OR`:

```sql
SELECT first_name, last_name
FROM employees
WHERE department = 'Sales' OR department = 'Marketing';
```

This query retrieves the first and last names of employees who work in either the Sales or Marketing departments.

### 5. **Sorting Results with ORDER BY**

The `ORDER BY` clause is used to sort the result set in either ascending (`ASC`) or descending (`DESC`) order.

#### Syntax:

```sql
SELECT column1, column2, ...
FROM table_name
ORDER BY column1 [ASC|DESC];
```

#### Example:

```sql
SELECT first_name, last_name, salary
FROM employees
ORDER BY salary DESC;
```

This query retrieves the first name, last name, and salary of employees, sorted by salary in descending order.

### 6. **Limiting the Number of Rows with LIMIT**

The `LIMIT` clause is used to specify the maximum number of records to return.

#### Syntax:

```sql
SELECT column1, column2, ...
FROM table_name
LIMIT number;
```

#### Example:

```sql
SELECT first_name, last_name
FROM employees
LIMIT 5;
```

This query retrieves the first 5 rows from the `employees` table.

### 7. **Aggregating Data with GROUP BY**

The `GROUP BY` clause is used to group rows that have the same values in specified columns into summary rows. Often used with aggregate functions like `COUNT`, `SUM`, `AVG`, `MAX`, `MIN`.

#### Syntax:

```sql
SELECT column1, AGGREGATE_FUNCTION(column2)
FROM table_name
GROUP BY column1;
```

#### Example:

```sql
SELECT department, COUNT(*)
FROM employees
GROUP BY department;
```

This query counts the number of employees in each department.

### 8. **Filtering Groups with HAVING**

The `HAVING` clause is used to filter groups based on conditions, similar to the `WHERE` clause but applied after the grouping.

#### Example:

```sql
SELECT department, COUNT(*)
FROM employees
GROUP BY department
HAVING COUNT(*) > 10;
```

This query returns departments that have more than 10 employees.

### 9. **Joining Tables**

Joins are used to combine rows from two or more tables based on a related column between them.

#### Example of an INNER JOIN:

```sql
SELECT employees.first_name, employees.last_name, departments.department_name
FROM employees
INNER JOIN departments ON employees.department_id = departments.department_id;
```

This query retrieves the first and last names of employees along with their department names by joining the `employees` and `departments` tables.

### 10. **Using Aliases for Columns and Tables**

Aliases are used to give a table or a column a temporary name for the duration of a query.

#### Syntax:

```sql
SELECT column_name AS alias_name
FROM table_name AS alias_name;
```

#### Example:

```sql
SELECT e.first_name, e.last_name, d.department_name
FROM employees AS e
INNER JOIN departments AS d ON e.department_id = d.department_id;
```

This query uses aliases `e` and `d` to shorten the table names in the `JOIN`.

### 11. **Subqueries**

A subquery is a query nested inside another query. Subqueries can be used in various places like `SELECT`, `WHERE`, and `FROM` clauses.

#### Example:

```sql
SELECT first_name, last_name
FROM employees
WHERE department_id = (SELECT department_id FROM departments WHERE department_name = 'Sales');
```

This query finds the first and last names of employees who work in the Sales department by using a subquery to get the `department_id`.

### 12. **Combining Results with UNION**

The `UNION` operator is used to combine the results of two or more `SELECT` statements. Each `SELECT` statement within the `UNION` must have the same number of columns in the result sets with similar data types.

#### Syntax:

```sql
SELECT column1, column2 FROM table1
UNION
SELECT column1, column2 FROM table2;
```

#### Example:

```sql
SELECT first_name, last_name FROM employees_sales
UNION
SELECT first_name, last_name FROM employees_marketing;
```

This query combines the list of employees from both the Sales and Marketing departments into a single result set.

### 13. **Pattern Matching with LIKE**

The `LIKE` operator is used to search for a specified pattern in a column.

#### Example:

```sql
SELECT first_name, last_name
FROM employees
WHERE first_name LIKE 'J%';
```

This query retrieves employees whose first name starts with 'J'.

### 14. **Handling NULL Values**

The `IS NULL` and `IS NOT NULL` operators are used to check for `NULL` values in a column.

#### Example:

```sql
SELECT first_name, last_name
FROM employees
WHERE middle_name IS NULL;
```

This query retrieves employees who do not have a middle name.

### 15. **Complex Queries**

You can combine multiple operations to perform more complex queries.

#### Example:

```sql
SELECT d.department_name, COUNT(e.employee_id) AS num_employees, AVG(e.salary) AS avg_salary
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id
WHERE e.hire_date > '2020-01-01'
GROUP BY d.department_name
HAVING num_employees > 5
ORDER BY avg_salary DESC
LIMIT 3;
```

This query retrieves the top 3 departments with more than 5 employees hired after January 1, 2020, sorted by the average salary in descending order.

### Conclusion

Querying data in MySQL is a powerful way to interact with your database, allowing you to retrieve, filter, sort, group, and analyze your data. By mastering the various aspects of querying, you can efficiently access and manipulate the information stored in your MySQL databases.
