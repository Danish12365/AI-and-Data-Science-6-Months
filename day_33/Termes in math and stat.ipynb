

## 1. Fundamental Definitions

### **Attribute (صفت)**

1. **General:** A quality, character, or characteristic ascribed to someone or something (e.g., leadership attributes).
2. **Object:** An object closely associated with a specific person or office (e.g., a scepter is the attribute of power).
3. **Computing:** A changeable property of a program component that can be set to different values (e.g., screen size, eye color).

### **Discrete (مجرد)**

* **Meaning:** Separate or divided; a distinct unit.
* **Example:** A room is a discrete space within a house.
* **In Statistics:** Quantitative data that can be **counted** but not measured (e.g., the number of students in a class).

### **Continuous**

* **Meaning:** A variable that can assume an infinite number of real values within a given interval.
* **Example:** Height, weight, temperature, and length.
* **Note:** Continuous data often changes over time (e.g., a baby's weight).

## 3. Correlation vs. Regression

### **Correlation (ارتباط)**

* **Definition:** Measures the strength and direction of the linear relationship between two variables.
* **Example:** Height and weight (taller people tend to be heavier).
* **Types:**
1. **Positive:** Both move in the same direction.
2. **Negative:** They move in opposite directions.
3. **No Correlation:** No relationship.



### **Regression (واپسی / پسپائی)**

* **Definition:** A statistical measure that describes how an independent variable () is associated with a dependent variable ().
* **Purpose:** To predict or explain one variable based on another.

> **⭐ One-Line Summary**
> * **Correlation** tells how strongly two variables move together.
> * **Regression** tells how one variable predicts or explains another.

## 5. Variance and Standard Deviation

### **Variance ()**

* How far a set of numbers are spread out from their mean. It is the average of **squared** differences.

### **Standard Deviation ()**

* The square root of the variance. It measures dispersion in the **same units** as the original data.
* **Low :** Data is clustered tightly around the mean.
* **High :** Data is spread out over a wide range.

### **Gradient Descent**

* An optimization algorithm used to train ML models by finding parameters (weights/biases) that result in the **lowest possible error**.

### **Derivative**

* The instantaneous rate of change.
