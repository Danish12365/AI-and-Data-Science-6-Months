In 2025, **Google Antigravity** has transitioned from being a famous browser Easter egg into a serious **AI-powered Integrated Development Environment (IDE)**. Released in November 2025 alongside Gemini 3, it is designed for "Agentic Workflows"—meaning the AI doesn't just suggest code; it actually performs tasks across your editor, terminal, and browser.

To use it "like a pro" for your Data Science projects, you need to shift your mindset from being a **coder** to being an **engineering manager**.

---

## 1. How to Use Antigravity on Your PC

Antigravity is a standalone application (a fork of VS Code) that you must install locally.

* **Installation:** Download the preview from `antigravity.google`. It supports Windows, macOS, and Linux.
* **The Two-Window Setup:** * **Editor View:** This looks like VS Code. Use this for manual tweaks and seeing your code.
* **Manager View (Mission Control):** This is where the magic happens. You spawn agents here to work on different parts of your project simultaneously.


* **Permissions (Standard vs. Turbo):** * **Standard:** The AI asks before running any command.
* **Turbo:** The AI can run terminal commands (like `pip install` or `python train.py`) automatically. *Pro tip: Use "Auto" mode so the AI only asks for risky commands.*

## 2. Using Antigravity for Data Science

Antigravity is uniquely powerful for Data Science because it can handle the "friction" of environment setup and long-running experiments.

### A. Autonomous Environment Setup

Instead of manually creating virtual environments, prompt an agent:

> *"Set up a Python 3.11 environment, install the latest versions of Polars, Scikit-Learn, and XGBoost, and create a folder structure for a heart disease prediction project."*
> The agent will open the terminal, run the commands, and create your directories while you do something else.

### B. "Browser-in-the-Loop" Research

Antigravity agents can use a browser. You can tell an agent:

> *"Go to Kaggle, find the most up-to-date Heart Disease dataset, download the CSV, and write a script to show the first 5 rows and a correlation heatmap."*
> The agent will actually navigate the web, handle the download, and write the code.

---

## 3. How to Use it Like a "Pro"

To get the most out of Antigravity, stop writing small prompts and start writing **"Missions."**

* **Multi-Agent Parallelism:** While one agent is cleaning your data, spawn a second agent in the Manager View to research the latest medical papers on heart disease indicators.
* **Trust via Artifacts:** Pros don't read every line of logs. Instead, check the **Artifacts** tab. Antigravity generates summaries, screenshots of plots, and "walkthroughs" of what changed.
* **Branching Strategy:** Always tell the agent to work on a **new Git branch**. This allows you to review the "Code Diffs" in the side panel before merging its work into your main project.
* **Automated Debugging:** If your model crashes with a memory error, don't fix it yourself. Right-click the error in the terminal and select **"Delegate to Agent."** It will analyze the traceback, check your PC's RAM, and adjust the batch size in your code.

## Comparison: Pro vs. Beginner Usage

| Feature | Beginner | Pro (Agentic) |
| --- | --- | --- |
| **Workflow** | Writes code line-by-line | Defines high-level "Missions" |
| **Terminal** | Manually runs `python script.py` | Sets agent to "Auto" to run & fix errors |
| **Research** | Searches Google in a separate tab | Sends an agent to the browser to fetch data |
| **Review** | Reads the whole file | Reviews the **Implementation Plan** & **Diffs** |

### Safety Warning

Because Antigravity can execute commands on your PC, never use **"Turbo" mode** on codebases you don't trust. Always review the **Task List** generated by the agent before hitting "Approve."
