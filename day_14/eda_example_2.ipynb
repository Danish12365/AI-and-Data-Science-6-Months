# 1️⃣ Missing Values (Nulls)

### 🔍 What are Missing Values?

Missing values mean data is not available:

* Age missing
* Cabin missing
* Embarked missing

### ✅ Check Missing Values

```python
# Missing values count
df.isnull().sum()
```

### ✅ Missing Values Percentage

```python
(df.isnull().sum() / len(df)) * 100
```

### 🔥 Visual Missing Map

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.isnull(), cbar=False)
plt.title("Missing Values Heatmap")
plt.show()
```

---

## ✅ Best Ways to Handle Missing Values (Modern Methods)

---

## 1. Drop (Only if very small missing)

```python
df.dropna(subset=['Embarked'], inplace=True)
```

---

## 2. Fill with Mean / Median (Numerical)

### Median is best if skewed

```python
df['Age'].fillna(df['Age'].median(), inplace=True)
```

---

## 3. Fill with Mode (Categorical)

```python
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

---

## 4. Advanced Modern ML Method: KNN Imputation

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=5)
df[['Age', 'Fare']] = imputer.fit_transform(df[['Age', 'Fare']])
```

---

## 5. Advanced Method: Iterative Imputer (MICE)

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imp = IterativeImputer()
df[['Age', 'Fare']] = imp.fit_transform(df[['Age', 'Fare']])
```

---

# 2️⃣ Duplicate Values

### 🔍 What are duplicates?

Same row repeated twice.

### ✅ Check Duplicates

```python
df.duplicated().sum()
```

### ✅ Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

---

# 3️⃣ Outliers Detection (Very Important)

---

## 🔍 What are Outliers?

Extreme unusual values:

* Fare = 500
* Age = 80+

Outliers affect:

❌ Mean
❌ Regression
❌ ML models

---

## ✅ Outlier Detection Using Boxplot

```python
sns.boxplot(x=df['Fare'])
plt.title("Fare Outliers")
plt.show()
```

---

## ✅ IQR Method (Best Practice)

### Step 1: Calculate limits

```python
Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(lower, upper)
```

---

### Step 2: Detect Outliers

```python
outliers = df[(df['Fare'] < lower) | (df['Fare'] > upper)]
outliers
```

---

# 4️⃣ Handling Outliers (Modern Best Methods)

---

# ✅ Method 1: Remove Outliers (Only if safe)

```python
df = df[(df['Fare'] >= lower) & (df['Fare'] <= upper)]
```

---

# ✅ Method 2: Cap / Winsorize (BEST PRACTICE)

### ⭐ Most used in ML today

Instead of deleting, we limit them.

```python
df['Fare'] = df['Fare'].clip(lower, upper)
```

---

## Winsorization Using scipy

```python
from scipy.stats.mstats import winsorize

df['Fare'] = winsorize(df['Fare'], limits=[0.01, 0.01])
```

Meaning:

* Bottom 1% capped
* Top 1% capped

---

# 5️⃣ Skewness Detection (Data Distribution Shape)

---

## 🔍 What is Skewness?

| Type | Meaning                    |
| ---- | -------------------------- |
| 0    | Normal Bell Curve          |
| +ve  | Right Tail (high outliers) |
| -ve  | Left Tail                  |

---

### ✅ Check Skewness

```python
df['Fare'].skew()
```

---

### ✅ Plot Distribution

```python
sns.histplot(df['Fare'], kde=True)
plt.title("Fare Distribution")
plt.show()
```

---

# 6️⃣ Fixing Skewness (Transformation)

---

## ✅ Log Transform (Best for Right Skew)

```python
import numpy as np

df['Fare_log'] = np.log1p(df['Fare'])
```

---

## ✅ Visualize After Transform

```python
sns.histplot(df['Fare_log'], kde=True)
plt.title("Log Transformed Fare")
plt.show()
```

---

# 7️⃣ Data Distribution (Univariate Analysis)

---

## Countplot for Survival

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()
```

---

## Age Distribution

```python
sns.histplot(df['Age'], kde=True, bins=30)
plt.title("Age Distribution")
plt.show()
```

---

# 8️⃣ Relationship Analysis (Bivariate)

---

## Survival by Gender

```python
sns.barplot(x='Sex', y='Survived', data=df)
plt.title("Survival Rate by Gender")
plt.show()
```

---

## Survival by Passenger Class

```python
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival Rate by Class")
plt.show()
```

---

# 9️⃣ Correlation Heatmap

```python
num_df = df.select_dtypes(include='number')

plt.figure(figsize=(10,6))
sns.heatmap(num_df.corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```
