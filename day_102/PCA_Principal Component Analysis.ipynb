## 🔑 Key Terms in Principal Component Analysis (PCA)

### Features (Dimensions)
In the context of machine learning and PCA, **features** are the measurable attributes or characteristics of the observed data. They correspond to the columns in a dataset.
* The term **dimensions** refers to the number of features. If you have a dataset with 50 different measurements (e.g., age, income, blood pressure, etc.), your data exists in a 50-dimensional space. PCA aims to reduce this number of dimensions.

---

### Variance and Covariance
These are fundamental statistical measures used to understand the spread and relationship of data.

* **Variance ($\sigma^2$):** Measures the spread of a single feature's data points around its mean. **High variance** means the data is widely spread out, indicating more information/signal. PCA seeks to maximize the variance captured by the new axes.
* **Covariance:** Measures the degree to which two different features (variables) change together.
    * **Positive Covariance:** The two features tend to increase or decrease together.
    * **Negative Covariance:** One feature tends to increase when the other decreases.
    * **Zero/Near-Zero Covariance:** The features are largely independent.

---

### Covariance Matrix
The **Covariance Matrix** is a square matrix that summarizes the covariance between all pairs of features in your dataset.
* The elements on the **main diagonal** are the **variances** of each individual feature ($\text{Cov}(X_i, X_i)$).
* The off-diagonal elements are the **covariances** between pairs of features ($\text{Cov}(X_i, X_j)$).
* This matrix is the starting point for PCA, as it encapsulates the relationships and spread of the data, which the new Principal Components must capture.

---

### Eigenvalues and Eigenvectors
These are derived directly from the covariance matrix and are central to defining the Principal Components.

* **Eigenvector (PC Direction):** A non-zero vector that, when multiplied by a matrix (the covariance matrix), results in a vector that is parallel to the original vector. In PCA, eigenvectors represent the **directions** (axes) of maximum variance in the data. These eigenvectors become the **Principal Components**.
* **Eigenvalue (PC Magnitude):** A scalar value associated with each eigenvector. It quantifies how much of the total variance in the data is captured along its corresponding eigenvector (Principal Component).
    * The largest eigenvalue corresponds to the **First Principal Component (PC1)**, which is the axis that captures the most variance.

---

### Principal Component (PC)
A **Principal Component** is a new axis (dimension) formed as a **linear combination** of the original features.
* The PCs are selected to be orthogonal to each other and ordered by the amount of variance they explain.
* PC1 captures the maximum variance, PC2 captures the next maximum variance orthogonal to PC1, and so on.

---

### Orthogonal (Perpendicular) Axes
**Orthogonal** means the new axes (Principal Components) are strictly **perpendicular** to one another (at a $90^\circ$ angle).
* In statistical terms, this means the Principal Components are **linearly uncorrelated**. This is a major goal of PCA, as it simplifies the data structure and removes redundancy.

---

### Projection
**Projection** is the final step in PCA where the original data points are mapped (transformed) from the old feature space onto the new, reduced subspace defined by the chosen Principal Components.
* If you keep only PC1 and PC2, the projection takes a data point from a high-dimensional space (e.g., 50 dimensions) and represents it as a single point in a 2-dimensional space, defined by its coordinates on PC1 and PC2.

---

### Scree Plots
A **Scree Plot** is a visualization tool used to help determine the optimal number of Principal Components to retain. 
* **How it works:** It plots the **eigenvalues** (or the explained variance ratio) against the corresponding Principal Component number.
* **Interpretation:** Analysts look for the **"elbow"**—the point where the curve bends and starts to flatten out. Components **before** the elbow are generally retained because they contribute significantly to explaining the total variance, while components after the elbow contribute minimally and are often considered noise.

---

### Loading Score (or Loading)
The **loading score** (or just **loading**) quantifies the correlation between an **original feature** and a **Principal Component**.
* **How it works:** Loadings are the coefficients of the linear combination that defines the PC. For example, for $\text{PC1} = w_1 X_1 + w_2 X_2 + \dots$, the weights $w_i$ are the loadings.
* **Interpretation:**
    * A loading close to **+1** or **-1** means the original feature strongly influences and correlates with that PC.
    * A loading near **0** means the original feature has little influence on that PC.
* Loadings are key to **interpreting** the PCs: they tell you which original variables are most responsible for the direction and variance of each new component.

---

### BI Plot
A **BI Plot** (short for **biplot**) is a graphical display that simultaneously shows the **scores of the observations** (the projected data points) and the **loadings of the original variables** (the vectors representing the original features) on the same Principal Component axes (usually PC1 and PC2).
* **Scores (Points):** The data points show the clustering or distribution of the observations in the reduced space.
* **Loadings (Vectors):** The vectors emanating from the origin show the direction and strength of the original features.
* **Interpretation:** If an observation point lies along the direction of a vector, it means that observation has a high value for the corresponding original feature. This is the most comprehensive visualization tool in PCA.

# PCA (Principal Component Analysis) 
---

# 1) What PCA does — short intuitive view

PCA finds a new set of orthogonal axes (directions) in the feature space such that:

* The first axis (first principal component, PC1) points in the direction of **maximum variance** in the data.
* The second axis (PC2) points in the direction of the next-highest variance **subject to being orthogonal** to PC1.
* And so on.

Projecting data onto the first few PCs captures most of the “information” (variance) while reducing dimensionality.

---

# 2) The basic math (step-by-step)

Given a data matrix (X) with (n) observations and (p) features (columns):

1. **Center** the data (subtract mean of each column):
   [
   \tilde{X} = X - \mathbf{1}\mu^\top
   ]
   where (\mu) is the vector of column means.

2. (Optional) **Scale** each feature (standardize) so each column has unit variance — recommended when features have different units.

3. Compute the **covariance matrix** (if data centered):
   [
   \Sigma = \frac{1}{n-1}\tilde{X}^\top \tilde{X} \quad (p\times p)
   ]

4. Solve the **eigenvalue decomposition** of (\Sigma):
   [
   \Sigma v_i = \lambda_i v_i
   ]
   where (v_i) are eigenvectors and (\lambda_i) are eigenvalues.

5. Sort eigenvalues (\lambda_1 \ge \lambda_2 \ge \dots \ge \lambda_p) and take corresponding eigenvectors. The eigenvectors are the **principal axes** (directions). The eigenvalues tell you how much variance each axis explains.

6. **Project** the data onto the top (k) eigenvectors (components):
   [
   Z = \tilde{X} V_k
   ]
   where (V_k = [v_1, \dots, v_k]) and (Z) is the (n\times k) matrix of scores.

Alternative: PCA via **SVD** on (\tilde{X}) (numerically stable):
[
\tilde{X} = U \Sigma_{svd} W^\top
]
Principal components relate to columns of (W) and singular values.

---

# 3) Key terms — explanations & intuition

### Features (dimensions)

* The original columns of your dataset (e.g., height, age, income). (p) features = (p)-dimensional space.
* PCA reduces the number of dimensions while preserving variance.

### Orthogonal (perpendicular) axes

* Orthogonal = vectors at right angles (dot product = 0).
* PCA axes (principal components) are orthogonal by construction; this avoids redundancy and makes components uncorrelated.

### Covariance / Covariance matrix

* **Covariance** between two variables (x) and (y): (\mathrm{cov}(x,y) = \frac{1}{n-1}\sum (x_i-\bar x)(y_i-\bar y)).

  * Positive: they move together; negative: move oppositely; zero: uncorrelated.
* **Covariance matrix** (\Sigma) is a (p\times p) matrix of pairwise covariances. PCA diagonalizes this matrix — i.e., finds directions where covariance is concentrated along axes.

### Eigenvalues and Eigenvectors

* For matrix (\Sigma):

  * **Eigenvector** (v): a direction that only gets scaled (not rotated) by (\Sigma).
  * **Eigenvalue** (\lambda): amount by which (\Sigma) stretches (v).
* In PCA:

  * Eigenvectors = principal axes (directions).
  * Eigenvalues = variance explained along that axis (magnitude of importance).

### Principal Component (PC)

* A linear combination of original features: (PC_j = v_j^\top \tilde{x}).
* PC1 explains the largest variance; PC2 the second-largest, etc.
* Each PC is orthogonal to the others.

### Projection

* Mapping a high-dimensional point (\tilde{x}) onto a PC axis via dot product: (z = v^\top \tilde{x}).
* Projection reduces dimensionality and creates the “scores” used for analysis or modeling.

### Scree plot

* A plot of eigenvalues (or % variance explained) versus component index.
* Purpose: visually find the "elbow" where adding further components yields diminishing returns.
* Often used to pick (k).

### Variance and covariance (recap)

* **Variance** = covariance of a variable with itself. Measures spread.
* PCA seeks directions maximizing variance; covariance between features determines the directions of PCs.

### Loading score (loadings)

* Loadings = the entries of eigenvectors (or components_ in sklearn). They show how strongly each original feature contributes to a PC.
* For PC1 with eigenvector (v_1 = [v_{11}, v_{12}, ..., v_{1p}]), (v_{1j}) is the loading of feature (j) on PC1.
* Large magnitude of loading (positive or negative) indicates strong influence.

### Biplot

* A combined scatter plot of:

  * **Scores**: projected observations in PC1-PC2 space (points).
  * **Loadings**: feature vectors/arrows showing direction and magnitude of original variables in that PC space.
* Biplot helps interpret which original features drive separation between observations.

---

# 4) Interpreting PCA results — practical tips

* **Standardize features** if they are on different scales or units. Otherwise large-scale features dominate variance.
* **Eigenvalue magnitude**: (\lambda_j) equals variance of data projected on (v_j).
* **Explained variance ratio**: (\frac{\lambda_j}{\sum_i \lambda_i}). Use cumulative explained variance to choose (k) (e.g., 90–95%).
* **Loadings sign**: positive loading means feature and PC are positively associated; negative means inverse relation.
* **Projection distortions**: PCA is linear — it cannot capture complicated non-linear manifolds.
* **Centering** is mandatory; scaling depends on your goal.

---

# 5) How many components to keep? Common rules

* **Scree (elbow) method**: look for point where slope levels off.
* **Cumulative variance threshold**: select smallest (k) with cumulative explained variance ≥ e.g., 90% (or 70–95% depending on application).
* **Kaiser rule** (for correlation matrix): keep components with eigenvalue > 1 (interpretable when data standardized).
* **Cross-validation or downstream task performance**: choose (k) that maximizes performance for your model (recommended for ML pipelines).
* **Interpretability**: keep fewer components if interpretation is important.

---

# 6) Limitations & cautions

* PCA is **linear** — won’t capture strong non-linear structure.
* Sensitive to **outliers** — consider robust PCA or outlier handling.
* **Scaling** affects results — be deliberate about standardization.
* Loadings can be hard to interpret if features are highly correlated or numerous.
* PCA components are combinations of all features (dense), which can reduce interpretability.

# 7) PCA — quick Python example (scikit-learn) with scree plot and biplot

```python
# Example: PCA on scaled data, scree plot and biplot
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# sample dataframe df with numeric columns
# df = pd.read_csv('your_data.csv')  # assume df exists
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
X = df[numeric_cols].dropna()  # simple; handle missing properly in practice

# 1. Standardize
scaler = StandardScaler()
Xs = scaler.fit_transform(X)

# 2. Fit PCA
pca = PCA()           # compute all components
Z = pca.fit_transform(Xs)

# 3. Scree plot (explained variance)
explained = pca.explained_variance_ratio_
cum_explained = np.cumsum(explained)

plt.figure(figsize=(8,4))
plt.plot(np.arange(1, len(explained)+1), explained, 'o-', label='Explained ratio')
plt.plot(np.arange(1, len(explained)+1), cum_explained, 's--', label='Cumulative')
plt.xlabel('Principal component')
plt.ylabel('Explained variance ratio')
plt.title('Scree plot')
plt.legend()
plt.show()

# 4. Biplot (PC1 vs PC2)
def biplot(scores, loadings, labels=None):
    xs = scores[:,0]
    ys = scores[:,1]
    plt.figure(figsize=(8,8))
    plt.scatter(xs, ys, alpha=0.6)
    n = loadings.shape[0]
    scalex = 1.0 / (xs.max() - xs.min())
    scaley = 1.0 / (ys.max() - ys.min())

    for i in range(n):
        plt.arrow(0, 0, loadings[i,0]*5, loadings[i,1]*5, 
                  color='r', alpha=0.7, head_width=0.05)
        if labels is None:
            plt.text(loadings[i,0]*5*1.15, loadings[i,1]*5*1.15, 
                     "Var"+str(i+1), color='g', ha='center', va='center')
        else:
            plt.text(loadings[i,0]*5*1.15, loadings[i,1]*5*1.15, 
                     labels[i], color='g', ha='center', va='center')
    plt.xlabel('PC1')
    plt.ylabel('PC2')
    plt.grid()
    plt.title('Biplot (PC1 vs PC2)')
    plt.show()

loadings = pca.components_.T[:, :2]  # p x 2 matrix of loadings for PC1 and PC2
biplot(Z, loadings, labels=numeric_cols)
```

Notes on the biplot code:

* `pca.components_` gives principal axes (each row = PC, columns = original features). We transpose to get loadings per variable.
* The arrow length/direction shows how each original variable maps to the PC space. Scaling factors (here `*5`) are arbitrary for visualization — tweak them.

## 📊 Principal Component Analysis (PCA)

Principal Component Analysis (PCA) is a fundamental, unsupervised statistical technique used primarily for **dimensionality reduction** and **data visualization**. It works by transforming a set of potentially correlated variables into a smaller set of linearly uncorrelated variables called **Principal Components (PCs)**.

---

## 🎯 The Core Goal of PCA

The main idea is to find the directions (axes) in the data that account for the **maximum variance**.

* **Dimensionality Reduction:** Reducing the number of features (dimensions) while retaining as much of the original variability (information) as possible.
* **Noise Reduction:** By focusing on the components with high variance, PCA effectively filters out the dimensions that primarily contain random noise.

---

## ⚙️ How PCA Works (The Steps)

PCA is based on projecting the data onto new orthogonal (perpendicular) axes:

1.  **Standardization/Scaling:** The original data must be standardized (mean $\mu=0$ and standard deviation $\sigma=1$) to prevent features with larger scales from dominating the process.
2.  **Covariance Matrix Calculation:** Calculate the **covariance matrix** to understand how the variables in the input data vary with respect to each other.
3.  **Eigenvalues and Eigenvectors:** Calculate the **eigenvectors** and **eigenvalues** of the covariance matrix.
    * **Eigenvectors** define the direction and magnitude of the new axes (the **Principal Components**).
    * **Eigenvalues** represent the magnitude or amount of variance captured along those new axes.
4.  **Selecting Principal Components:** Sort the eigenvectors by their corresponding eigenvalues in descending order. The eigenvector with the highest eigenvalue is the **First Principal Component (PC1)**, which captures the most variance.
5.  **Projection:** Select the top $k$ eigenvectors (the desired number of dimensions) and transform the original data points onto this new subspace. This new dataset contains the reduced number of features, the Principal Components.

---

## 💡 Key Concepts

### Principal Components (PCs)

The Principal Components are the new features. They are linear combinations of the original features and are always **orthogonal** (at 90 degrees) to each other, meaning they are statistically **uncorrelated**.

* **PC1:** Captures the largest amount of variance in the data.
* **PC2:** Captures the second largest amount of variance, subject to being orthogonal to PC1.
* ...and so on.

### Explained Variance Ratio

This is the most critical metric for deciding how many components to keep.

* It is the proportion of the dataset's variance that each principal component accounts for.
* By summing the explained variance of the top $k$ components, you can determine how much information (variance) is retained in the reduced dataset. For example, if the first two components explain 95% of the variance, you can reduce the dimensionality from $D$ to 2 while losing only 5% of the data's variability.

---

## 🐍 Implementation and Applications

### Applications

* **Data Visualization:** Reducing complex, high-dimensional data (e.g., 50 features) to 2 or 3 Principal Components for plotting and visual analysis.
* **Preprocessing for Machine Learning:** Using PCs as input features for models like K-Means or Support Vector Machines (SVMs) to improve performance and reduce training time.
* **Noise and Redundancy Removal:** Eliminating components that explain very little variance, assuming that low-variance dimensions are mostly noise.

### Choosing the Number of Components ($k$)

Two common methods are used to determine $k$:

1.  **Scree Plot:** Plot the eigenvalues against the component number. Look for the "elbow" point where the eigenvalues level off. Components before the elbow are typically kept.
2.  **Cumulative Explained Variance Plot:** Choose the number of components that accounts for a desired percentage of total variance (e.g., $k$ components that cumulatively explain $\ge 90\%$ of the variance). 

## Singular Value Decomposition (SVD) 📐

**Singular Value Decomposition (SVD)** is a powerful and general matrix factorization technique that decomposes any rectangular matrix into three component matrices. It is one of the most widely used methods in linear algebra, with applications ranging from data compression to recommendation systems.

It is guaranteed to exist for *any* $m \times n$ matrix, unlike other decompositions (like eigenvalue decomposition) which require the matrix to be square.

---

### The SVD Formula

The decomposition of an $m \times n$ matrix $\mathbf{A}$ is given by:

$$\mathbf{A} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^{\text{T}}$$

Where:

* $\mathbf{A}$ is the original $m \times n$ matrix (e.g., a data matrix where rows are observations and columns are features).
* $\mathbf{U}$ is an $m \times m$ **orthogonal matrix** whose columns are the **left singular vectors** of $\mathbf{A}$.
* $\mathbf{\Sigma}$ (Sigma) is an $m \times n$ **diagonal matrix** containing the **singular values** of $\mathbf{A}$ on its diagonal. The singular values are always real, non-negative, and are typically arranged in descending order.
* $\mathbf{V}^{\text{T}}$ is the transpose of an $n \times n$ **orthogonal matrix** $\mathbf{V}$, whose columns are the **right singular vectors** of $\mathbf{A}$.

---

### 🔑 Key Components and Interpretation

The three resulting matrices have distinct roles in interpreting the original data matrix $\mathbf{A}$:

| Matrix | Dimensions | Interpretation in Data Science |
| :--- | :--- | :--- |
| $\mathbf{U}$ (Left Singular Vectors) | $m \times m$ | Represents the **relationships between the rows** (observations/documents/users). |
| $\mathbf{\Sigma}$ (Singular Values) | $m \times n$ | Represents the **strength or importance** of each dimension or factor. |
| $\mathbf{V}^{\text{T}}$ (Right Singular Vectors) | $n \times n$ | Represents the **relationships between the columns** (features/terms/items). |

The singular values in $\mathbf{\Sigma}$ are related to the eigenvalues of $\mathbf{A}\mathbf{A}^{\text{T}}$ and $\mathbf{A}^{\text{T}}\mathbf{A}$. Specifically, the singular values $\sigma_i$ are the square roots of the non-zero eigenvalues ($\sigma_i = \sqrt{\lambda_i}$).

### 📉 SVD for Dimensionality Reduction (Truncated SVD)

One of the most practical uses of SVD is dimensionality reduction, often called **Truncated SVD**.

Because the singular values are ordered from largest to smallest, the first few columns of $\mathbf{U}$ and $\mathbf{V}^{\text{T}}$ (corresponding to the largest singular values) capture the most variance and signal in the data.

1.  **Low-Rank Approximation:** To reduce the dimensionality from $r$ (the rank of $\mathbf{A}$) to $k$ ($k < r$), we simply keep only the **top $k$ singular values** and their corresponding vectors.
2.  **Compressed Matrix:** The resulting approximation $\mathbf{A}_k$ is calculated as:
    $$\mathbf{A}_k = \mathbf{U}_k \mathbf{\Sigma}_k \mathbf{V}_k^{\text{T}}$$
    where $\mathbf{U}_k$ is $m \times k$, $\mathbf{\Sigma}_k$ is $k \times k$, and $\mathbf{V}_k^{\text{T}}$ is $k \times n$. This yields a reduced data representation that is the **best rank-$k$ approximation** of the original matrix $\mathbf{A}$ in a least-squares sense.

### 🌐 Key Applications of SVD

* **Principal Component Analysis (PCA):** SVD is often the primary computational method used to perform PCA efficiently. The right singular vectors $\mathbf{V}$ are the same as the Principal Components (eigenvectors of $\mathbf{A}^{\text{T}}\mathbf{A}$).
* **Recommendation Systems:** Used in collaborative filtering to decompose user-item interaction matrices (matrix factorization) to discover latent features that explain user preferences.
* **Latent Semantic Analysis (LSA):** Used in Natural Language Processing (NLP) to reduce the dimensionality of term-document matrices, revealing underlying semantic concepts (topics).
* **Image Compression:** A grayscale image can be represented as a matrix. By applying Truncated SVD, a high-quality approximation of the image can be stored using much less data (fewer singular values).

# **1. What is SVD (High-Level Intuition)?**

If you have a matrix **A** (for example, a dataset: rows = samples, columns = features),

[
A_{m \times n}
]

SVD breaks it into **3 special matrices**:

[
A = U \Sigma V^T
]

### **Meaning of the matrices**

| Matrix        | Size         | Meaning                                                                             |
| ------------- | ------------ | ----------------------------------------------------------------------------------- |
| **U**         | (m \times m) | Left singular vectors → directions in **row space**                                 |
| **Σ** (Sigma) | (m \times n) | Diagonal matrix with **singular values** → how much variance each direction carries |
| **Vᵀ**        | (n \times n) | Right singular vectors → directions in **feature space**                            |

### **Intuition**

Think of SVD as telling us:

* **“What are the important directions in my data?”**
* **“How much variance does each direction carry?”**
* **“Can I reduce the number of dimensions without losing information?”**

---

# **2. Singular Values — The Heart of SVD**

Singular values are the diagonal elements of Σ:

[
\sigma_1 \ge \sigma_2 \ge \ldots \ge \sigma_p
]

Where ( p = \min(m, n) ).

### **Meaning**

* Large singular value → direction carries **strong information**
* Small singular value → direction carries **noise or redundancy**

Exactly like PCA eigenvalues.

---

# **3. Geometric Interpretation**

When SVD decomposes A, it performs 3 actions:

1️⃣ Rotate data to a new direction (using Vᵀ)
2️⃣ Stretch/squash the data along axes (using Σ)
3️⃣ Rotate again to final orientation (using U)

This explains why SVD is a generalization of PCA.

---

# **4. Mathematical Explanation**

Given matrix A:

[
A = U \Sigma V^T
]

### **U columns (left singular vectors)**

* Orthogonal
* Represent axes of **maximum variation in samples**

### **V columns (right singular vectors)**

* Orthogonal
* Represent axes of **maximum variation in features**
* Equivalent to **principal components** in PCA

### **Σ (sigma)**

Contains singular values (sorted from high → low).
The **square** of singular values equals **eigenvalues** of the covariance matrix:

[
\sigma_i^2 = \lambda_i
]

---

# **5. SVD vs PCA (Connection)**

### **PCA uses SVD under the hood**.

If you standardize your data matrix **X**, PCA is equivalent to:

[
\text{PCA components} = V
]
[
\text{Explained variance} = \sigma_i^2
]

Thus:

* **PCA = SVD applied to covariance matrix**
* **PCA directions = right singular vectors (V)**

# **6. Types of SVD**

### **1. Full SVD**

[
A = U_{m \times m}\Sigma_{m \times n} V^T_{n \times n}
]

### **2. Reduced SVD (most used in ML)**

Only keeps non-zero singular values:

[
A = U_{m \times k} \Sigma_{k \times k} V^T_{k \times n}
]

Where k = rank(A)

### **3. Truncated SVD** (Dimensionality Reduction)

Keep only top **K** singular values:

[
A \approx U_{m \times K}\Sigma_{K \times K}V^T_{K \times n}
]

This is used in:

* Topic modeling
* Recommender systems
* Image compression

The primary mathematical technique used for Principal Component Analysis (PCA) is **Singular Value Decomposition (SVD)**. However, because PCA is fundamentally about finding the **eigenvectors** (principal components) of the **covariance matrix**, any method that solves this eigenvalue problem can be used.

Here are the other common methods used in PCA, often depending on the size and nature of the data:

## 1. Eigendecomposition (Spectral Decomposition)

This is the classical, direct method for solving PCA.

* **How it works:** It involves explicitly calculating the **covariance matrix ($\mathbf{\Sigma}$)** of the centered data matrix ($\mathbf{X}$) and then finding its eigenvalues and eigenvectors.
    $$\mathbf{\Sigma} = \frac{1}{n-1} \mathbf{X}^{\text{T}} \mathbf{X}$$
    The eigenvectors of $\mathbf{\Sigma}$ are the **Principal Components**, and their corresponding eigenvalues represent the variance captured by each component.
* **Best for:** Small to medium-sized datasets where the covariance matrix can be easily computed and stored in memory.
* **Limitation:** Calculating the covariance matrix can be computationally expensive and numerically unstable for very large datasets, especially those with many features.

---

## 2. Iterative Methods (e.g., Power Iteration)

When dealing with extremely large or sparse matrices, especially in data where only the first few principal components are needed, iterative solvers are preferred over direct decomposition.

* **Power Iteration:** This method is used to find the largest eigenvalue and its corresponding eigenvector of a matrix (like the covariance matrix). By deflating the matrix, it can then find the second largest, and so on.
* **Best for:** Finding only the top $k$ principal components quickly, without computing the entire decomposition. This is common when $k$ is much smaller than the total number of features.

---

## 3. Randomized PCA

Randomized algorithms use random projections to approximate the structure of the data and find the principal components.

* **How it works:** It uses random matrices to quickly project the data into a lower-dimensional space, where the standard SVD or Eigendecomposition can be applied much faster.
* **Best for:** Datasets that are extremely large and high-dimensional, where memory and computation time are major constraints. It provides a good balance between accuracy and speed, especially for obtaining the first few dominant components.

---

## 🛠️ Summary: SVD vs. Eigendecomposition

While Eigendecomposition is the conceptual heart of PCA, **SVD is the computational workhorse** for most modern software packages (like Scikit-learn):

* **SVD bypasses the covariance matrix:** When $\mathbf{A} = \mathbf{U} \mathbf{\Sigma} \mathbf{V}^{\text{T}}$, the columns of $\mathbf{V}$ (the right singular vectors) are mathematically identical to the eigenvectors of the covariance matrix $\mathbf{A}^{\text{T}}\mathbf{A}$. Therefore, SVD calculates the PCs directly from the centered data matrix $\mathbf{A}$ without explicitly forming the potentially huge covariance matrix.
* **Numerical Stability:** SVD is generally more numerically stable and less prone to floating-point errors than direct Eigendecomposition, especially for matrices that are not full rank.

## t-SNE (t-distributed Stochastic Neighbor Embedding) 📉

t-SNE is a **non-linear, unsupervised dimensionality reduction technique** widely used for **visualization**. Its primary goal is to take high-dimensional data and map it to a low-dimensional space (typically 2 or 3 dimensions) such that the **local structure** of the data is preserved.

Unlike methods like PCA, which focus on maximizing global variance, t-SNE prioritizes keeping similar data points close together and dissimilar points far apart in the low-dimensional map.

---

## ⚙️ How t-SNE Works

t-SNE's mechanism is based on converting high-dimensional Euclidean distances between data points into **conditional probabilities** that represent similarities.

### 1. High-Dimensional Similarity (Gaussian Distribution)

In the high-dimensional space, the similarity between two data points, $x_i$ and $x_j$, is defined by a **Gaussian (Normal) distribution**. The conditional probability $p_{j|i}$ (the probability that $x_i$ would choose $x_j$ as its neighbor) is calculated as:

$$p_{j|i} = \frac{\exp(-\|x_i - x_j\|^2 / 2\sigma_i^2)}{\sum_{k \neq i} \exp(-\|x_i - x_k\|^2 / 2\sigma_i^2)}$$

* The parameter $\sigma_i$ (standard deviation) is set individually for each point $x_i$ such that the **perplexity** (discussed below) is satisfied.
* The joint probability, $P_{ij}$, is then calculated as the average of the conditional probabilities: $P_{ij} = (p_{j|i} + p_{i|j}) / 2n$.

### 2. Low-Dimensional Similarity (Student's t-distribution)

In the low-dimensional target space (the map), the similarity between the corresponding points, $y_i$ and $y_j$, is defined using the **t-distribution with 1 degree of freedom (which is Cauchy distribution)**:

$$q_{ij} = \frac{(1 + \|y_i - y_j\|^2)^{-1}}{\sum_{k \neq l} (1 + \|y_k - y_l\|^2)^{-1}}$$

* Using a heavy-tailed t-distribution helps address the "**crowding problem**" (where projected data points become squished together in the center), which often plagues methods that use Gaussian distributions for both spaces.

### 3. Minimizing Divergence (Kullback-Leibler)

The algorithm then uses **gradient descent** to move the points $y_i$ in the low-dimensional space to minimize the difference between the high-dimensional joint probabilities ($P_{ij}$) and the low-dimensional joint probabilities ($Q_{ij}$).

This difference is measured using the **Kullback-Leibler (KL) Divergence**:

$$\text{KL}(P \| Q) = \sum_{i} \sum_{j} P_{ij} \log \left( \frac{P_{ij}}{Q_{ij}} \right)$$

Minimizing the KL divergence ensures that the map $Q$ is a good representation of the original data structure $P$.

---

## 🔑 Key Parameter: Perplexity

The **Perplexity** is arguably the most important hyperparameter in t-SNE.

* **Definition:** Perplexity is a smooth measure of the effective number of neighbors surrounding a data point. It is mathematically related to the local Gaussian variance ($\sigma_i$).
* **Role:** It determines the balance between focusing on **local structure** (small $\sigma_i$) and **global structure** (large $\sigma_i$).
* **Rule of Thumb:** Perplexity values are typically between **5 and 50**.
    * **Low Perplexity:** Focuses too much on local density, potentially leading to noise and fragmented clusters.
    * **High Perplexity:** Focuses too much on global structure, potentially merging genuinely separate clusters into a single blob.

---

## ⚠️ Interpretation and Limitations

* **Cluster Size/Density:** The *distances* between clusters on the t-SNE map are **not meaningful**. t-SNE focuses on local structure, so the area or density of a cluster in the plot does not reflect its size or density in the original high-dimensional space.
* **Distance between Clusters:** The size of the gap between two clusters is **not interpretable**. It only suggests that the two groups are well-separated.
* **Computational Cost:** t-SNE is computationally expensive, especially for very large datasets (millions of points), which often necessitates using approximation techniques or running PCA first to reduce the input dimensions.
* **Non-deterministic:** Due to the random initialization and non-convex nature of the optimization, different runs can produce slightly different, though topologically similar, visualizations.

t-SNE:

1. Finds for each point **which points are close neighbors**

3. Then tries to recreate that neighborhood structure in 2D or 3D

# 🔥 **Why t-SNE Was Invented?**

PCA fails when data is **non-linear**, complex, curved, or clustered.
t-SNE solves this by:

* using probability distributions
* using a heavy-tailed t-distribution in low-dimensional space
* preserving **local distances** instead of global variance

---

# 🧠 **How t-SNE Works (Step-by-Step Deep Explanation)**

---

# **STEP 1 — Compute Pairwise Distances**

For every pair of points (x_i) and (x_j):

[
d_{ij} = |x_i - x_j|^2
]

This is in the **high-dimensional space**.

---

# **STEP 2 — Convert Distances to Conditional Probabilities**

t-SNE wants to know:

👉 “What is the probability that point j is a neighbor of point i?”

Using Gaussian distribution:

[
p_{j|i} = \frac{\exp(-d_{ij}/2\sigma_i^2)}{\sum_{k \ne i} \exp(-d_{ik}/2\sigma_i^2)}
]

where:

* (\sigma_i) = variance of Gaussian (changes per point)
* chosen via **perplexity**

---

# ✔ **Perplexity — the Most Important Hyperparameter**

Perplexity controls the number of effective neighbors.

[
Perplexity = 2^{H(P_i)}
]

Common values:

* 5 → very local clusters
* 30 → standard
* 50–100 → large-scale structure

---

# **STEP 3 — Make Symmetric Probabilities**

[
p_{ij} = \frac{p_{i|j} + p_{j|i}}{2n}
]

This creates a joint probability distribution **in high-dimensional space**.

---

# **STEP 4 — Create Low-Dimensional Points (2D or 3D)**

Randomly initialize points:

[
y_1, y_2, \ldots, y_n
]

These are the 2D/3D projected points.

---

# **STEP 5 — Compute Low-Dimensional Probabilities**

Except now we use a **Student’s t-distribution with 1 degree of freedom**:

[
q_{ij} = \frac{(1 + |y_i - y_j|^2)^{-1}}{\sum_{k \ne l} (1 + |y_k - y_l|^2)^{-1}}
]

Why t-distribution?

* Heavy tails
* Prevents all points collapsing
* Separates clusters better

---

# **STEP 6 — Minimize KL Divergence**

We want ( q_{ij} ) to match ( p_{ij} ):

[
KL(P \parallel Q) = \sum_i \sum_j p_{ij} \log \frac{p_{ij}}{q_{ij}}
]

This is minimized using:

* Gradient descent
* Momentum
* Learning rate

---

# 🧩 **What t-SNE Is Optimizing?**

It wants:

* **close points to stay close**
* **far points to move far apart**

A perfect technique to visualize high-dimensional clusters.

---

# 🌈 **Final Visualization**

t-SNE outputs 2D or 3D clusters that reflect the real neighborhood relationships.

In machine learning, **divergence** is a measure of how different one probability distribution is from another. The most common type, especially in t-SNE, is the **Kullback-Leibler (KL) Divergence**, also known as relative entropy.

Think of it as a **"penalty score"** for using an approximation. If you have a "true" probability distribution, $P$, and an approximation of it, $Q$, the KL divergence measures the amount of information you lose when you use $Q$ to represent $P$.

* A divergence of **0** means the two distributions are identical.
* A **higher** divergence means the approximation is worse, and more information is lost.

---

## 🏛️ The Relationship with Perplexity in t-SNE

The relationship is direct and fundamental: **Perplexity** is a key hyperparameter you set to help *define* the high-dimensional probability distribution ($P$) that the **Divergence** then tries to match.

Here’s the step-by-step connection:

1.  **Goal of t-SNE:** The algorithm's main goal is to create a low-dimensional map ($Q$) that is as similar as possible to the high-dimensional data ($P$).
2.  **Measuring Similarity:** t-SNE measures the "similarity" of data points as probabilities.
    * **$P$ (High-Dim):** The probability that point $x_i$ would pick $x_j$ as its neighbor.
    * **$Q$ (Low-Dim):** The probability that point $y_i$ would pick $y_j$ as its neighbor.
3.  **Role of Perplexity:** Perplexity is a value you set (e.g., 30) that defines the "effective number of neighbors" for each point. The algorithm uses this value to calibrate the variance ($\sigma_i$) of the Gaussian distribution used to calculate the high-dimensional probabilities ($P$).
    * **Low Perplexity (e.g., 5):** The model focuses only on the *very* closest neighbors, emphasizing local structure.
    * **High Perplexity (e.g., 50):** The model "looks" further out and considers more neighbors, trying to preserve more of the global structure.
4.  **Role of Divergence:** The KL divergence is the **cost function** t-SNE tries to minimize. It calculates the "penalty" for the mismatch between the high-dim probabilities (the $P$ you defined with perplexity) and the low-dim probabilities ($Q$). 

In short: **Perplexity $\rightarrow$ helps create $P$ (the target) $\rightarrow$ KL Divergence $\rightarrow$ measures how different $Q$ (the map) is from $P$.**

The algorithm iteratively adjusts the map ($Q$) to minimize this divergence, pulling points together or pushing them apart until the low-dimensional relationships best reflect the high-dimensional relationships you defined with your perplexity setting.

# ✅ **What is KL Divergence? (Simple + Deep Explanation)**

KL Divergence measures how different two probability distributions are.

In t-SNE, we compare:

* **High-dimensional probability distribution** → (P_{ij})
* **Low-dimensional probability distribution** → (Q_{ij})

KL divergence is:

[
KL(P \parallel Q) = \sum_{i} \sum_{j} p_{ij} \log \frac{p_{ij}}{q_{ij}}
]

### ✔ If (P) and (Q) are similar → KL divergence is LOW

### ✔ If (P) and (Q) are different → KL divergence is HIGH

t-SNE tries to **minimize** KL divergence using gradient descent.

---

# 🧠 **Why does t-SNE use KL Divergence?**

KL divergence helps t-SNE keep:

* **close points** → close
* **far points** → far

Because KL heavily penalizes situations where:

* a pair is close in high-dim space,
* but far away in low-dim space.

That’s why t-SNE preserves **local structure** extremely well.

But the variance (\sigma_i) of each point must be chosen.

t-SNE chooses (\sigma_i) such that:

[

]

Where (H(P_i)) is entropy of distribution around point (i).
