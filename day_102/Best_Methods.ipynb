# 1️⃣ **K-Means Clustering**

### ✔ Where we use it in ML?

* Customer segmentation
* Market analysis
* Image compression
* Grouping similar users
* Fraud pattern grouping

### ✔ Where in Preprocessing?

* To find **initial clusters** for hierarchical or GMM
* To create **cluster_labels** as features for ML models

### ✔ Example:

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5)
df['cluster'] = kmeans.fit_predict(df[['Age','Annual Income','Spending Score']])
```

---

# 2️⃣ **Hierarchical Clustering**

### ✔ Where we use it?

* When you want **dendrograms**
* When number of clusters is unknown
* Medical data clustering (gene expression)

### ✔ Preprocessing?

* Used to decide **optimal number of clusters** before using K-means

### ✔ Example:

```python
import scipy.cluster.hierarchy as sch
sch.dendrogram(sch.linkage(df[['Age','Annual Income','Spending Score']], method='ward'))
```

---

# 3️⃣ **DBSCAN**

### ✔ Where used?

* Density-based clustering
* Good for arbitrarily shaped clusters
* Finds outliers automatically

### ✔ Preprocessing?

* Removing noise
* Removing outliers before training ML models

### ✔ Example:

```python
from sklearn.cluster import DBSCAN
db = DBSCAN(eps=5, min_samples=5).fit(df[['Annual Income','Spending Score']])
```

---

# 4️⃣ **OPTICS**

### ✔ Improved DBSCAN

* Works with **varying density**
* No need for a single eps
* Produces reachability plot

### ✔ Use cases:

* Geospatial patterns
* Network traffic clustering

### ✔ Example:

```python
from sklearn.cluster import OPTICS
opt = OPTICS(min_samples=10).fit(df[['Annual Income','Spending Score']])
```

---

# 5️⃣ **HDBSCAN (Best Density Clustering)**

### ✔ Why best?

* No need to choose eps
* Finds clusters + noise
* Works on large data
* Handles variable density better than DBSCAN/OPTICS

### ✔ Use cases:

* NLP embeddings (BERT vector clustering)
* Medical imaging
* Fraud detection

### ✔ Example:

```python
import hdbscan
clusterer = hdbscan.HDBSCAN(min_cluster_size=20)
df['cluster'] = clusterer.fit_predict(df[['Annual Income','Spending Score']])
```

---

# 6️⃣ **BIRCH**

### ✔ Designed for massive datasets (>1M rows)

* Uses CF trees
* Incremental clustering

### ✔ Use cases:

* Real-time clustering
* Big e-commerce datasets

### ✔ Example:

```python
from sklearn.cluster import Birch
birch = Birch(n_clusters=5)
df['cluster'] = birch.fit_predict(df[['Age','Income']])
```

---

# 7️⃣ **CLARA**

### ✔ K-means on samples

* For huge datasets
* Memory-efficient

* Clustering millions of customer records

*(Available in R; not in scikit-learn)*

# 8️⃣ **GMM (Gaussian Mixture Models)**

### ✔ Soft clustering

* Gives probabilities
* Assumes data is Gaussian mixture

### ✔ Use cases:

* Customer segmentation with uncertainty
* Speech processing (GMM-HMM)
* Financial modeling

### ✔ Example:

```python
from sklearn.mixture import GaussianMixture
gmm = GaussianMixture(n_components=4)
df['cluster'] = gmm.fit_predict(df[['Age','Income']])
```
