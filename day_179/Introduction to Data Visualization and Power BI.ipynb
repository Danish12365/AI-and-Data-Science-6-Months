# 🔷 1. Introduction to Data Visualization

👉 **Data Visualization = turning raw data into charts, graphs, dashboards**

---

## 🧠 Why It Matters

* Humans understand visuals faster than numbers
* Helps in decision-making
* Reveals patterns, trends, and outliers

---

## 📊 Examples

* Sales trend → Line chart
* Category comparison → Bar chart
* Geographic data → Map

---

## 🔥 Simple Definition

> Data Visualization = “Data ko visually present karna for better understanding”

---

# 🔷 2. What is Power BI?

👉 **Power BI** is a tool used to:

* Analyze data
* Create dashboards
* Share insights

---

## 🧩 Components

* Power BI Desktop → Build reports
* Power BI Service → Share online
* Power BI Mobile → View dashboards

---

## 🔥 Key Features

* Drag & Drop interface
* Strong data modeling
* DAX (Data Analysis Expressions)
* Real-time dashboards

---
