Git and GitHub are **must-have skills for Data Scientists**—not just for coding, but for **project management, collaboration, and portfolio building**.


---

# 🔷 1. What is Git?

👉 **Git** = Version Control System

It helps you:

* Track changes in code
* Go back to previous versions
* Work safely without losing work

📌 Think:

> Git = “History + Backup + Control”

---

# 🔷 2. What is GitHub?

👉 **GitHub** = Cloud platform for Git repositories

* Store your code online
* Share projects
* Collaborate with others
* Build your **portfolio**

📌 Think:

> GitHub = “Google Drive for Code + Collaboration”

# 🔷 4. Basic Git Workflow (MOST IMPORTANT)

This is the **core cycle**:

```
Working Directory → Staging Area → Repository
```

### Commands:

```bash
git init        # Start repo
git add .       # Add files
git commit -m "message"   # Save snapshot
```

---

# 🔷 5. Connect with GitHub

## Step 1: Create Repo on GitHub

Go to 👉 GitHub
Create new repository

---

## Step 2: Link Local to GitHub

```bash
git remote add origin https://github.com/username/repo.git
git branch -M main
git push -u origin main
```

# 🔷 6. Important Commands for Data Scientists

## 📌 Check status

```bash
git status
```

---

## 📌 See changes

```bash
git diff
```

---

## 📌 View history

```bash
git log
```

---

## 📌 Undo changes

```bash
git checkout file_name
```

---

## 📌 Remove file

```bash
git rm file_name
```

# 🔷 7. Branching (VERY IMPORTANT)

👉 Branch = separate version of project

### Why?

* Try new model safely
* Test features

```bash
git branch new-feature
git checkout new-feature
```

Or:

```bash
git checkout -b new-feature
```

---

## 🔀 Merge Branch

```bash
git checkout main
git merge new-feature
```

# 🔷 8. GitHub for Data Science Projects

## 📁 Typical Project Structure

```
project/
│
├── data/
├── notebooks/
├── src/
├── models/
├── README.md
├── requirements.txt
```

---

## 📌 Important Files

### 1. README.md (VERY IMPORTANT)

Explain:

* Problem
* Dataset
* Approach
* Results

---

### 2. requirements.txt

```bash
pip freeze > requirements.txt
```

---

### 3. .gitignore

Ignore:

```
venv/
__pycache__/
*.csv
*.pkl
```
