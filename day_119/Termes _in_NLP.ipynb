# 1. Tokenization

## What is it?

Tokenization means breaking text into smaller units called **tokens**.

Example:

```text
I love Data Science
```

After tokenization:

```python
["I", "love", "Data", "Science"]
```

These tokens may be:

* Words
* Sentences
* Characters
* Subwords

---

## Types

### Word Tokenization

```text
I love AI
```

↓

```python
["I","love","AI"]
```

---

### Sentence Tokenization

```text
I love AI. It is amazing.
```

↓

```python
[
"I love AI.",
"It is amazing."
]
```

---

### Character Tokenization

```text
AI
```

↓

```python
["A","I"]
```

---

### Subword Tokenization (Used by BERT & GPT)

```text
unbelievable
```

↓

```python
["un","believ","able"]
```

Why?

Because models cannot memorize every word.

---

# 2. Stemming

## What is it?

Stemming reduces words to their root form.

Example:

```text
playing
played
plays
```

↓

```text
play
```

---

### Examples

```text
studies
```

↓

```text
studi
```

Notice:

❌ Not always a real word.

---

### Popular Stemmer

* Porter Stemmer
* Snowball Stemmer

---

# 3. Lemmatization

## What is it?

Lemmatization converts a word into its dictionary form (lemma).

Example:

```text
better
```

↓

```text
good
```

---

### Examples

```text
running
```

↓

```text
run
```

```text
studies
```

↓

```text
study
```

Unlike stemming:

✔ Produces valid words

---

## Stemming vs Lemmatization

| Stemming              | Lemmatization       |
| --------------------- | ------------------- |
| Fast                  | Slower              |
| Rule based            | Dictionary based    |
| May create fake words | Produces real words |
| Less accurate         | More accurate       |

---

# 5. Bag of Words (BoW)

## Idea

Convert text into numbers.

Machine learning models understand numbers, not words.

---

Sentences:

```text
I love AI
I love Python
```

Vocabulary:

```text
[I, love, AI, Python]
```

---

Vector Representation:

| Word   | S1 | S2 |
| ------ | -- | -- |
| I      | 1  | 1  |
| love   | 1  | 1  |
| AI     | 1  | 0  |
| Python | 0  | 1  |

---

Problem:

BoW ignores:

* Word order
* Meaning
* Context

---

# 6. TF-IDF

## Full Form

Term Frequency – Inverse Document Frequency

---

## Why Needed?

BoW treats all words equally.

TF-IDF gives importance to rare words.

---

### Term Frequency (TF)

How often a word appears.

```text
AI AI AI Data
```

TF(AI) = 3

---

### Inverse Document Frequency (IDF)

How unique a word is.

Word:

```text
the
```

Appears everywhere.

Low importance.

Word:

```text
Transformer
```

Appears rarely.

High importance.

---

### Formula

[
TF-IDF = TF \times IDF
]

---

# 7. Named Entity Recognition (NER)

## What is NER?

Finding important entities in text.

---

Sentence:

```text
Muhammad Danish works at KE in Karachi.
```

NER identifies:

| Word            | Entity       |
| --------------- | ------------ |
| Muhammad Danish | Person       |
| KE              | Organization |
| Karachi         | Location     |

---

Entity Types

* Person
* Organization
* Country
* City
* Date
* Money

---

# 8. POS (Parts of Speech Tagging)

---

| Word     | Tag        |
| -------- | ---------- |
| The      | Determiner |

| is       | Verb       |

---

Common POS Tags

* Noun
* Verb
* Adjective
* Adverb
* Pronoun
* Preposition

# 12. Word Embeddings

## Problem with BoW

BoW doesn't understand meaning.

---

Example:

```text
King
Queen
Man
Woman
```

BoW sees them as unrelated.

---

Word Embeddings convert words into vectors.

Example:

```text
King = [0.5, 0.8, ...]
Queen = [0.6, 0.9, ...]
```

Similar words have similar vectors.

---

Popular Embeddings

### Word2Vec

Created by Tomas Mikolov

---

### GloVe

Created by Stanford University

---

### FastText

Created by Meta AI

→ right context
