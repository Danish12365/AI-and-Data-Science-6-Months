Probability Theory is the mathematical framework for quantifying uncertainty. It allows us to transition from "guessing" to making "calculated risks." As a Data Scientist, you use this every day—whether you're calculating the p-value of a test or determining the likelihood of a customer churning.

---

## 1. The Three Approaches to Probability

Before diving into the math, you must understand how we "assign" probability:

* **Theoretical (Classical):** Based on physical symmetry. If you toss a fair coin, the probability of heads is $1/2$ because there are 2 equal outcomes.
* **Empirical (Frequentist):** Based on data and observation. If you toss a coin 1,000 times and it lands on heads 520 times, the probability is $0.52$. This is how most Machine Learning models work.
* **Subjective (Bayesian):** Based on personal belief updated by evidence. "I am 80% sure it will rain because the clouds are dark."

---

## 2. Key Terminology

To speak the language of probability, you need these four terms:

1.  **Experiment:** An action with an uncertain result (e.g., tossing a die).
2.  **Outcome:** A single possible result (e.g., rolling a 4).
3.  **Sample Space ($S$):** The set of all possible outcomes. For a die, $S = \{1, 2, 3, 4, 5, 6\}$.
4.  **Event ($E$):** A specific subset of outcomes we care about (e.g., "Rolling an even number").

The fundamental formula is:
$$P(E) = \frac{\text{Number of Favorable Outcomes}}{\text{Total Number of Outcomes in } S}$$

---

## 3. The Rules of Probability (Axioms)

For any event $A$ and $B$, these rules must always hold true:

### A. The Range
Probability is always between 0 and 1 (or 0% and 100%).
$$0 \leq P(A) \leq 1$$

### B. The Complement Rule
The probability of something **not** happening is $1$ minus the probability of it happening.
$$P(A^c) = 1 - P(A)$$

### C. The Addition Rule (OR Logic)
Used when you want to find the probability of $A$ **OR** $B$ occurring.
* **If Mutually Exclusive (cannot happen together):** $P(A \cup B) = P(A) + P(B)$
* **If Not Mutually Exclusive:** $P(A \cup B) = P(A) + P(B) - P(A \cap B)$
    *(We subtract the intersection so we don't count it twice!)*



### D. The Multiplication Rule (AND Logic)
Used when you want to find the probability of $A$ **AND** $B$ occurring.
* **If Independent:** $P(A \cap B) = P(A) \times P(B)$
* **If Dependent:** $P(A \cap B) = P(A) \times P(B|A)$

---

## 4. Conditional Probability & Independence

This is the most important concept for your **Naive Bayes** and **Bayesian Optimization** work.

**Conditional Probability** is the probability of $A$ occurring **given** that $B$ has already occurred. It is written as $P(A|B)$.
$$P(A|B) = \frac{P(A \cap B)}{P(B)}$$

* **Independent Events:** Knowing $B$ happened tells you nothing about $A$. (e.g., Tossing a coin twice).
* **Dependent Events:** Knowing $B$ happened changes the probability of $A$. (e.g., Drawing a card from a deck and not putting it back).

## 5. Probability Distributions

In Data Science, we don't just look at single events; we look at the behavior of variables.

| Distribution | Type | Use Case |
| :--- | :--- | :--- |
| **Normal (Gaussian)** | Continuous | Height, IQ, Model Errors (Residuals). |
| **Binomial** | Discrete | Yes/No outcomes (e.g., Will a user click an ad?). |
| **Poisson** | Discrete | Frequency of events over time (e.g., Customers per hour). |
| **Bernoulli** | Discrete | A single trial with two outcomes (Success/Failure). |



---

## 6. Law of Large Numbers (LLN)
This law states that as you repeat an experiment more times, the **average** of the results will get closer and closer to the **expected value**.
* **Example:** If you flip a coin 10 times, you might get 7 heads (70%). But if you flip it 10,000 times, you will almost certainly be very close to 5,000 heads (50%).

### 1. Probability Mass Function (PMF)
**The Concept:** PMF is for **Discrete** data (things you can count: 1, 2, 3...). 
It gives you the **exact probability** of a specific outcome.

* **Analogy:** Think of a **Bar Chart** of a dice roll.
* **The Math:** If you ask, "What is the probability of rolling exactly a 4?", the PMF gives you the height of that specific bar (which is $1/6$).
* **Key Rule:** If you add up all the "masses" (the heights of all bars), they must equal **1**.



---

### 2. Probability Density Function (PDF)
**The Concept:** PDF is for **Continuous** data (things you measure: 1.52, 1.521, 1.5211...).
Because there are infinite decimals between 1 and 2, the probability of an "exact" number (like exactly 1.750000... meters tall) is technically **0**.

* **How it works:** Instead of an exact point, we look at the **Area Under the Curve** between two points.
* **Analogy:** Think of a **Hill** (The Bell Curve). The "Density" is the height of the hill at a certain spot, but the "Probability" is the amount of dirt (area) under the hill between two markers.
* **The Math:** To find the probability that a movie's IMDB rating is between 7.0 and 8.0, you calculate the area under the PDF curve between those two values.



---

### 3. Cumulative Distribution Function (CDF)
**The Concept:** The CDF is "The Accumulator." It works for **both** discrete and continuous data.
It doesn't ask "What is the chance of X?", it asks **"What is the chance of X or anything smaller?"**

* **Analogy:** Think of a **Staircase** (for discrete) or a **Rising Slope** (for continuous). It always starts at 0 and ends at 1.
* **The Math:** If you want to know the probability that a bank customer has *at most* 3 complaints, you look at the CDF value at 3. It adds up the probabilities of 0, 1, 2, and 3 complaints.
* **Why use it?** It is perfect for finding **Percentiles** (e.g., "90% of movies have a rating below this value").



---

### 4. Empirical Distribution Function (EDF / ECDF)
**The Concept:** The EDF is the **Real-World version** of the CDF. 
While a CDF is a perfect mathematical formula (Theory), the EDF is built from your **Actual Data** (Practice).

* **How it works:** You take your data points (e.g., ages of 100 people), sort them from smallest to largest, and create a step-graph. Each data point "steps up" the probability by $1/N$.
* **Analogy:** If you have 10 data points, each point represents a 10% step. If you plot these steps, you get an Empirical (observed) version of the theoretical CDF.
* **Use Case:** We use this to check if our real data "fits" a theoretical distribution (like checking if your Bank Churn data is normally distributed).
