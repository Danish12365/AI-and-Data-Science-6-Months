# Data Types & Categories for Analysis

## I. Fundamental Data Types

### 1. **Data Types by Structure**

| Type | Description | Examples | Common Use |
|------|-------------|----------|------------|
| **Structured** | Highly organized, predefined format, fits in tables | Databases, CSV files, Excel spreadsheets | Traditional analytics, SQL queries |
| **Semi-structured** | Some organization but flexible schema | JSON, XML, YAML, NoSQL databases | Web APIs, log files |
| **Unstructured** | No predefined format or organization | Text, images, videos, audio, social media posts | NLP, computer vision, sentiment analysis |

### 2. **Data Types by Measurement Scale**

| Type | Description | Mathematical Operations | Examples |
|------|-------------|------------------------|----------|
| **Nominal** | Categories without order | =, ≠ | Colors (Red, Blue), Gender, Country |
| **Ordinal** | Categories with order but unknown intervals | =, ≠, <, > | Education level, Customer satisfaction ratings |
| **Interval** | Numeric with equal intervals but no true zero | =, ≠, <, >, +, - | Temperature (°C), Calendar dates, IQ scores |
| **Ratio** | Numeric with true zero point | All operations (+, -, ×, ÷) | Height, Weight, Income, Age |

### 3. **Data Types by Source**

| Type | Examples |
|------|----------|
| **Internal** | CRM data, Sales records, Employee data |
| **External** | Market research, Government data, Social media |
| **Primary** | Data collected firsthand for specific research |
| **Secondary** | Data collected by others, reused for new analysis |
| **Real-time** | Sensor data, Stock prices, Live traffic |
| **Historical** | Archived data, Past sales, Historical records |

## II. Types of Data Analysis

### The Analytics Hierarchy (From Basic to Advanced)

```mermaid
graph TD
    A[What happened?<br>Descriptive Analytics] --> B[Why did it happen?<br>Diagnostic Analytics]
    B --> C[What will happen?<br>Predictive Analytics]
    C --> D[What should we do?<br>Prescriptive Analytics]
```

### 1. **Descriptive Analytics**
**"What happened?"**

- **Purpose:** Summarize historical data to understand past performance
- **Key Techniques:**
  - Data aggregation
  - Summary statistics (mean, median, mode, standard deviation)
  - Data visualization (charts, graphs, dashboards)
  - Reporting and KPIs
- **Examples:**
  - Monthly sales reports
  - Website traffic analysis
  - Social media engagement metrics
  - Customer demographics summary
- **Tools:** Excel, Tableau, Power BI, SQL

### 2. **Diagnostic Analytics**
**"Why did it happen?"**

- **Purpose:** Identify causes and relationships behind observed outcomes
- **Key Techniques:**
  - Correlation analysis
  - Regression analysis
  - Root cause analysis
  - Drill-down analysis
  - Data mining
- **Examples:**
  - Why did sales drop in Q3?
  - What factors affect customer churn?
  - Which marketing channel drove the most conversions?
  - What caused the spike in website bounce rate?
- **Tools:** R, Python (Pandas, Statsmodels), SAS, SPSS

Both SAS and SPSS are industry-leading software platforms used for statistical analysis and data management. 
Full Forms

SAS: Statistical Analysis System.
SPSS: Statistical Package for the Social Sciences. 

Key Differences
Developer: SAS was developed by the SAS Institute. SPSS was originally created by SPSS Inc. and is now owned and developed by IBM.

Primary Use Case: SAS is widely used in healthcare, finance, and enterprise settings for advanced analytics and handling very large datasets. SPSS is the standard in academic and social science research due to its intuitive, user-friendly interface.

Interface: SAS is primarily command-based (programming-heavy), whereas SPSS is known for its point-and-click graphical user interface (GUI).

### 3. **Predictive Analytics**
**"What will happen?"**

- **Purpose:** Forecast future outcomes based on historical patterns
- **Key Techniques:**
  - Machine learning algorithms
  - Time series forecasting
  - Classification models
  - Regression models
  - Neural networks
- **Examples:**
  - Sales forecasting for next quarter
  - Customer churn prediction
  - Credit risk scoring
  - Equipment failure prediction
  - Demand forecasting
- **Tools:** Python (Scikit-learn, TensorFlow), R, SAS, MATLAB

### 4. **Prescriptive Analytics**
**"What should we do?"**

- **Purpose:** Recommend optimal actions to achieve desired outcomes
- **Key Techniques:**
  - Optimization algorithms
  - Simulation modeling
  - Decision analysis
  - Recommendation systems
  - A/B testing frameworks
- **Examples:**
  - Optimal pricing strategy
  - Inventory optimization
  - Route optimization for delivery
  - Personalized product recommendations
  - Resource allocation planning
- **Tools:** Python (PuLP, Gurobi), IBM Decision Optimization, AnyLogic


hese tools represent the gold standard for prescriptive analytics, but they serve different needs ranging from open-source coding to heavy-duty enterprise simulation. 

1. Python Ecosystem: PuLP & Gurobi
Python is the most common language for building custom prescriptive models because it bridges the gap between data science and mathematical optimization.

PuLP (The Modeler): An open-source Python library used to formulate linear programming problems. Think of it as the "translator" that turns your business logic into math. It doesn't solve the problem itself; it passes it to a "solver" like CBC (included) or Gurobi.

Gurobi (The Solver): Widely considered the world's fastest mathematical solver. While it has its own Python API (gurobipy), it is a commercial product. It is designed to handle millions of variables and complex constraints (like non-linear or integer programming) that would crash free tools. 

2. IBM Decision Optimization (The Enterprise Platform)
This is a full-stack suite powered by the famous CPLEX engine. It is built for teams that need to deploy optimization into production software rather than just running a script on a laptop. 
Modeling Assistant: Includes a "Natural Language" tool that lets non-coders build optimization models using simple English.

Scenario Management: Allows users to run "What-If" scenarios side-by-side to see how changing a single constraint (e.g., a 10% budget cut) impacts the final recommendation.
Watson Integration: Seamlessly connects with IBM's AI tools to use Predictive outputs (e.g., demand forecasts) as inputs for Prescriptive decisions. 

3. AnyLogic (The Simulation Specialist)
While the other tools focus on "Mathematical Programming" (equations), AnyLogic focuses on Simulation Modeling.
Dynamic Environments: It is best for scenarios where variables change over time in ways that equations can't easily capture—like the flow of people in an airport or the movement of trucks in a global supply chain.
Hybrid Approach: You can run an AnyLogic simulation and "call" a Python optimizer (like PuLP or Gurobi) mid-run to find the best possible move at that specific moment.

| Analysis Type | Complexity | Time Orientation | Business Value | Typical Questions |
|---------------|------------|------------------|----------------|-------------------|
| **Descriptive** | Low | Past | Foundation | What were our sales last month? |
| **Diagnostic** | Medium | Past | Understanding | Why did sales increase in region X? |

## V. Practical Implementation Framework

### 1. **Start with the Right Questions:**
- **Descriptive:** "What is our current state?"
- **Diagnostic:** "Why are we in this state?"
- **Predictive:** "What could happen next?"
- **Prescriptive:** "What's the best action to take?"

### 2. **Choose Appropriate Tools:**
- **Descriptive:** Excel, BI Tools, SQL
- **Diagnostic:** Statistical software, Data mining tools
- **Predictive:** ML platforms, Forecasting tools
- **Prescriptive:** Optimization software, Simulation tools

### 3. **Build Maturity Gradually:**
```
Level 1: Descriptive Analytics → Basic reporting
Level 2: Diagnostic Analytics → Root cause analysis
Level 3: Predictive Analytics → Forecasting models
Level 4: Prescriptive Analytics → Automated decision-making
```

### 4. **Common Pitfalls to Avoid:**
- Jumping to predictive without solid descriptive foundation
- Using inappropriate data types for analysis
- Ignoring data quality issues
- Overcomplicating simple descriptive questions
- Not aligning analysis type with business objectives
