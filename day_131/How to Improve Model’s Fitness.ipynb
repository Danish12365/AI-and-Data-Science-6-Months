# 🔷 1. What is “Model Fitness”?

👉 Model fitness means:

* How well your model **learns patterns**
* How well it **predicts new (unseen) data**

### 📊 Two key things:

| Term             | Meaning                                          |
| ---------------- | ------------------------------------------------ |
| **Underfitting** | Model is too simple (bad performance everywhere) |
| **Overfitting**  | Model memorizes training data (bad on test data) |

---

# 🔷 2. Goal of Improving Fitness

👉 Balance between:

* Bias (underfitting)
* Variance (overfitting)

📌 Ideal:

> High accuracy on BOTH training and testing data
