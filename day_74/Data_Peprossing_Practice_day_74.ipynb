# DATA PREPROCESSING STEPS

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "CustomerID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Name": ["John Doe", "Jane Smith", "Ali Khan", "Sara Ali", "Michael Brown", "Lily Chen", "David Lee", "Maria Gomez"],
    "Age": [25, 30, None, 29, 35, 28, 40, None],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "AnnualIncome": [50000, 60000, 40000, None, 45000, 75000, 30000, 55000],
    "SpendingScore": [75, 80, 60, 85, 50, 90, 30, None],
    "Country": ["USA", "Canada", "Pakistan", "UK", "USA", "China", "Australia", "Mexico"],
    "PurchaseCategory": ["Electronics", "Clothing", "Groceries", "Clothing", "Groceries", None, "Electronics", "Clothing"]
}

df = pd.DataFrame(data)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

# Step 1: Data Cleaning
1.1 Handle Missing Values

We will address missing values in the dataset, considering different strategies based on the data type and distribution.

```python

# Handle missing values without using inplace=True to avoid the FutureWarning
df['Age'] = df['Age'].fillna(df['Age'].median())
df['AnnualIncome'] = df['AnnualIncome'].fillna(df['AnnualIncome'].mean())
df['SpendingScore'] = df['SpendingScore'].fillna(df['SpendingScore'].mean())
df['PurchaseCategory'] = df['PurchaseCategory'].fillna(df['PurchaseCategory'].mode()[0])

print("\nData after handling missing values:\n", df)

```

```python
# Categorical column - Fill with mode
df['PurchaseCategory'] = df['PurchaseCategory'].fillna(df['PurchaseCategory'].mode()[0])

print("\nData after handling missing values:\n", df)
```

# 1.2 Handle Noisy Data and Outliers

We can detect outliers using techniques like the Interquartile Range (IQR) method.

```python
# Outlier detection using IQR for 'AnnualIncome'
Q1 = df['AnnualIncome'].quantile(0.25)
Q3 = df['AnnualIncome'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
df = df[(df['AnnualIncome'] >= lower_bound) & (df['AnnualIncome'] <= upper_bound)]
print("\nData after removing outliers:\n", df)
```

# 1.3 Remove Duplicates

Remove any duplicate rows if present.

```python
df.duplicated()
```

```python
# Check and remove duplicates
df.drop_duplicates(inplace=True)
print("\nData after removing duplicates:\n", df)
```

# 1.4 Normalize Data

We’ll normalize AnnualIncome using Min-Max scaling.

```python
sns.histplot(df['AnnualIncome'], kde=True)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df['AnnualIncome'] = scaler.fit_transform(df[['AnnualIncome']])
print("\nData after normalization:\n", df)
```

# Step 3: Data Transformation
3.1 Feature Encoding

We will encode categorical data such as Gender and Country.

```python
# One-hot encode 'Gender' and 'Country'
df = pd.get_dummies(df, columns=['Gender', 'Country'], drop_first=True)
print("\nData after encoding:\n", df)
```

# 3.3 Feature Engineering

Create new features if necessary, such as combining SpendingScore and AnnualIncome into an EconomicScore.

```python
# Feature engineering: Economic Score
df['EconomicScore'] = df['AnnualIncome'] * df['SpendingScore']
print("\nData after feature engineering:\n", df)
```
