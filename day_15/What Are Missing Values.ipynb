# 1️⃣ What Are Missing Values?

Missing values occur when **no data is stored** for a variable.

### Common representations

* `NaN` (Python / Pandas)
* `NULL` (SQL)
* Empty cells
* Special codes: `-999`, `0`, `"?"`

📌 **Important**: Missing ≠ Zero ≠ Empty string

---

# 2️⃣ Why Missing Values Matter

Missing data can:

* Bias your analysis
* Reduce model accuracy
* Break ML algorithms
* Lead to wrong business decisions

👉 **Handling missing values is NOT optional**

---

# 3️⃣ Types of Missing Data (VERY IMPORTANT)

Understanding this decides **which method to use**.

## 1️⃣ MCAR – Missing Completely At Random

Example:

* Sensor failed randomly
* User skipped question randomly

✔ Safe to remove or simple impute

---

## 2️⃣ MAR – Missing At Random

Example:

* Income missing more for younger users
* Medical test missing depending on age

✔ Use **smart/statistical/ML imputation**

---

## 3️⃣ MNAR – Missing Not At Random

Example:

* People with high income don’t report salary
* Sick patients skip health forms

⚠ Most dangerous
✔ Requires **domain knowledge + advanced methods**

---

# 4️⃣ First Step: Always Analyze Missing Values

### In Pandas

```python
df.isnull().sum()
df.isnull().mean() * 100
```

### Visual tools (Modern Practice)

* Missingno library

```python
import missingno as msno
msno.matrix(df)
msno.heatmap(df)
```

📌 **This is now standard in real projects**

---

# 5️⃣ Classical Methods (Still Important)

## 1️⃣ Delete Rows

```python
df.dropna()
```

✔ Use when:

* Missing < 5%
* Large dataset
  ❌ Risk: data loss

---

## 2️⃣ Delete Columns

✔ Use when:

* Column has > 40–50% missing
* Column not business-critical

---

## 3️⃣ Simple Imputation

### Mean / Median / Mode

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
df['age'] = imputer.fit_transform(df[['age']])
```

| Data Type        | Best Choice |
| ---------------- | ----------- |
| Numeric (Normal) | Mean        |
| Numeric (Skewed) | Median      |
| Categorical      | Mode        |

---

# 6️⃣ Modern & Best-Practice Methods (🔥 Important)

These are **expected in ML projects today**.

---

## 1️⃣ Missing Indicator (Very Powerful)

Instead of hiding missing values, **use them as information**.

```python
df['age_missing'] = df['age'].isnull().astype(int)
```

✔ Widely used in:

* Finance
* Healthcare
* Fraud detection

---

## 2️⃣ KNN Imputation

Uses similarity between rows.

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=5)
df_imputed = imputer.fit_transform(df)
```

✔ Better than mean
❌ Slow for large datasets

---

## 3️⃣ Iterative Imputer (MICE) ⭐

**Industry-standard modern method**

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

imputer = IterativeImputer()
df_imputed = imputer.fit_transform(df)
```

✔ Models each feature as a function of others
✔ Very strong for MAR data

---

## 4️⃣ Model-Based Imputation

Train a model to predict missing values.

Example:

* Random Forest
* XGBoost

✔ Used in **high-stakes ML systems**

# 🔟 Best Practices Used Today (2025)

✅ Always analyze missingness first
✅ Add missing indicators
✅ Use **median / KNN / Iterative** for ML
✅ Never impute using **target variable**
✅ Fit imputer **only on training data**
✅ Use pipelines

```python
from sklearn.pipeline import Pipeline
```

# 1️⃣1️⃣ Interview Golden Answer (Short Version)

> “First, I analyze missingness type (MCAR, MAR, MNAR).
> For small missingness, I use median/mode.
> For ML, I prefer Iterative Imputer or KNN with missing indicators.
> For tree models, I often keep missing values.
> I always fit imputers on training data only.”

❌ Removing too much data
❌ Ignoring missing patterns
❌ Data leakage during imputation
