# NAIVE Bayes algorithm

Naive Bayes Algorithm is a classification algorithm based on Bayes Theorem. It is called naive because it assumes that the features in a dataset are independent of each other. This assumption is not true in real life but it simplifies the computation and gives good results in most of the cases.

## Bayes Theorem

Bayes Theorem is a mathematical formula used for calculating conditional probability. It is defined as:

$$P(A|B) = \frac{P(B|A)P(A)}{P(B)}$$

where A and B are events and P(B) != 0

## Naive Bayes Algorithm

Naive Bayes Algorithm is based on Bayes Theorem. It is defined as:

$$P(y|x_1,x_2,...,x_n) = \frac{P(x_1,x_2,...,x_n|y)P(y)}{P(x_1,x_2,...,x_n)}$$

where y is the class variable and x1, x2, ..., xn are the features.

The algorithm assumes that the features are independent of each other. So, the above equation can be written as:

$$P(y|x_1,x_2,...,x_n) = \frac{P(x_1|y)P(x_2|y)...P(x_n|y)P(y)}{P(x_1,x_2,...,x_n)}$$

The denominator is constant for a given input. So, the equation can be written as:

$$P(y|x_1,x_2,...,x_n) \propto P(x_1|y)P(x_2|y)...P(x_n|y)P(y)$$

The class with the highest probability is the output of the algorithm.

````python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.datasets import load_iris
````

````python
# load the dataset

iris = load_iris()
X = iris.data
y = iris.target

# train test split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
````

````python
# model initialize
gnb = GaussianNB()

# train the model
gnb.fit(X_train, y_train)

# predict the test data
y_pred = gnb.predict(X_test)

# evaluate the model
print("Accuracy Score: ", accuracy_score(y_test, y_pred))
print("Confusion Matrix: \n", confusion_matrix(y_test, y_pred))
print("Classification Report: \n", classification_report(y_test, y_pred))
````

````python
# model initialize
mnb = MultinomialNB()

# train the model
mnb.fit(X_train, y_train)

# predict the test data
y_pred = mnb.predict(X_test)

# evaluate the model
print("Accuracy Score: ", accuracy_score(y_test, y_pred))
print("Confusion Matrix: \n", confusion_matrix(y_test, y_pred))
print("Classification Report: \n", classification_report(y_test, y_pred))
````

````python
# model initialize
bnb = BernoulliNB()

# train the model
bnb.fit(X_train, y_train)

# predict the test data
y_pred = bnb.predict(X_test)

# evaluate the model
print("Accuracy Score: ", accuracy_score(y_test, y_pred))
print("Confusion Matrix: \n", confusion_matrix(y_test, y_pred))
print("Classification Report: \n", classification_report(y_test, y_pred))
````

The **Naive Bayes** algorithm is a family of probabilistic classifiers based on **Bayes' Theorem**. It is famous for being incredibly fast, simple to implement, and surprisingly effective for high-dimensional data like text.

The "Naive" part of the name comes from a bold assumption: it assumes that every feature in your dataset is **completely independent** of every other feature.

---

## 1. The Mathematical Foundation: Bayes' Theorem

At its core, Naive Bayes calculates the probability that an event will occur, given that another event has already occurred.

$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

In machine learning terms, we use it like this:

* **$P(Class|Features)$:** The probability that a row belongs to a certain class (e.g., "Spam") given its features (the words in the email).
* **$P(Features|Class)$:** The likelihood—how often these features appear in that class.
* **$P(Class)$:** The "Prior" probability—how common the class is in general.
* **$P(Features)$:** The "Evidence"—how common the features are across all data.

---

## 2. Why is it called "Naive"?

Imagine you are identifying a fruit. It is **red**, **round**, and **about 3 inches in diameter**.

* A "Smart" model knows that if a fruit is red and round, it’s more likely to be an apple.
* **Naive Bayes** treats these features as if they have nothing to do with each other. It calculates the probability of "Red," the probability of "Round," and the probability of "3 inches" separately and multiplies them together.

Even though this assumption is almost always false in the real world (features are usually correlated), Naive Bayes still performs remarkably well, especially for **Sentiment Analysis** and **Spam Filtering**.

---

## 3. The Different "Flavors" of Naive Bayes

Depending on the nature of your features, you must choose the right version:

| Version | Best For... | Example |
| --- | --- | --- |
| **Gaussian Naive Bayes** | Continuous/Numerical data that follows a Normal Distribution. | Predicting Iris flower species based on petal length/width. |
| **Multinomial Naive Bayes** | Discrete data, specifically **word counts**. | Spam detection or classifying news articles into categories. |
| **Bernoulli Naive Bayes** | Binary/Boolean features (Yes/No). | Identifying if a specific word is *present* or *absent* in a document. |

---

## 4. Pros and Cons

**Pros:**

* **Lightning Fast:** It is much faster than Random Forests or SVMs.
* **Scalability:** It can handle millions of features easily (very important for NLP).
* **Small Data:** It works well even if you don't have a massive training set.

**Cons:**

* **Independence Assumption:** If your features are highly correlated (e.g., "Age" and "Years of Experience"), Naive Bayes will struggle.
* **Zero Frequency Problem:** If a category in the test set wasn't in the training set, the probability becomes 0, and the model can't make a prediction. (We fix this with **Laplace Smoothing**).

At its most basic level, **Bayes' Theorem** is a mathematical formula used to determine the probability of an event based on prior knowledge of conditions that might be related to that event.

In data science, we use it to update our "beliefs" as we get new evidence.

---

## 1. The Core Concept: "Updating Beliefs"

Imagine you are a doctor. A patient comes in with a cough.

* **Prior Probability:** Before checking anything, you know that 5% of people in the city have the flu.
* **New Evidence:** The patient has a high fever.
* **Posterior Probability:** Now that you have the "evidence" (fever), your "belief" that they have the flu increases from 5% to, say, 60%.

Bayes' Theorem is the math that tells you exactly how much that probability should change.

---

## 2. The Formula

The formula looks like this:

$$P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

### Breaking down the terms:

* **$P(A|B)$ (Posterior):** The probability of event $A$ happening **given** that $B$ has already happened. (e.g., Probability of Flu given a Fever).
* **$P(B|A)$ (Likelihood):** The probability of $B$ happening given that $A$ is true. (e.g., How many people with Flu actually get a Fever?).
* **$P(A)$ (Prior):** The initial probability of $A$ before seeing any evidence. (e.g., General rate of Flu in the population).
* **$P(B)$ (Evidence):** The total probability of $B$ happening under all conditions. (e.g., How many people in total have a fever, for any reason?).

---

## 3. Real-World Example: Medical Testing

This is the classic "False Positive" problem that every Data Scientist must understand.

**The Scenario:**

* **1%** of the population has a rare disease ($P(A) = 0.01$).
* A test is **99%** accurate ($P(B|A) = 0.99$).
* The test has a **5%** false positive rate ($P(B|\text{not } A) = 0.05$).

**The Question:** If you test positive, what is the actual chance you have the disease?
Most people guess 99%. **Bayes' Theorem says it is only about 16%.**

**Why?** Because the disease is so rare ($1\%$) that the $5\%$ of healthy people getting false positives far outnumbers the $1\%$ of sick people getting true positives.
