# 🔷 1. Business Intelligence (BI) with Power BI

👉 **Business Intelligence (BI)** = using data to make **smart business decisions**

---

## 🧠 Simple Definition

> BI = “Raw data → Insights → Decisions”

---

## 📊 Example (Real Life)

A company wants to know:

* Which product sells most?
* Which region gives more profit?
* Why sales dropped last month?

👉 Power BI helps answer these using dashboards

---

## 🔥 BI Process in Power BI

1. Collect data
2. Clean data
3. Analyze data
4. Visualize data
5. Share insights

---

# 🔷 2. Seven Steps Workflow in Power BI

👉 This is the **real workflow used in companies**

---

## 🪜 Step 1: Connect Data

* Import data (Excel, SQL, CSV)

---

## 🪜 Step 2: Transform Data

* Clean data using Power Query
* Handle missing values

---

## 🪜 Step 3: Data Modeling

* Create relationships between tables

---

## 🪜 Step 4: Create Measures (DAX)

* Calculate KPIs
* Example: Total Sales, Profit

---

## 🪜 Step 5: Build Visualizations

* Charts, graphs, maps

---

## 🪜 Step 6: Create Dashboard

* Combine visuals into one screen

---

## 🪜 Step 7: Share & Publish

* Upload to Power BI Service

---

📌 This workflow = **end-to-end BI process**

# 🔷 4. Connect Data in Power BI

👉 First and most important step

---

## 🪜 Step-by-Step

---

### 🟢 Method 1: Excel File

1. Open Power BI
2. Click **Home → Get Data → Excel**
3. Select file
4. Click **Load**

---

---

### 🔵 Method 2: CSV File

1. Home → Get Data → Text/CSV
2. Select file
3. Load

---

---

### 🟡 Method 3: Database (SQL)

1. Home → Get Data → SQL Server
2. Enter server details
3. Connect

---

---

### 🔴 Method 4: Web / API

1. Home → Get Data → Web
2. Enter URL

---

---

## 🔧 Load vs Transform

---

### 🟢 Load

👉 Directly load data

---

### 🔵 Transform (Recommended)

👉 Open Power Query to:

* Clean data
* Remove errors
* Change data types
