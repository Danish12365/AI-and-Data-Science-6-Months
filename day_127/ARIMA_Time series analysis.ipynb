ARIMA is one of the **most important models in Time Series Analysis**, especially for forecasting data like stock prices, sales, temperature, etc. 

---

# 🔷 1. What is ARIMA?

**ARIMA = AutoRegressive Integrated Moving Average**

It is a statistical model used to **predict future values based on past values**.

👉 It has 3 parts:

* **AR (AutoRegressive)** → uses past values
* **I (Integrated)** → makes data stationary
* **MA (Moving Average)** → uses past errors

We write it as:

[
ARIMA(p, d, q)
]

---

# 🔷 2. Understanding Each Component Deeply

## 🟢 (1) AR – AutoRegressive (p)

👉 Uses previous values to predict current value.

Example:

[
y_t = c + \phi_1 y_{t-1} + \phi_2 y_{t-2} + \dots
]

✔ Meaning:

* Today depends on yesterday, day before, etc.

📌 Real-life:

* Stock price today depends on previous days

---

## 🟡 (2) I – Integrated (d)

👉 Makes data **stationary**

### ❓ What is Stationary Data?

Data whose:

* Mean is constant
* Variance is constant
* No trend

📌 Most real-world data is **NOT stationary**, so we apply:

### Differencing:

[
y'*t = y_t - y*{t-1}
]

✔ If still not stationary → do it again

👉 `d = number of times differencing applied`

---

## 🔵 (3) MA – Moving Average (q)

👉 Uses past errors (residuals)

[
y_t = \mu + \theta_1 e_{t-1} + \theta_2 e_{t-2}
]

✔ Meaning:

* Model corrects itself using past mistakes

---

# 🔷 3. When to Use ARIMA?

Use ARIMA when:

✅ Data is time-based
✅ Data shows trend (can be removed)
✅ No strong seasonality

❌ Not ideal for:

* Strong seasonal data → use SARIMA
* Non-linear patterns → use ML/DL models

# 🔷 4. ARIMA Workflow (VERY IMPORTANT)

This is what interviewers and real projects expect:

---

## 🧩 Step 1: Load Data

```python
import pandas as pd

data = pd.read_csv("data.csv", parse_dates=['Date'], index_col='Date')
```

---

## 📊 Step 2: Plot Data

```python
import matplotlib.pyplot as plt

data.plot()
plt.show()
```

---

## 🔍 Step 3: Check Stationarity

Use **ADF Test (Augmented Dickey-Fuller)**

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(data['value'])
print(result[1])  # p-value
```

✔ If p-value > 0.05 → NOT stationary
✔ If p-value < 0.05 → stationary

---

## 🔄 Step 4: Make Data Stationary

```python
data_diff = data.diff().dropna()
```

Repeat until stationary

---

## 📉 Step 5: Find p and q (IMPORTANT)

Use:

* **ACF (AutoCorrelation Function)** → find q
* **PACF (Partial AutoCorrelation)** → find p

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

plot_acf(data_diff)
plot_pacf(data_diff)
```

📌 Rule:

* PACF cutoff → p
* ACF cutoff → q

---

## 🧠 Step 6: Build ARIMA Model

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(data['value'], order=(p, d, q))
model_fit = model.fit()
print(model_fit.summary())
```

---

## 🔮 Step 7: Forecast

```python
forecast = model_fit.forecast(steps=10)
print(forecast)
```

---

# 🔷 5. Full Example (Simple)

```python
from statsmodels.tsa.arima.model import ARIMA
import pandas as pd

# Load data
data = pd.read_csv("data.csv", index_col='Date', parse_dates=True)

# Build model
model = ARIMA(data['value'], order=(1,1,1))
model_fit = model.fit()

# Forecast
forecast = model_fit.forecast(steps=5)

print(forecast)
```

---

# 🔷 6. Important Concepts (VERY IMPORTANT)

## ⚠️ 1. Overfitting

Too high p, q → model memorizes data
👉 Keep it simple

---

## ⚠️ 2. Residuals Check

```python
model_fit.plot_diagnostics()
```

✔ Residuals should be:

* Random
* No pattern

---

## ⚠️ 3. AIC (Model Selection)

Lower AIC = better model

```python
print(model_fit.aic)
```

---

# 🔷 7. Advanced Version: Auto ARIMA

Automatically finds best (p, d, q)

```python
from pmdarima import auto_arima

model = auto_arima(data['value'], seasonal=False)
print(model.summary())
```
