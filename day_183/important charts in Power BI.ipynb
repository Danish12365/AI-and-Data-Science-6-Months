# 🔷 1. Pie Charts in Power BI

---

## 🧠 What is Pie Chart?

👉 Shows **parts of a whole (percentage distribution)**

---

## 📊 Example

* Sales by Category:

  * Electronics → 40%
  * Clothing → 30%
  * Food → 30%

---

## 🪜 How to Create

1. Select **Pie Chart**
2. Drag:

   * Category → Legend
   * Sales → Values

---

## ✅ When to Use

✔ Few categories (max 4–5)
✔ Show percentage share

---

## ❌ When NOT to Use

❌ Too many categories
❌ Small differences

---

## 🔥 Pro Tips

* Show **percent labels**
* Use clear colors
* Highlight important slice

---

# 🔷 2. Donut Charts in Power BI

---

## 🧠 What is Donut Chart?

👉 Same as pie, but with **hole in center**

---

## 📊 Difference

| Pie                 | Donut                  |
| ------------------- | ---------------------- |
| Full circle         | Circle with hole       |
| Less space for info | Can show KPI in center |

---

## 🪜 How to Create

1. Select **Donut Chart**
2. Drag same fields

---

## 🔥 Pro Advantage

👉 Add **Total Value in center**

Example:

* Total Sales = 1M

---

## ⚠️ Tip

👉 Use donut when:

* You want **clean design + KPI focus**

---

# 🔷 3. Tree Map in Power BI

---

## 🧠 What is Tree Map?

👉 Shows **hierarchical data using rectangles**

---

## 📊 Example

* Category → Sub-category → Sales

---

## 🪜 How to Create

1. Select **Treemap**
2. Drag:

   * Category → Group
   * Sales → Values

---

## 📌 What It Shows

* Size = value
* Color = category

---

## ✅ When to Use

✔ Many categories
✔ Compare proportions

---

## ❌ Avoid When

❌ Need precise comparison
❌ Very small differences

---

## 🔥 Pro Tips

* Use **color grouping**
* Avoid too many levels
* Add labels

---

# 🔷 4. Line Charts in Power BI

---

## 🧠 What is Line Chart?

👉 Shows **trend over time**

---

## 📊 Example

* Monthly sales
* Daily website visits

---

## 🪜 How to Create

1. Select **Line Chart**
2. Drag:

   * Date → Axis
   * Sales → Values

---

## 🔥 Add More Power

* Add **Category → Legend** (multiple lines)
* Add **Markers**

---

## ✅ When to Use

✔ Time-based data
✔ Trends & patterns

---

## ❌ Avoid When

❌ No time dimension
❌ Too many lines

---

## 🔥 Pro Tips

* Use **continuous date axis**
* Highlight key trends
* Add moving average (advanced)

---

# 🔷 5. Comparison (VERY IMPORTANT)

| Chart   | Best Use              |
| ------- | --------------------- |
| Pie     | Simple % distribution |
| Donut   | % + KPI display       |
| Treemap | Many categories       |
| Line    | Trends over time      |
