Swagger docs in FastAPI are one of its **biggest advantages**—you get **automatic, interactive API documentation** without writing extra code.

Let’s break it down clearly 👇

---

# 🔷 1. What are Swagger Docs?

👉 Swagger Docs = **Interactive API documentation**

In FastAPI, they are **auto-generated**.

---

## 📌 URL

When you run your FastAPI app:

```text
http://127.0.0.1:8000/docs
```

👉 This opens Swagger UI

---

# 🔷 2. Why Swagger Docs are Powerful

✅ Test APIs directly from browser
✅ No need for Postman
✅ Shows request & response formats
✅ Helps frontend developers

---

# 🔷 3. Example (Auto Swagger)

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/books/")
def get_books():
    return [{"title": "Python"}]
```

👉 Swagger automatically shows:

* Endpoint `/books/`
* Method GET
* Response

---

# 🔷 4. With Request Body (Better Example)

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class Book(BaseModel):
    title: str
    author: str
    price: float

@app.post("/books/")
def create_book(book: Book):
    return book
```

---

## 🔥 Swagger will show:

* Input form (JSON)
* Required fields
* Data types

---

# 🔷 5. Interactive Testing (Best Feature)

👉 In Swagger UI:

1. Click endpoint
2. Click **"Try it out"**
3. Enter data
4. Click **Execute**

📌 Example JSON:

```json
{
  "title": "AI Book",
  "author": "Danish",
  "price": 20
}
```

# 🔷 6. Add Description & Metadata

Make your API **professional** 👇

```python
app = FastAPI(
    title="Book Store API",
    description="API for managing books",
    version="1.0.0"
)
```

---

# 🔷 7. Add Endpoint Descriptions

```python
@app.post("/books/", summary="Create a new book", description="Add a book to the store")
def create_book(book: Book):
    return book
```

---

# 🔷 8. Add Response Models

```python
@app.post("/books/", response_model=Book)
def create_book(book: Book):
    return book
```

👉 Swagger shows correct output format

---

# 🔷 9. Path & Query Parameters

```python
@app.get("/books/{book_id}")
def get_book(book_id: int):
    return {"book_id": book_id}
```

👉 Swagger shows:

* Required parameter
* Type (int)

---

# 🔷 10. Alternative Docs (ReDoc)

👉 FastAPI also provides:

```text
http://127.0.0.1:8000/redoc
```

* Cleaner UI
* Better for reading

---

# 🔷 11. Customize Docs URL

```python
app = FastAPI(docs_url="/swagger", redoc_url="/redoc")
```

👉 Now:

* `/swagger` → Swagger UI
* `/redoc` → ReDoc

---

# 🔷 12. Disable Docs (Production)

```python
app = FastAPI(docs_url=None, redoc_url=None)
```

👉 Used for security

---

# 🔷 13. Advanced Features

---

## 🔐 Add Authentication in Swagger

```python
from fastapi.security import OAuth2PasswordBearer

oauth2_scheme = OAuth2PasswordBearer(tokenUrl="token")
```

👉 Swagger shows login option

---

## 📦 Tag Grouping

```python
@app.get("/books/", tags=["Books"])
```

👉 Organizes endpoints
