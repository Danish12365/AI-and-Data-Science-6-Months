## 🔥 Why Tableau is Important

# 🔷 5. Load Data in Tableau Public

---

## 📥 Step-by-Step:

1. Open Tableau
2. Click **“Text File / Excel”**
3. Select your dataset
4. Click **Open**

---

👉 Example:

* CSV file
* Excel file

---

# 🔷 7. Data Types Check in Tableau

---

## 📊 Types:

| Type    | Example    |
| ------- | ---------- |
| String  | Name       |
| Number  | Sales      |
| Date    | Order Date |
| Boolean | True/False |

---

## 🔍 How to Check:

👉 Look at icon beside column name:

* ABC → Text
* # → Number
* Calendar → Date

---

## 🔧 Change Data Type:

Right click column → Change Data Type

---

# 🔷 8. Data Filtering in Tableau

---

## 📌 Method 1: Drag to Filter Shelf

1. Drag field → Filters
2. Select values

---

## 📌 Method 2: Quick Filter

Right click field → Show Filter

---

## 💡 Example:

Filter:

* Region
* Category
* Date

---

# 🔷 9. Sorting in Tableau

---

## 🔄 Sort Data:

👉 Click sort icon in toolbar

OR

Right click field → Sort

---

## 📊 Example:

* Sales descending
* Alphabetical order
