# 🤖 K-Means Clustering: An Unsupervised Learning Algorithm

**K-means clustering** is a popular **unsupervised machine learning algorithm** used to partition a dataset into a pre-defined number, $K$, of distinct, non-overlapping groups (clusters).

The core objective is to group data points with similar characteristics together, where similarity is typically measured by the **Euclidean distance** between points.

-----

## 🎯 How the K-Means Algorithm Works

The K-means algorithm is an **iterative process** that works to minimize the **Within-Cluster Sum of Squares (WCSS)**, which is the total squared distance between each data point and the **centroid** (center) of the cluster it is assigned to.

The process follows these steps:

1.  **Initialization**:

      * The user specifies the number of clusters, $K$.
      * The algorithm randomly selects $K$ data points from the dataset to serve as the initial **centroids** (cluster centers).

2.  **Assignment Step**:

      * Each data point in the dataset is assigned to the nearest centroid based on a distance metric, most commonly the Euclidean distance. This forms the initial clusters.

3.  **Update Step**:

      * The centroids are recomputed by calculating the **mean (average) position** of all the data points currently within each cluster.

4.  **Repeat**:

      * Steps 2 and 3 are repeated. The algorithm continues to iterate, reassigning points and recalculating centroids, until a condition is met (e.g., cluster assignments no longer change, centroids stop moving, or a max number of iterations is reached).

## 💻 Python Implementation Example (Scikit-learn)

Here is a typical Python implementation of K-means clustering in a Jupyter Notebook using `scikit-learn` for demonstration.

### 1\. Import Libraries and Generate Sample Data

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import warnings

# Suppress all warnings for clean notebook output
warnings.filterwarnings("ignore")

# 1. Generate a synthetic dataset with 4 distinct clusters (K=4)
# X contains the features, y contains the true (but ignored by KMeans) cluster labels
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Visualize the generated data
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], s=50) # Plot all data points
plt.title("Sample Data for Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True, alpha=0.3)
plt.show()
```

### 2\. Determine Optimal $K$ using the Elbow Method

The **Elbow Method** is a heuristic used to find the optimal number of clusters by plotting the WCSS for a range of $K$ values. The "elbow" point, where the rate of decrease slows down significantly, is a good candidate for $K$.

```python
# Calculate WCSS (Inertia) for different numbers of clusters
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=0)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_) # inertia_ is the WCSS

# Plot the Elbow Graph
plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Inertia)')
plt.grid(True, alpha=0.3)
plt.show()

print(f"Based on the Elbow curve, the optimal K is likely 4 (where the bend occurs).")
```

### 3\. Apply K-Means Clustering

```python
# Create the KMeans model with the optimal K=4
kmeans = KMeans(n_clusters=4, init='k-means++', max_iter=300, n_init=10, random_state=0)

# Fit the model and get the cluster labels for each data point
y_kmeans = kmeans.fit_predict(X)

# Get the final coordinates of the centroids
centroids = kmeans.cluster_centers_
```

### 4\. Visualize the Final Clusters

```python
# Plot the data points colored by their assigned cluster
plt.figure(figsize=(8, 5))
plt.title("K-Means Clustering Results (K=4)")

# Scatter plot for each cluster
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1], s=50, c='red', label='Cluster 1')
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1], s=50, c='blue', label='Cluster 2')
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1], s=50, c='green', label='Cluster 3')
plt.scatter(X[y_kmeans == 3, 0], X[y_kmeans == 3, 1], s=50, c='cyan', label='Cluster 4')

# Plot the centroids
plt.scatter(centroids[:, 0], centroids[:, 1], s=200, c='yellow', marker='*', label='Centroids', edgecolor='black')

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 💡 Important Terms in Clustering

Let’s explain all those technical words in simple language 👇

---

### 🔹 **Intra-cluster Distance**

* “Intra” means **within**.
* It measures **how close or tight the data points are inside the same cluster**.
* **Smaller intra-cluster distance = better cluster** (more similar points).

🧠 Example:
If you have a cluster of apples, all apples should be close together — that means low intra-cluster distance.

---

### 🔹 **Inter-cluster Distance**

* “Inter” means **between**.
* It measures **how far apart the clusters are from each other**.
* **Larger inter-cluster distance = better clustering** (clusters are well-separated).

🧠 Example:
The cluster of apples 🍎 should be far from bananas 🍌 → that means high inter-cluster distance.

---

### 📊 Ideal Clustering:

👉 **Low Intra-cluster Distance** (tight groups)
👉 **High Inter-cluster Distance** (clear separation)

---

### 🔹 **Centroid**

* The **center point** of a cluster (average of all points in that cluster).
  Used mostly in **K-Means**.

---

### 🔹 **Similarity**

* It shows **how much two data points are alike**.
* Higher similarity → more likely to be in the same cluster.

Example: Two customers with almost same age & spending pattern are “similar.”

---

### 🔹 **Distance**

* Opposite of similarity — measures how far apart two data points are.
* Common distance measures:

  * **Euclidean Distance** (straight line distance)
  * **Manhattan Distance** (distance along axes)
  * **Cosine Similarity** (angle between two vectors)

👉 In clustering, smaller distance = more similar points.

---

### 🔹 **Feature Scaling**

Before clustering, we **standardize** the features (using StandardScaler or MinMaxScaler).
Because distance calculations can be biased if features have different scales (e.g., age in years vs income in dollars).

## 🧠 **K-Means Clustering (Full Explanation)**

### 🔹 What is K-Means Clustering?

K-means clustering is a **popular unsupervised machine learning algorithm** used to group data into a **predefined number of clusters ((k))** based on similarity.

It tries to **minimize the distance between data points and their cluster centers (centroids)** — meaning, it groups together points that are “close” to each other in terms of numerical distance (usually **Euclidean distance**).

So, in simple words:
👉 *K-means tries to find natural groups in data, where each group (cluster) has points that are similar to one another.*

---

### ⚙️ **How the Algorithm Works**

K-means works through an **iterative optimization** process. It continuously refines cluster boundaries until they stabilize.

#### **Step 1: Initialization**

* You choose the number of clusters, (k).
* The algorithm then randomly selects (k) data points from your dataset as **initial centroids**.

#### **Step 2: Assignment Step**

* Each data point is assigned to the **nearest centroid** based on the chosen **distance metric** (usually Euclidean distance).
  This creates the initial cluster groupings.

#### **Step 3: Update Step**

* The algorithm then **recalculates the centroid** of each cluster by taking the **mean position** of all data points in that cluster.

#### **Step 4: Repeat**

* The **assignment** and **update** steps are repeated until:

  * Centroids stop changing significantly, or
  * The cluster assignments stop changing, or
  * The maximum number of iterations is reached.

The algorithm is guaranteed to converge — though not necessarily to the *global* best solution (sometimes a *local* one).

---

### 📏 **Key Terms in K-Means**

| **Term**                                 | **Meaning**                                                                                             |
| ---------------------------------------- | ------------------------------------------------------------------------------------------------------- |
| **Centroid**                             | The “center” point of a cluster — calculated as the mean of all points in that cluster.                 |
| **Cluster**                              | A group of data points that are similar to one another based on distance.                               |
| **Distance Metric**                      | A measure of how far apart two points are. Common: *Euclidean distance*.                                |
| **Intra-cluster Distance**               | The distance between data points **within the same cluster**. Lower = better (points are closer).       |
| **Inter-cluster Distance**               | The distance between **different cluster centroids**. Higher = better (clusters are well-separated).    |
| **WCSS (Within-Cluster Sum of Squares)** | The sum of squared distances between each point and its cluster centroid — used to measure compactness. |

✅ **Goal of K-Means:**
Minimize **intra-cluster distance (tight clusters)** and maximize **inter-cluster distance (separation)**.

---

### 🎯 **Applications of K-Means**

1. **Customer Segmentation:**
   Group customers by buying habits or demographics.
   Example: “budget,” “premium,” and “occasional” buyers.

2. **Image Segmentation & Compression:**
   Group similar pixels (e.g., color intensity) to simplify or compress images.

3. **Anomaly Detection:**
   Detect outliers (e.g., fraudulent transactions that don’t fit any cluster).

4. **Document Clustering:**
   Group news articles or research papers by topic.

5. **Route Optimization:**
   Cluster delivery addresses to plan efficient delivery routes.

---

### ⚖️ **Advantages and Disadvantages**

#### ✅ **Advantages**

* Simple and easy to implement.
* Fast and efficient on large datasets.
* Works well when clusters are **spherical and evenly sized**.
* Guarantees convergence (it always stops after some iterations).

#### ❌ **Disadvantages**

* You must **predefine (k)** (the number of clusters).
* Sensitive to **initial centroid positions** (can produce different results).
* Struggles with **non-spherical clusters** or varying densities.
* Sensitive to **outliers**, which can distort centroids.

---

## 🔍 **Choosing the Right Methods for K-Means Clustering**

Selecting the right **number of clusters ((k))** and **initialization method** is essential for good performance.

---

### 🔹 **Methods for Choosing the Number of Clusters ((k))**

#### **1. Elbow Method**

* Runs K-means for a range of (k) values.
* Plots **WCSS (Within-Cluster Sum of Squares)** vs (k).
* The “elbow point” — where the rate of WCSS decrease sharply slows down — indicates the **optimal (k)**.
  📈 *Intuition:* After a point, adding more clusters doesn’t improve much.

#### **2. Silhouette Method**

* Measures how well each data point fits in its cluster compared to others.
* Produces a **Silhouette Score** ranging from **-1 to +1**:

  * +1 → well-clustered
  * 0 → overlapping clusters
  * -1 → wrongly assigned
* The (k) with the **highest average silhouette score** is considered optimal.

#### **3. Gap Statistic**

* Compares the WCSS of your data’s clustering to a **null reference dataset** (data with no cluster structure).
* A larger **gap value** means better clustering.

#### **4. Davies-Bouldin Index**

* Measures average similarity between clusters.
* **Lower values** indicate better separation and compactness (better clustering).

---

### 🔹 **Methods for Centroid Initialization**

#### **1. Forgy Method**

* Randomly selects (k) points from the dataset as initial centroids.
  ✅ Pros: Simple and fast.
  ❌ Cons: Can result in poor performance if initial centroids are badly chosen.

#### **2. K-Means++**

* A smarter initialization method designed to improve results.
  🧩 How it works:

  1. Randomly pick the first centroid.
  2. For the remaining centroids, choose points **farther from existing centroids** with higher probability.
     ✅ Pros: Reduces chances of poor initialization, improves convergence.
     ✅ Commonly used in most modern K-means implementations (like in scikit-learn).

---

### 🧮 **Mathematical Objective**

K-means minimizes the **sum of squared distances (SSD)** between each point (x_i) and its assigned centroid (\mu_j):

[
J = \sum_{j=1}^{k} \sum_{i \in C_j} ||x_i - \mu_j||^2
]

where:

* (C_j) = cluster (j)
* (\mu_j) = centroid of cluster (j)
* (||x_i - \mu_j||^2) = squared Euclidean distance

The algorithm stops when this **cost function (J)** no longer decreases significantly.

# which type of input variable we can take in k-means clustering

Let’s go step-by-step in a simple, deep, and practical way 👇

---

## 🧠 **1️⃣ Basic Idea Recap**

K-means clustering works by **measuring distances** (usually **Euclidean distance**) between data points and centroids.
Because of that, it requires variables that make **distance calculation meaningful** — i.e., numerical values.

So, K-means works best when your **input variables are numeric and continuous**.

---

## 🧮 **2️⃣ Type of Input Variables in K-Means**

| **Variable Type**               | **Can We Use It?**       | **Explanation**                                                                                                                                                          |
| ------------------------------- | ------------------------ | ------------------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| **Numerical (Continuous)**      | ✅ Yes (Best Type)        | These are values that can take any number — like *age, income, height, temperature*. K-means is designed for such data because distance calculations make sense.         |
| **Numerical (Discrete)**        | ⚠️ Yes (Sometimes)       | You can use discrete numeric values (*e.g., number of purchases, count of visits*), but only if they represent **quantitative differences**, not categories.             |
| **Categorical (Nominal)**       | ❌ Not directly           | Example: *Gender (Male/Female)*, *City (Lahore, Karachi)*. K-means doesn’t work well here because distance between “Karachi” and “Lahore” is meaningless numerically.    |
| **Ordinal (Ranked Categories)** | ⚠️ Sometimes (carefully) | Example: *Education Level (Primary, Secondary, College)* — you can encode these as numbers (1, 2, 3) but distances are not always meaningful, so interpret with caution. |

---

## 🧰 **3️⃣ Why K-Means Needs Numerical Data**

K-means minimizes the **Euclidean distance** between points and cluster centroids:

[
\text{Distance} = \sqrt{(x_1 - c_1)^2 + (x_2 - c_2)^2 + ... + (x_n - c_n)^2}
]

👉 For this formula to make sense, the variables must be **numerical**.
If you try to plug in words (like “Male” or “Blue”), the algorithm cannot calculate distances properly.

---

## ⚙️ **4️⃣ Preprocessing Non-Numerical Data for K-Means**

If you have **categorical data**, you can still use K-means — but you must **convert it into numbers** using **encoding techniques**:

### 🔹 **One-Hot Encoding**

Turns each category into a binary column (0 or 1).
Example:

| Gender | Male | Female |
| ------ | ---- | ------ |
| Male   | 1    | 0      |
| Female | 0    | 1      |

✅ Good for nominal (unordered) data.

---

### 🔹 **Label Encoding**

Assigns each category a numeric value.
Example:

| City      | Encoded |
| --------- | ------- |
| Lahore    | 1       |
| Karachi   | 2       |
| Islamabad | 3       |

⚠️ Be careful — this implies **ordering**, which K-means might interpret as “distance.”

---

### 🔹 **Feature Scaling**

After encoding, you must **normalize or standardize** data (e.g., using *MinMaxScaler* or *StandardScaler*), because K-means is **scale-sensitive**.

Example:
If you have *Age (20–60)* and *Income (20,000–500,000)*,
Income dominates because it has a larger range — scaling fixes this problem.

---

## 📊 **5️⃣ When K-Means Is Not Suitable**

K-means is **not suitable** when:

* Your dataset is mostly **categorical** (like gender, region, department).
* You have **mixed types** (both numeric and categorical) and cannot meaningfully scale them.

In those cases, use **alternative clustering algorithms**:

| **Algorithm**               | **When to Use**                                               |
| --------------------------- | ------------------------------------------------------------- |
| **K-Modes**                 | For purely categorical data.                                  |
| **K-Prototypes**            | For mixed data (numerical + categorical).                     |
| **Hierarchical Clustering** | For small datasets or when number of clusters (k) is unknown. |
| **DBSCAN**                  | For non-spherical clusters or uneven densities.               |

---

| Discrete Numeric     | No. of Orders, Visits | ⚠️ Yes (carefully)  | Scale if needed                          |
| Ordinal              | Low, Medium, High     | ⚠️ Yes              | Label Encode + Scale                     |
| Nominal              | Gender, City, Color   | ❌ No (directly)     | One-Hot Encode or use K-Modes            |

## How various metrics for selecting optimal number of clusters are calculated
## The thumb rules to select optimal number of clusters using these metrics
## How to implement the cluster validation methods in python
## How to interpret results of these methods

Excellent 👏 — this is one of the most **important** and **often misunderstood** parts of clustering:

> ✅ “How to choose the optimal number of clusters (k), calculate metrics, and interpret them in practice.”

Let’s go step-by-step — starting from **concepts → calculations → Python implementation → interpretation** — in an easy and deep way 💡

---

## 🧠 1️⃣ What Is Cluster Validation?

**Cluster Validation** means **checking how good your clusters are** — whether your K-Means (or other clustering) algorithm has actually grouped the data meaningfully.

We use **cluster validation metrics** to:

* Determine the **optimal number of clusters (k)**
* Evaluate the **quality of clustering**

There are two main types:

1. **Internal Validation:** Uses the data itself (no external labels).
   👉 e.g., Elbow Method, Silhouette Score, Davies-Bouldin Index, Calinski–Harabasz Index.
2. **External Validation:** Requires true labels (used in supervised learning).
   👉 e.g., Rand Index, Mutual Information — *not applicable for pure unsupervised tasks*.

---

## ⚙️ 2️⃣ Key Metrics for Selecting Optimal Number of Clusters (k)

Let’s go one by one 👇

---

### 🔹 **(1) Elbow Method (Using WCSS)**

**Concept:**
K-means tries to minimize **WCSS (Within-Cluster Sum of Squares)** — the total squared distance between each point and its cluster centroid.

[
WCSS = \sum_{j=1}^{k} \sum_{i \in C_j} ||x_i - \mu_j||^2
]

As (k) increases:

* WCSS **decreases** (clusters get smaller and tighter)
* But after a point, improvement becomes marginal

👉 The “**elbow point**” (bend) is the **optimal k** — beyond it, adding clusters doesn’t significantly improve results.

**Thumb Rule:**

> Choose (k) at the “bend” of the curve — where WCSS starts to flatten.

---

### 🔹 **(2) Silhouette Score**

**Concept:**
Measures how well each point fits into its cluster vs. others.

For each data point (i):
[
s(i) = \frac{b(i) - a(i)}{\max(a(i), b(i))}
]

Where:

* (a(i)) = mean distance to all other points in the **same cluster**
* (b(i)) = mean distance to points in the **nearest neighboring cluster**

Range: **-1 → +1**

* **+1:** Perfectly assigned (well-separated)
* **0:** Borderline (overlapping clusters)
* **-1:** Misclassified

**Thumb Rule:**

> Choose the (k) with the **highest average silhouette score** — that’s the best-defined clustering.

---

### 🔹 **(3) Davies–Bouldin Index (DBI)**

**Concept:**
Measures **similarity between clusters** — lower values are better.

[
DBI = \frac{1}{k} \sum_{i=1}^{k} \max_{j \neq i} \frac{S_i + S_j}{M_{ij}}
]

Where:

* (S_i): Average distance between points in cluster (i) and its centroid (cluster compactness)
* (M_{ij}): Distance between centroids (i) and (j) (cluster separation)

**Thumb Rule:**

> Lower DBI = better clustering
> (Compact, well-separated clusters)

---

### 🔹 **(4) Calinski–Harabasz Index (CH Score / Variance Ratio Criterion)**

**Concept:**
Measures ratio of **between-cluster variance** to **within-cluster variance**.

[
CH = \frac{\text{Between-cluster dispersion}}{\text{Within-cluster dispersion}}
]

**Thumb Rule:**

> Higher CH score = better cluster separation and compactness.

---

### 🔹 **(5) Gap Statistic**

**Concept:**
Compares the total within-cluster variation for different (k) values to what is expected under a **null (random) distribution**.

[
Gap(k) = E_n[\log(W_k)] - \log(W_k)
]

Where:

* (E_n[\log(W_k)]): Expected log(WCSS) for a reference dataset with no clusters
* (W_k): Observed WCSS for k clusters

**Thumb Rule:**

> Optimal (k) is the one that maximizes the **Gap statistic**.

---

## 🧩 3️⃣ Summary Table of Metrics

| **Metric**            | **Goal**                                  | **Best Value**      | **Thumb Rule**           |
| --------------------- | ----------------------------------------- | ------------------- | ------------------------ |
| **Elbow (WCSS)**      | Find point of diminishing returns         | N/A                 | Choose (k) at the "bend" |
| **Silhouette**        | Maximize cluster separation & compactness | +1 (best)           | Highest average score    |
| **Davies-Bouldin**    | Minimize intra-cluster similarity         | 0 (best)            | Lowest score             |
| **Calinski-Harabasz** | Maximize variance ratio                   | Higher = better     | Highest score            |
| **Gap Statistic**     | Compare vs. null reference                | Larger gap = better | Choose max gap           |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

# ✅ Step 1: Generate synthetic data
X, y = make_blobs(n_samples=500, centers=4, cluster_std=0.6, random_state=0)

# ✅ Step 2: Define a function for Gap Statistic
# (Scikit-learn doesn't have this built-in, so we calculate it manually)
def compute_gap_statistic(X, k, n_refs=3):
    # Fit Actual Data
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    actual_inertia = kmeans.inertia_
    
    # Generate Reference Datasets (Random uniform distribution)
    ref_inertias = []
    for i in range(n_refs):
        # Create random data with same shape and bounds as X
        random_data = np.random.uniform(low=np.min(X, axis=0), 
                                        high=np.max(X, axis=0), 
                                        size=X.shape)
        kmeans_ref = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans_ref.fit(random_data)
        ref_inertias.append(kmeans_ref.inertia_)
    
    # Calculate Gap: Log(Expected Inertia) - Log(Actual Inertia)
    expected_log_inertia = np.mean(np.log(ref_inertias))
    actual_log_inertia = np.log(actual_inertia)
    gap = expected_log_inertia - actual_log_inertia
    return gap

# ✅ Step 3: Loop through K values
K = range(2, 10)
wcss = []
silhouette = []
dbi = []
ch = []
gaps = []

print("Running clustering metrics (this may take a moment due to Gap Stat)...")

for k in K:
    # Standard K-Means
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    labels = kmeans.labels_

    # 1. Elbow Method (Inertia)
    wcss.append(kmeans.inertia_)
    
    # 2. Silhouette Score
    silhouette.append(silhouette_score(X, labels))
    
    # 3. Davies-Bouldin Index
    dbi.append(davies_bouldin_score(X, labels))
    
    # 4. Calinski-Harabasz Score
    ch.append(calinski_harabasz_score(X, labels))
    
    # 5. Gap Statistic (New)
    gap_val = compute_gap_statistic(X, k)
    gaps.append(gap_val)

# ✅ Step 4: Plotting All Methods
plt.figure(figsize=(10, 10))

# Plot 1: Elbow Method
plt.subplot(3, 2, 1)
plt.plot(K, wcss, 'bo-', linewidth=2)
plt.title('1. Elbow Method (WCSS)')
plt.xlabel('k')
plt.ylabel('Inertia (Lower is better)')
plt.grid(True)

# Plot 2: Silhouette Score
plt.subplot(3, 2, 2)
plt.plot(K, silhouette, 'ro-', linewidth=2)
plt.title('2. Silhouette Score')
plt.xlabel('k')
plt.ylabel('Score (Higher is better)')
plt.grid(True)

# Plot 3: Gap Statistic (New)
plt.subplot(3, 2, 3)
plt.plot(K, gaps, 'co-', linewidth=2)
plt.title('3. Gap Statistic')
plt.xlabel('k')
plt.ylabel('Gap Value (Higher is better)')
plt.grid(True)

# Plot 4: Davies-Bouldin
plt.subplot(3, 2, 4)
plt.plot(K, dbi, 'go-', linewidth=2)
plt.title('4. Davies-Bouldin Index')
plt.xlabel('k')
plt.ylabel('Score (Lower is better)')
plt.grid(True)

# Plot 5: Domain Knowledge / Heuristic (Visual Inspection)
# Visualizing the actual data to let "Human Logic" decide
plt.subplot(3, 2, 5)
plt.scatter(X[:, 0], X[:, 1], s=30, alpha=0.7, c='gray', label='Raw Data')
plt.title('5. Domain Knowledge (Visual Inspection)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True)

# Plot 6: Calinski-Harabasz (Moved here)
plt.subplot(3, 2, 6)
plt.plot(K, ch, 'mo-', linewidth=2)
plt.title('6. Calinski-Harabasz Index')
plt.xlabel('k')
plt.ylabel('Score (Higher is better)')
plt.grid(True)

plt.tight_layout()
plt.show()
````

Here is the complete code. I have added the **Gap Statistic** (which requires a custom calculation since it's not built into `sklearn` by default) and a **Visual Inspection** step to represent **Domain Knowledge/Heuristics**.


### 📝 Summary of Added Methods

1.  **Gap Statistic:**

      * **Code Implemented:** `compute_gap_statistic` function.
      * **How it works:** It creates random noise (reference data) and checks if your data clusters better than the noise.
      * **Interpretation:** A higher Gap value indicates a better K. Ideally, you choose the $K$ where the Gap value maximizes or stops growing significantly.

2.  **Domain Knowledge (Visual Inspection):**

      * **Code Implemented:** `plt.scatter` (Plot 5).
      * **How it works:** "Domain Knowledge" isn't a formula; it's using context. By visualizing the raw data (or a PCA projection if you had many columns), you can visually count the blobs.
      * **Interpretation:** If the quantitative metrics say $K=5$ but you look at Plot 5 and clearly see 4 distinct blobs, your **Domain Knowledge** overrides the math.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# ✅ Step 1: Generate Data (Blobs)
X, y = make_blobs(n_samples=500, centers=4, cluster_std=0.6, random_state=0)

# ✅ Step 2: Define the comparison
# We use max_iter=1 and n_init=1 to pause the algorithm right at the start
# so we can see the INITIAL positions of the centroids.

# Method 1: Forgy (Random Initialization)
# 'random' chooses k observations (rows) at random from data for the initial centroids.
kmeans_forgy = KMeans(n_clusters=4, init='random', n_init=1, max_iter=1, random_state=2) 
kmeans_forgy.fit(X)

# Method 2: K-Means++
# 'k-means++' selects initial cluster centers for k-mean clustering in a smart way to speed up convergence.
kmeans_pp = KMeans(n_clusters=4, init='k-means++', n_init=1, max_iter=1, random_state=2)
kmeans_pp.fit(X)

# ✅ Step 3: Plotting
plt.figure(figsize=(10, 10))

# --- Plot 1: Forgy (Random) ---
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], s=30, c='lightgray', marker='o', label='Data Points')
# Plot Centroids
plt.scatter(kmeans_forgy.cluster_centers_[:, 0], kmeans_forgy.cluster_centers_[:, 1], 
            s=300, c='red', marker='X', edgecolors='black', label='Initial Centroids')
plt.title('Method 1: Forgy (Random Init)\n(Risk: Centroids clumped together)')
plt.legend()
plt.grid(True, alpha=0.3)

# --- Plot 2: K-Means++ ---
plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], s=30, c='lightgray', marker='o', label='Data Points')
# Plot Centroids
plt.scatter(kmeans_pp.cluster_centers_[:, 0], kmeans_pp.cluster_centers_[:, 1], 
            s=300, c='blue', marker='X', edgecolors='black', label='Initial Centroids')
plt.title('Method 2: K-Means++\n(Smart: Centroids spread out)')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

Here is the Python code that demonstrates the difference between these two initialization methods.

To visualize this effectively, I have set the algorithm to run for only **1 iteration**. This allows you to see exactly where the centroids **start** (the initialization) before the algorithm corrects itself.


### 🔍 How to Interpret the Result

1.  **Forgy (Random) Plot:**

      * Look at the **Red X** markers.
      * Because it picks points purely at random, you might see two centroids landing inside the *same* blob, or two centroids very close to each other.
      * **The Problem:** The algorithm will have to work much harder (more iterations) to move those centroids to the correct clusters.

2.  **K-Means++ Plot:**

      * Look at the **Blue X** markers.
      * K-Means++ ensures the centroids are **spread apart** across the data space.
      * **The Benefit:** It is much more likely to land one centroid in each of the 4 distinct blobs right from the start, leading to faster and more accurate results.
