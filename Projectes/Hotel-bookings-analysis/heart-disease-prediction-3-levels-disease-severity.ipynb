### 🧐 Initial Data Observation (Heart Disease Dataset)

### 1. 🎯 Target Variable (Disease Severity)
* The target variable, **`num`**, is **multiclass** with five categories (0, 1, 2, 3, 4).
    * It is significantly **imbalanced**: Class 0 ("No Disease") is the majority (411 samples), while classes 2, 3, and 4 combined (severe disease) total only 244 samples. This necessitates using **stratified splitting** and **robust macro-average metrics** (like Macro F1-score) during modeling.

### 2. 📝 Missing Data (The Primary Challenge)
* **Widespread Missingness:** Seven columns have missing values, with **`ca`** (major vessels) and **`thal`** (thalassemia) being particularly problematic, missing $\sim 66\%$ and $\sim 53\%$ of data, respectively.
* **Imputation Strategy Required:** Simple imputation (like mean/median/mode) for these heavily missing features, especially `ca` and `thal`, carries a significant risk of introducing bias. A strategy like **creating missingness indicators** before imputation is strongly recommended.

### 3. 📊 Feature Engineering and Cleaning Needs
* **Object Type (Categorical):** Eight columns are of `object` type (e.g., `sex`, `cp`, `slope`, `thal`). These must be properly encoded (e.g., One-Hot Encoding) for use in most models.
* **Implausible Zeros:** The `trestbps` (resting blood pressure) and `chol` (cholesterol) columns contain a small number of **$0.0$ values** in the unique value list, which are medically impossible. These should be treated as **missing values ($\text{NaN}$)** and imputed.
* **Outliers:** The `trestbps`, `chol`, and `oldpeak` columns show a few extreme outliers (e.g., $trestbps=200.0$, $chol=603.0$, $oldpeak=6.2$) that exceed reasonable clinical bounds and may require **winsorization** or clipping.
* **Dataset Feature:** The **`dataset`** column identifies the origin of the data (Cleveland, Hungary, etc.). This feature may be highly predictive of the target (as it could reflect demographic or clinical differences between hospitals) and should be included in the model.

```python
# EDA summary & initial cleaning


# Standardize column names (lowercase)
df.columns = [c.strip() for c in df.columns]

# Known columns (from provided dataset)
categorical_cols = ['sex','dataset','cp','fbs','restecg','exang','slope','thal']
numeric_cols = [c for c in df.columns if c not in categorical_cols + ['num','id']]
print('\nNumeric cols:', numeric_cols)
print('Categorical cols (expected):', categorical_cols)

# Replace impossible zeros with NaN for clinical measures
for c in ['trestbps','chol','thalch']:
    if c in df.columns:
        n_zero = (df[c] == 0).sum()
        if n_zero > 0:
            print(f"Marking {n_zero} zeros in {c} as NaN")
            df.loc[df[c] == 0, c] = np.nan

# Create simple missingness indicators for features with >1% missing
for c in df.columns:
    if df[c].isna().mean() > 0.01:
        df[c + '_miss'] = df[c].isna().astype(int)

# Drop 'id' from modeling
if 'id' in df.columns:
    print('Dropping id for modeling (kept in raw DF)')

# Show target distribution
print('\nTarget class distribution:')
print(df['num'].value_counts(dropna=False))
```

# STEP 3: Prepare target: Option B - Collapse 2/3/4 -> 2 (severe)

```python
# ==========================================
# STEP 3: Prepare target: Option B - Collapse 2/3/4 -> 2 (severe)
# ==========================================

COLLAPSE_SEVERE = True
if COLLAPSE_SEVERE:
    y = df['num'].replace({2:2,3:2,4:2})
    print('Collapsing classes 2,3,4 into single class 2 (severe). New class counts:')
    print(y.value_counts())
else:
    y = df['num']

# Features X (drop id and num)
X = df.drop(columns=['num','id']) if 'id' in df.columns else df.drop(columns=['num'])

# --- Indices of the misclassified samples ---
error_indices = [531, 493]

# --- Extract the feature data for the error cases ---
X_errors = X.loc[error_indices]

print("\n--- Feature Data for Error Cases (True Label = 1, Predicted = 0) ---")
print("These samples were misclassified as 'No Disease' (0) instead of 'Moderate Disease' (1).")
print(X_errors.to_string())
```

```python
# --- 1. Custom Winsorization Function ---
# Function to cap extreme outliers at specified percentiles (e.g., 1st and 99th).
def winsorize_array(X, lower_p=0.01, upper_p=0.99):
    arr = np.array(X, dtype=float, copy=True) # Convert input to 2D numpy array.
    if arr.ndim == 1:
        arr = arr.reshape(-1,1) # Ensure it's 2D for scikit-learn.
    out = arr.copy()

    for i in range(out.shape[1]): # Iterate through each feature column.
        col = out[:, i]
        if np.all(np.isnan(col)):
            continue

        # Calculate the percentile bounds, ignoring NaNs.
        low = np.nanquantile(col, lower_p)
        high = np.nanquantile(col, upper_p)

        # Clip (cap) values: Outliers below 'low' become 'low', above 'high' become 'high'.
        out[:, i] = np.clip(col, low, high)
    return out

# Wrap the custom function as a scikit-learn transformer.
wins = FunctionTransformer(winsorize_array, kw_args={'lower_p':0.01,'upper_p':0.99}, validate=False)


# --- 2. OneHotEncoder (OHE) Setup ---
# Using the modern sparse_output=False syntax
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Select feature lists that exist in the current DataFrame (X).
num_features = [c for c in numeric_cols if c in X.columns]
cat_features = [c for c in categorical_cols if c in X.columns]


# --- 3. Preprocessing Pipelines ---
# Pipeline for Numerical Features:
num_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy='median')),  # 1. Fill NaNs with the median.
    ('winsor', wins),                              # 2. Cap outliers (Winsorize).
    ('scale', StandardScaler())                     # 3. Scale to mean=0, std=1.
])

# Pipeline for Categorical Features:
cat_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy='most_frequent')), # 1. Fill NaNs with the mode.
    ('ohe', ohe)                                         # 2. Convert categories to binary columns.
])

# --- 4. Final Column Transformer ---
# Combine the pipelines and apply them to the correct feature sets.
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_features), # Apply numerical steps to numerical features.
    ('cat', cat_pipeline, cat_features)  # Apply categorical steps to categorical features.
],
remainder='drop') # Drop any features not explicitly listed (optional but clean).

print('\nPreprocessor ready. Num features:', len(num_features), 'Cat features:', len(cat_features))
```

# STEP 5: Train/test split

```python
# ==========================================
# STEP 5: Train/test split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE)
print('\nTrain/test shapes:', X_train.shape, X_test.shape)
```

# STEP 6: Model candidates

```python
# ==========================================
# STEP 6: # Model candidates
# We have a balanced selection: RandomForest, LightGBM, XGBoost, SVC, Logistic
# ==========================================

# --- 1. MODEL DEFINITIONS ---
# This dictionary initializes five different classification models.
# Note: RANDOM_STATE and preprocessor are assumed to be defined globally.

# --- ADJUSMENT: Add class_weight to penalize errors on minority classes ---
# --- 1. MODEL DEFINITIONS ---
# This dictionary initializes five different classification models.
# Note: RANDOM_STATE and preprocessor are assumed to be defined globally.
models = {

    'RandomForest': RandomForestClassifier(n_jobs=-1, random_state=RANDOM_STATE),

    'CatBoost': CatBoostClassifier(verbose=0, random_state=RANDOM_STATE),

    'XGB': XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=RANDOM_STATE),

    'SVC': SVC(probability=True, random_state=RANDOM_STATE),

    'Logistic': LogisticRegression(multi_class='multinomial', max_iter=2000, random_state=RANDOM_STATE
                         )

}


# --- 2. HYPERPARAMETER GRIDS ---
# This dictionary defines the search space for GridSearchCV for each model.
# The keys must match the model names in the 'models' dictionary.
# The parameter names follow the format: 'model__[parameter_name]', as they are part of a Pipeline.
param_grids = {
    'RandomForest': {'model__n_estimators':[200,400], 'model__max_depth':[8,16,None]},
    'CatBoost': {'model__iterations':[100,200], 'model__learning_rate':[0.01,0.1]},
    'XGB': {'model__n_estimators':[100,200], 'model__max_depth':[3,5], 'model__learning_rate':[0.01,0.1]},
    'SVC': {'model__C':[0.1,1], 'model__kernel':['rbf']}, # C is the regularization parameter, rbf is the Gaussian kernel.
    'Logistic': {'model__C':[0.1,1,10]} # C is the inverse of regularization strength.
}

# --- 3. EVALUATION SETUP ---
scoring = 'f1_macro' # The metric used to judge the performance of model during grid search. f1_macro is robust for multi-class/imbalanced data.
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE) # Defines the cross-validation strategy: 3 folds, ensuring class balance (Stratified).

results = [] # Initialize a list to store the performance of all models.

# --- 4. MODEL TRAINING AND TUNING LOOP ---
for name, clf in models.items():
    print('\n=== Training', name, '===')

    # Create a pipeline: first apply preprocessing, then fit the classifier.
    pipe = Pipeline([('preproc', preprocessor), ('model', clf)])

    # GridSearchCV searches over the defined 'param_grids' for the optimal parameters.
    # n_jobs=-1 runs the search in parallel using all CPU cores.
    grid = GridSearchCV(pipe,
                        param_grids.get(name, {}), # Get the grid for the current model name.
                        scoring=scoring,
                        cv=cv,
                        n_jobs=-1)

    # Fit the GridSearchCV object to the training data (X_train, y_train).
    # This triggers the cross-validated search across all parameter combinations.
    grid.fit(X_train, y_train)

    best = grid.best_estimator_ # Retrieve the best performing model (with optimal hyperparameters) from the search.
    y_pred = best.predict(X_test) # Make predictions on the unseen test data.

    # Calculate final test metrics
    acc = accuracy_score(y_test, y_pred)
    f1m = f1_score(y_test, y_pred, average='macro')

    # Output results
    print('Best CV (search.best_score_):', grid.best_score_)
    print('Test f1_macro:', f1m, 'accuracy:', acc)
    print(classification_report(y_test, y_pred, zero_division=0)) # Detailed per-class metrics

    # Store results for final comparison
    results.append({'model':name, 'best':best, 'cv':grid.best_score_, 'f1_macro':f1m, 'accuracy':acc})

# --- 5. FINAL COMPARISON AND SELECTION ---
# Convert the list of results into a DataFrame.
res_df = pd.DataFrame(results).sort_values('f1_macro', ascending=False).reset_index(drop=True)
print('\nModel comparison:')
# Display the summary table
print(res_df[['model','cv','f1_macro','accuracy']])

# Choose the model that performed best on the test set based on the 'f1_macro' metric.
best_row = res_df.loc[0]
best_name = best_row['model']
best_model = best_row['best']

print('\nBest model chosen:', best_name)
```

```python
# --- OPTUNA TUNING (If the best model is tree-based) ---

# Calculate class weights for handling imbalance
# This creates a dictionary mapping class labels (0, 1, 2) to their respective weights.
class_weights_array = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weights_dict = dict(zip(np.unique(y_train), class_weights_array))

# Check if the best model chosen by GridSearchCV is one of the supported tree-based estimators
if best_name in ('LGBM','XGB','RandomForest'):
    print('Running Optuna tuning for', best_name)

    # Define the objective function that Optuna will minimize/maximize.
    def objective(trial):
        # Dynamically define the hyperparameter search space using trial.suggest_X() methods.
        if best_name == 'LGBM':
            params = {
                # Suggest integer values for number of estimators (trees)
                'n_estimators': trial.suggest_int('n_estimators', 100, 500),
                # Suggest float values for learning rate, searching on a logarithmic scale
                'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
                'max_depth': trial.suggest_int('max_depth', 3, 12),
                'num_leaves': trial.suggest_int('num_leaves', 16, 128),
                'min_data_in_leaf': trial.suggest_int('min_data_in_leaf', 10, 50),
                'feature_fraction': trial.suggest_float('feature_fraction', 0.5, 1.0)
            }
            # Instantiate the LightGBM model with the suggested parameters
            model = LGBMClassifier(random_state=RANDOM_STATE, n_jobs=-1, **params)
        elif best_name == 'XGB':
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 100, 400),
                'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
                'max_depth': trial.suggest_int('max_depth', 3, 8)
            }
            # Instantiate the XGBoost model
            model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=RANDOM_STATE, **params)
        else: # RandomForest
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 100, 500),
                'max_depth': trial.suggest_int('max_depth', 4, 20)
            }
            # Instantiate the RandomForest model, preserving class weights for imbalance
            model = RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1, class_weight=class_weights_dict, **params)

        # Create a temporary pipeline for this trial
        pipe = Pipeline([('preproc', preprocessor), ('model', model)])
        # Evaluate the pipeline using cross-validation (CV) and f1_macro score
        scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring='f1_macro')
        # Return the mean CV score, which Optuna attempts to maximize
        return float(np.mean(scores))

    # Initialize Optuna study, setting the goal to maximize the score
    study = optuna.create_study(direction='maximize')
    # Run the optimization for a specified number of trials
    study.optimize(objective, n_trials=40, show_progress_bar=True)

    print('Best optuna:', study.best_params, 'best_value:', study.best_value)

    # --- Build and Fit Final Tuned Model ---
    # Re-instantiate the final model using the best hyperparameters found by Optuna
    if best_name == 'LGBM':
        tuned = LGBMClassifier(random_state=RANDOM_STATE, n_jobs=-1, **study.best_params)
    elif best_name == 'XGB':
        tuned = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=RANDOM_STATE, **study.best_params)
    else:
        # Pass class_weights to the tuned Random Forest model
        tuned = RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1, class_weight=class_weights_dict, **study.best_params)

    # Create the final best pipeline with the tuned model
    best_model = Pipeline([('preproc', preprocessor), ('model', tuned)])
    # Train the final pipeline on the entire training dataset
    best_model.fit(X_train, y_train)
    print('Optuna tuned model fitted')
else:
    # Skip tuning if the best model is not one of the supported types (e.g., SVC, Logistic)
    print('Optuna not run (not supported for the chosen model)')
```

# STEP 7: Calibration for probability outputs

```python
# ==========================================
# STEP 7: Calibration for probability outputs
# ==========================================

print('Calibrating probabilities using CalibratedClassifierCV (sigmoid)')
# Need to fit calibrated model on transformed features
base_clf = best_model.named_steps['model'] if isinstance(best_model, Pipeline) else best_model

# transform training data
X_train_trans = preprocessor.fit_transform(X_train)
calib = CalibratedClassifierCV(base_clf, cv='prefit', method='sigmoid')


# if base_clf not fitted, fit
if not hasattr(base_clf, 'classes_'):
    # fit base pipeline if not fitted
    best_model.fit(X_train, y_train)
    base_clf = best_model.named_steps['model']
    X_train_trans = preprocessor.transform(X_train)
calib.fit(X_train_trans, y_train)

# final pipeline: preprocessor -> calibrated
final_pipeline = Pipeline([('preproc', preprocessor), ('calib', calib)])

# Evaluate final pipeline
y_pred = final_pipeline.predict(X_test)
y_prob = final_pipeline.predict_proba(X_test)
print('Final (calibrated) evaluation:')
print('Accuracy:', accuracy_score(y_test, y_pred))
print('F1 macro:', f1_score(y_test, y_pred, average='macro'))
print(classification_report(y_test, y_pred, zero_division=0))
print('Confusion matrix:\n', confusion_matrix(y_test, y_pred))
```

# STEP 8: Feature names extraction and permutation importance

```python
# ==========================================
# STEP 8: Feature names extraction and permutation importance
# ==========================================

print('Building feature names from preprocessor for importance plots')
# Get the original column names from the test set. These are the features permutation_importance shuffles.
input_feat_names = X_test.columns.tolist()

print('Total input features for permutation importance:', len(input_feat_names))

print('Running permutation importance (may be slow)')

# Calculate Permutation Importance: Measures how model performance decreases when a feature is randomly shuffled.
# The calculation is performed on the fully fitted 'final_pipeline' using the hold-out 'X_test' and 'y_test'.
perm = permutation_importance(
    final_pipeline,
    X_test,
    y_test,
    n_repeats=10,
    random_state=RANDOM_STATE,
    n_jobs=-1
)
# Create a DataFrame to display the results neatly
imp_df = pd.DataFrame({
    'feature': input_feat_names,
    'importance_mean': perm.importances_mean,
    'importance_std': perm.importances_std
})
# Sort the DataFrame to put the most important features (highest mean decrease in score) at the top
imp_df = imp_df.sort_values('importance_mean', ascending=False).reset_index(drop=True)
display(imp_df.head(30)) # Show the importance scores

# Keep the transformed feature names for SHAP later (these are the columns used by the model internally)
feat_names = []
# Start with the names of the original numerical features
feat_names.extend(num_features)

# Extract the names of the columns created by One-Hot Encoding (OHE) for categorical features
ohe_step = preprocessor.named_transformers_['cat'].named_steps['ohe']
cat_ohe_names = list(ohe_step.get_feature_names_out(cat_features))
# Add the OHE feature names to the list
feat_names.extend(cat_ohe_names)

print('Total transformed features (for SHAP/internal representation):', len(feat_names))

# plot
plt.figure(figsize=(8,6))
# Create a horizontal bar plot of the mean feature importance
sns.barplot(x='importance_mean', y='feature', data=imp_df)
plt.title('Permutation importance (top features)')
plt.show() #
```
