## **3. Sweetviz (The Comparison Specialist)**
**Best for comparing datasets** (Train vs Test, Before vs After)

### **Installation**
```bash
pip install sweetviz
```

### **Basic Usage**
```python
import sweetviz as sv
import pandas as pd

# Load datasets
df_train = pd.read_csv("train.csv")
df_test = pd.read_csv("test.csv")

# 1. Analyze single dataset
report = sv.analyze(df_train, target_feat='target_column')
report.show_html("single_report.html")

# 2. Compare two datasets
compare_report = sv.compare(
    [df_train, "Train"],
    [df_test, "Test"],
    target_feat="target_column"
)
compare_report.show_html("comparison_report.html")

# 3. Compare with source dataset
report = sv.compare_intra(
    df_train,
    df_train["split_column"] == "train",  # Condition
    ["Train", "Test"],
    target_feat='target_column'
)
```

### **Advanced Features**
```python
# Custom configuration
feature_config = sv.FeatureConfig(
    skip=["id", "unnecessary_column"],  # Skip columns
    force_cat=["zip_code"],  # Force categorical
    force_text=["description"]  # Force text analysis
)

# Create detailed report
report = sv.analyze(
    df_train,
    target_feat='target_column',
    feat_cfg=feature_config,
    pairwise_analysis='on'  # Enable correlation analysis
)

# Show in notebook
report.show_notebook()
```

## **🔄 Comparing Multiple Tools Side-by-Side**

```python
import pandas as pd
from datetime import datetime

class EDAComparator:
    """Compare different EDA tools on the same dataset"""
    
    def __init__(self, df):
        self.df = df
        self.results = {}
        
    def run_ydata_profiling(self):
        """Run ydata-profiling"""
        from ydata_profiling import ProfileReport
        start = datetime.now()
        profile = ProfileReport(self.df, minimal=True, title="Quick Analysis")
        html_path = f"reports/ydata_report_{datetime.now().strftime('%Y%m%d_%H%M%S')}.html"
        profile.to_file(html_path)
        elapsed = (datetime.now() - start).total_seconds()
        return {"tool": "ydata-profiling", "time": elapsed, "output": html_path}
    
    def run_skimpy(self):
        """Run skimpy"""
        from skimpy import clean_columns, skim
        start = datetime.now()
        # Clean column names
        df_clean = clean_columns(self.df)
        result = skim(df_clean)
        elapsed = (datetime.now() - start).total_seconds()
        return {"tool": "skimpy", "time": elapsed, "output": "console"}
    
    def run_all(self):
        """Run all EDA tools"""
        tools = [
            self.run_skimpy,
            self.run_ydata_profiling,
            # Add other tools as needed
        ]
        
        for tool_func in tools:
            try:
                result = tool_func()
                self.results[result['tool']] = result
                print(f"✅ {result['tool']}: {result['time']:.2f} seconds")
            except Exception as e:
                print(f"❌ Error with {tool_func.__name__}: {e}")
        
        return self.results

# Usage
# df = pd.read_csv("your_data.csv")
# comparator = EDAComparator(df.sample(10000))  # Use sample for speed
# results = comparator.run_all()
```

### **2. Memory-Efficient EDA**
```python
def memory_efficient_eda(df):
    """Perform EDA without loading everything in memory"""
    
    # 1. Basic stats without full load
    print("Dataset shape:", df.shape)
    print("\nColumn dtypes:")
    print(df.dtypes)
    
    # 2. Memory usage
    print("\nMemory usage per column (MB):")
    for col in df.columns:
        mem = df[col].memory_usage(deep=True) / 1024**2
        print(f"  {col}: {mem:.2f} MB")
    
    # 3. Sample-based analysis
    df_sample = df.sample(min(10000, len(df)))
    
    # 4. Use skimpy for quick analysis
    from skimpy import skim
    skim(df_sample)
    
    # 5. Save to parquet for faster subsequent loads
    df_sample.to_parquet("sample_data.parquet")
```

### **3. Chunk-based Processing**
```python
def chunk_based_analysis(filepath, chunksize=10000):
    """Analyze data in chunks to avoid memory issues"""
    
    chunk_stats = []
    
    for i, chunk in enumerate(pd.read_csv(filepath, chunksize=chunksize)):
        print(f"Processing chunk {i+1}...")
        
        # Collect basic stats from each chunk
        stats = {
            'chunk': i+1,
            'rows': len(chunk),
            'columns': len(chunk.columns),
            'missing_total': chunk.isnull().sum().sum(),
            'memory_mb': chunk.memory_usage(deep=True).sum() / 1024**2
        }
        chunk_stats.append(stats)
        
        # Process first chunk in detail
        if i == 0:
            from skimpy import skim
            print("\nFirst chunk analysis:")
            skim(chunk)
    
    # Aggregate stats
    total_stats = pd.DataFrame(chunk_stats)
    print("\nAggregated Statistics:")
    print(total_stats.describe())
```

## **📝 Practical Implementation Guide**

### **Complete EDA Pipeline**
```python
import pandas as pd
import numpy as np
from pathlib import Path

class AutomatedEDA:
    """Complete automated EDA pipeline"""
    
    def __init__(self, data_path):
        self.data_path = data_path
        self.df = None
        self.reports_dir = Path("eda_reports")
        self.reports_dir.mkdir(exist_ok=True)
        
    def load_data(self):
        """Load data with automatic format detection"""
        path = Path(self.data_path)
        
        if path.suffix == '.csv':
            self.df = pd.read_csv(path)
        elif path.suffix == '.parquet':
            self.df = pd.read_parquet(path)
        elif path.suffix in ['.xlsx', '.xls']:
            self.df = pd.read_excel(path)
        else:
            raise ValueError(f"Unsupported file format: {path.suffix}")
        
        print(f"✅ Loaded {len(self.df)} rows, {len(self.df.columns)} columns")
        return self.df
    
    def quick_overview(self):
        """Fast overview using skimpy"""
        try:
            from skimpy import clean_columns, skim
            df_clean = clean_columns(self.df)
            print("\n" + "="*50)
            print("QUICK OVERVIEW (skimpy)")
            print("="*50)
            skim(df_clean)
        except ImportError:
            print("skimpy not installed, using pandas describe")
            print(self.df.describe(include='all'))
    
    def detailed_analysis(self):
        """Detailed analysis with ydata-profiling"""
        try:
            from ydata_profiling import ProfileReport
            timestamp = pd.Timestamp.now().strftime("%Y%m%d_%H%M%S")
            report_path = self.reports_dir / f"detailed_report_{timestamp}.html"
            
            print("\n" + "="*50)
            print("GENERATING DETAILED REPORT...")
            print("="*50)
            
            profile = ProfileReport(
                self.df,
                title=f"EDA Report - {timestamp}",
                explorative=True,
                minimal=False
            )
            profile.to_file(report_path)
            print(f"✅ Report saved: {report_path}")
            
        except ImportError:
            print("ydata-profiling not installed")
    
    def interactive_exploration(self):
        """Interactive exploration options"""
        print("\n" + "="*50)
        print("INTERACTIVE EXPLORATION OPTIONS")
        print("="*50)
        print("1. For GUI-based exploration: dtale.show(df)")
        print("2. For visualization builder: pygwalker.walk(df)")
        print("3. For train/test comparison: Use Sweetviz")
    
    def generate_summary(self):
        """Generate text summary"""
        summary = f"""
        DATA SUMMARY
        {'='*40}
        Dataset: {self.data_path}
        Shape: {self.df.shape[0]} rows × {self.df.shape[1]} columns
        Memory Usage: {self.df.memory_usage(deep=True).sum() / 1024**2:.2f} MB
        
        COLUMN INFORMATION:
        {'-'*40}
        """
        
        for col in self.df.columns:
            dtype = self.df[col].dtype
            missing = self.df[col].isnull().sum()
            missing_pct = (missing / len(self.df)) * 100
            unique = self.df[col].nunique()
            
            summary += f"\n{col}:"
            summary += f"\n  Type: {dtype}"
            summary += f"\n  Missing: {missing} ({missing_pct:.1f}%)"
            summary += f"\n  Unique values: {unique}"
            
            if np.issubdtype(dtype, np.number):
                summary += f"\n  Range: {self.df[col].min():.2f} to {self.df[col].max():.2f}"
                summary += f"\n  Mean: {self.df[col].mean():.2f}"
        
        # Save summary
        summary_path = self.reports_dir / "data_summary.txt"
        with open(summary_path, 'w') as f:
            f.write(summary)
        
        print(summary)
        print(f"\n✅ Summary saved: {summary_path}")
    
    def run_full_pipeline(self):
        """Execute complete EDA pipeline"""
        print("🚀 STARTING AUTOMATED EDA PIPELINE")
        print("="*60)
        
        # 1. Load data
        self.load_data()
        
        # 2. Quick overview
        self.quick_overview()
        
        # 3. Detailed analysis (if dataset < 100K rows)
        if len(self.df) < 100000:
            self.detailed_analysis()
        else:
            print("\n⚠️  Dataset too large for detailed report, using sample...")
            df_sample = self.df.sample(min(50000, len(self.df)))
            from ydata_profiling import ProfileReport
            profile = ProfileReport(df_sample, minimal=True)
            profile.to_file(self.reports_dir / "sampled_report.html")
        
        # 4. Generate summary
        self.generate_summary()
        
        # 5. Show interactive options
        self.interactive_exploration()
        
        print("\n" + "="*60)
        print("✅ EDA PIPELINE COMPLETED")
        print(f"📁 Reports saved in: {self.reports_dir.absolute()}")

# Usage
# eda = AutomatedEDA("your_dataset.csv")
# eda.run_full_pipeline()
```
